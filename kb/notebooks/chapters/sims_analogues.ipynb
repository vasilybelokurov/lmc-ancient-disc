# Simulated analogues observed like the LMC

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/vasilybelokurov/lmc-ancient-disc/blob/main/kb/notebooks/chapters/sims_analogues.ipynb)

Every number on this page is written by a script: `scripts/paper_text_sims.py` (`results/macros_sims.tex`) and
`scripts/paper_text_letter.py` (`results/macros_letter.tex`), from the result tables; the code cells are the figure script
`scripts/paper_sims_kinematics.py`, step by step.

The kinematics chapter found that the oldest, most metal-poor red giants of the LMC rotate at
$v_{\rm rot}(4\,{\rm kpc})\approx$ 57 km/s with a vertical velocity dispersion $\sigma_z\approx$ 32 km/s:
rotation dominates even in the most metal-poor population. Here $v_{\rm rot}(4\,{\rm kpc})$ is the mean azimuthal velocity of a
population of stars, from its fitted rotation curve, at cylindrical radius $R$ = 4 kpc in the plane of the disc; $\sigma_z$ is the
dispersion of the velocities perpendicular to that plane, at the same radius; and [Fe/H], the iron abundance, is the logarithm of a
star's iron-to-hydrogen ratio relative to the Sun's. The ratio $v_{\rm rot}/\sigma_z$, the rotational support, measures how far a
population is supported against gravity by rotation rather than by random motions: a cold disc has $v_{\rm rot}/\sigma_z$ of several,
a hot spheroid well below 1. The giants are divided, as in the Letter, into five bins of [Fe/H] with edges at -1.5, -1.1, -0.8
and -0.5.

Here we ask whether galaxies of the LMC's mass formed in cosmological simulations have the same property. A like-for-like answer
needs the simulations measured as the LMC was: the same tracer, the same sky positions and errors, the same metallicity bins and the
same kinematic model. We therefore 'observe' every simulated galaxy and compare rotation, dispersion and rotational support against
metallicity.

## The simulations

We use three public suites of hydrodynamical simulations, which follow dark matter, gas and stars together, with LMC-mass galaxies
and star particles (each a population of stars of one age and metallicity) at redshift $z=0$.

- [TNG50-1](https://www.tng-project.org/) ([Nelson et al. 2019a](https://doi.org/10.1093/mnras/stz2306);
  [Pillepich et al. 2019](https://doi.org/10.1093/mnras/stz2338); [Nelson et al. 2019b](https://doi.org/10.1186/s40668-019-0028-x)),
  the highest-resolution run of the IllustrisTNG cosmological volumes: all 681 central galaxies (the main galaxy of
  their group, not a satellite) with bound stellar mass 1.5–5 × 10⁹ M☉ and subhalo mass (all mass bound to the galaxy)
  $5\times10^{10}$–$5\times10^{11}$ M☉; baryonic mass resolution 8.5 × 10⁴ M☉ (Pillepich et al. 2019).
- Auriga, level 3 ([Grand et al. 2024](https://doi.org/10.1093/mnras/stae1598)):
  zoom-in simulations, which resimulate one halo of a cosmological volume at high resolution; we use the five dwarf-galaxy zooms
  Au2, 8, 9, 10 and 11 (haloes of ≈10¹¹ M☉).
- FIRE-2 ([Wetzel et al. 2023](https://doi.org/10.3847/1538-4365/acb99a)): the public
  zooms m11d, m11e and m11h at 7100 M☉ resolution.

The TNG50 analogues are selected by mass alone, as $z=0$ centrals; none is required to have an SMC-like companion or an LMC-like
orbit. Their dark-matter particles have 4.5 × 10⁵ M☉ (Pillepich et al. 2019); those of the Auriga and FIRE-2 zooms,
3.4–4.6 × 10⁴ and $3.9\times10^{4}$ M☉ (read from the snapshots), 12 times less. Stellar metallicities are
$[{\rm Fe/H}]=\log_{10}[(X_{\rm Fe}/X_{\rm H})/(X_{\rm Fe}/X_{\rm H})_\odot]$, with $X$ the mass fractions of iron and hydrogen of the
particle and the solar abundance of [Asplund et al. (2021)](https://doi.org/10.1051/0004-6361/202140445). The formed (birth) mass of
a star particle is the mass of stars it formed, before stellar evolution returned part of it to the gas. It is taken from the
snapshot where available (Auriga) and otherwise from the present mass divided by the present-to-formed mass fraction of a stellar
population with the [Kroupa (2001)](https://doi.org/10.1046/j.1365-8711.2001.04022.x) initial mass function, of the particle's age and
metallicity (the conversion of the stellar-mass chapter); for Auriga this estimate is 6–7 per cent below the snapshot's birth
masses.

## Observing the simulations

For every simulated galaxy we build a mock version of the sample of red giants and fit it exactly as the data. The data are the
giants with SDSS-V BOSS or APOGEE spectra (two spectroscopic surveys; BOSS is the optical, APOGEE the near-infrared one) and Gaia
proper motions, described in the data chapters.

1. *Frame and placement.* Star particles within 50 kpc of the galaxy centre are rotated so that $z$ lies along the angular
   momentum of the stars within 10 kpc, with velocities relative to their mean. The galaxy is then placed at the LMC: the centre,
   inclination (the tilt of the disc plane to the sky), line of nodes (where the disc plane crosses the sky), systemic proper motion
   and line-of-sight velocity (the motion of the LMC's centre) of the joint kinematic fit (kinematics chapter),
   distance 50.1 kpc, with the simulation's sense of rotation mapped onto the LMC's. Each particle keeps its own 3D position, so
   the thickness of the simulated galaxy enters the projection as it does for the LMC.
2. *Tracer.* Each particle contributes $E=m_{\rm formed}\,\eta_{\rm BOSS}(\tau,[{\rm Fe/H}])$ red giants in the BOSS selection box,
   where $m_{\rm formed}$ is its formed mass, $\tau$ its age and $\eta_{\rm BOSS}$ the number of giants in the box per unit formed mass
   of the stellar-mass chapter.
3. *Metallicity.* The simulated metallicity distributions differ from the LMC's in zero point and shape, and the difference is partly
   a matter of nucleosynthetic yields. We therefore compare populations of the same *rank*: each particle's [Fe/H] is mapped onto the
   LMC scale so that the cumulative formed mass below it, as a fraction of the galaxy's formed mass within 12° of its centre
   on the sky, equals the LMC's (stellar-mass chapter). A simulated bin then holds the same fraction of its galaxy's stellar mass as
   the LMC bin of the same [Fe/H]: the most metal-poor bin, [Fe/H] < -1.5, holds the 2.1 per cent of the formed mass lowest in
   metallicity. These are old stars (median age 12.1 Gyr), but not simply the oldest: only 45 per cent of the oldest
   stars holding the same fraction of the mass by age fall in this bin. The mapping is applied to the LMC's measured distribution,
   and BOSS errors are then added (step 4), so the simulated bins are blurred somewhat more than the LMC's; this mixes metal-richer,
   more rotating stars into the metal-poor bin.
4. *Where and how well.* The real giants fix where we look. Each real giant at 2° < $\rho$ < 9°, where $\rho$ is the
   angular distance from the LMC centre, lies in a cell of the fit frame (1 kpc in in-plane radius × 45° in azimuth); for
   each, one mock giant is drawn from the particles of the same cell with probability $\propto E$, inside the same velocity window
   (100 < $v_{\rm los}$ < 450 km/s, with $v_{\rm los}$ the line-of-sight velocity). It inherits the real giant's proper-motion
   covariance, scaled by the Gaia error factor $k$ fitted for the real bin (the factor by which the kinematic model inflates the
   catalogued proper-motion errors), and its line-of-sight error, and its measured [Fe/H] is the mapped value plus an error: none for
   an APOGEE giant, a random BOSS − APOGEE difference of the stars observed by both surveys for a BOSS giant. The mock sample
   therefore has the size, sky distribution, error budget and metallicity blurring of the real one. Its [Fe/H] bins differ in size
   from the real ones (for [Fe/H] < -1.5, a median 1222 mock giants against 2168), because the analogues' stars of a
   given [Fe/H] rank differ in age and so in giants per unit mass.
5. *Fit.* Each mock [Fe/H] bin is fitted on its own with the ellipsoid model of the kinematics chapter (a thin rotating disc whose
   velocity ellipsoid has dispersions $\sigma_R$, $\sigma_\phi$, $\sigma_z$ along the radial, azimuthal and vertical directions),
   by maximum likelihood with all parameters free (stage 1 of the fit); bins with more than 8000 mock giants are fitted on a
   random 8000.

The procedure recovers the particles' own kinematics (Figure 9.1), measured as the mean $v_\phi$ and $\sigma_z$ of the
particles that became the mock giants, within 0.5 kpc of $R$ = 4 kpc. Over all 3445 analogue bins the fitted
$v_{\rm rot}(4\,{\rm kpc})$ exceeds the particles' by +0.7 km/s on the median (16th–84th percentiles -2.2 to +5.0 km/s) and
$\sigma_z$ by 1.5 per cent. The metal-poor bin is the noisiest (-5 to +12 km/s in $v_{\rm rot}$, -8 to +12 per
cent in $\sigma_z$), but its $v_{\rm rot}/\sigma_z$ is recovered to +0.07 (-0.11 to +0.24). The model does not allow
counter-rotation, so a bin whose particles counter-rotate is fitted with $v_{\rm rot}=0$.

### The mock observation, step by step

`scripts/sims_observe.py` carries out steps 1–5 for every analogue and records the particle values against which the fits are
checked. Its function `setup` reads what all analogues share: the frame of the joint kinematic fit, the real giants with their errors,
cells and error factors $k$, the differences between the BOSS and APOGEE [Fe/H] of the stars observed by both surveys, the LMC's
cumulative formed mass against [Fe/H], and the number of BOSS giants per unit formed mass. `particles` does steps 1–3 for one
analogue: it turns the galaxy into the frame of its spin, places it at the LMC, and gives every particle its LMC-scale [Fe/H] and its
expected number of giants $E$. `observe` does step 4: for each real giant it draws one particle of the same cell with probability
proportional to $E$ and gives it that giant's errors. The main loop splits the mock giants into the five [Fe/H] bins, takes the mean
$v_\phi$ and the dispersions of their particles near $R$ = 4 kpc, and fits each bin (step 5).

Each analogue is written to a file of its own, so that the work can be shared among parallel processes and restarted;
`python scripts/sims_observe.py merge` then joins the files into `results/sims_observed.csv`, the table from which all the figures of
this chapter are drawn.

#### Helpers: sampling cells and the LMC's metallicity scale
`sky_cell` gives the cell (in-plane radius and azimuth in the fitted disc frame) of any sky position, so that real giants and
simulated particles are put in the same cells. `lmc_knots` is the LMC's cumulative formed-mass fraction against [Fe/H], to which
the simulated metallicities are matched (step 3).

```python
def sky_cell(p, ra, dec):
    """Cell index (R // CELL_R, phi // CELL_PHI) of sky positions in the frame of p, with R and phi as lmcold.kinfit computes them
    (line of sight to the disc plane).

    p: kinfit parameter vector (kinfit.PARAMS); ra, dec: (N,) [deg]. Returns (N,) int, 1000 x radial cell + azimuthal cell, with
    the in-plane radius R [kpc] and the azimuth phi [deg, 0-360, from the line of nodes] where the line of sight meets the disc plane."""
    n = len(ra)
    # Line of sight of each position meets the disc plane of frame p: R [kpc], unit radial vector eR and disc normal nn.
    d = kinfit.data(ra, dec, np.zeros(n), np.zeros(n), np.ones(n), np.ones(n), np.zeros(n))   # positions only; motions are dummies
    _, _, _, _, R, eR, _, nn = (np.asarray(v) for v in kinfit._model(p, d['rhat'], d['ea'], d['ed']))
    # Unit vector along the line of nodes of the frame (centre, inclination, position angle theta of p).
    _, _, _, line, _ = (np.asarray(v) for v in kinfit.geo.symmetry_axis(*(float(c) for c in kinfit.centre(p[0], p[1])),
                                                                        np.deg2rad(p[P('inc')]), np.deg2rad(p[P('theta')])))
    phi = np.rad2deg(np.arctan2(eR @ np.cross(nn[0], line), eR @ line)) % 360   # deg, azimuth from the line of nodes
    return (R // CELL_R).astype(int) * 1000 + (phi // CELL_PHI).astype(int)


def lmc_knots():
    """The LMC's cumulative formed-mass fraction against [Fe/H] (results/giant_mass.csv): ([Fe/H] knots from LMC_FEH_FLOOR, fraction
    of the formed mass within RHO_MASS below each).

    Returns two arrays of equal length: [Fe/H] [dex] (LMC_FEH_FLOOR, then the upper edges of the mass bins but the last) and the
    fraction (0 at the floor, rising to below 1)."""
    # One row per [Fe/H] bin: feh_hi (upper edge), cumulative (formed mass below that edge).
    m = pd.read_csv(ROOT / 'results/giant_mass.csv')
    # Knots: the floor plus every upper edge but the last; fractions normalised by the total formed mass.
    return np.r_[LMC_FEH_FLOOR, m.feh_hi.to_numpy()[:-1]], np.r_[0, m.cumulative.to_numpy()[:-1] / m.cumulative.iloc[-1]]
```

#### Inputs shared by all analogues
`setup` reads everything that does not depend on the analogue: the frame of the joint fit, the real giants (their positions, errors,
survey and cell, and the error factor $k$ of their [Fe/H] bin), the BOSS − APOGEE [Fe/H] differences, the LMC metallicity knots, and
the grids of giants per formed mass and of surviving mass.

```python
def setup():
    """Inputs shared by all analogues: frame vector, real giants (one table, with cell, bin, k), BOSS - APOGEE differences, LMC
    cumulative-mass knots, n_box and survival grids.

    Returns (frame: kinfit parameter vector of the joint stage-2 fit; real: DataFrame, one row per real giant at 2-9 deg with ra, dec
    [deg], pmra_error, pmdec_error [mas/yr], pmra_pmdec_corr, evlos [km/s], survey, k, cell; dfeh: (M,) BOSS - APOGEE [Fe/H] [dex] of
    the stars in both surveys; knots: lmc_knots(); nbox: BOSS rows of results/giant_conversion.csv; survival: the surviving-mass
    grid of lmcold.surviving_mass)."""
    frame = next(iter(joint_params().values()))   # the shared parameters (centre, orientation, systemic motion) are the same in every population's vector
    t = pd.read_csv(ROOT / 'results/kin_mcmc_data_ellipsoid.csv')
    ks = t[(t.stage == 1) & (t.param == 'k')].set_index('population')['median']   # proper-motion error factor of each real bin
    real = []
    for j, (lab, c) in enumerate(samples.giants()):
        c = ring(c)   # 2 < rho < 9 deg, as the kinematic fits
        real.append(pd.DataFrame({k: c[k] for k in ('ra', 'dec', 'pmra_error', 'pmdec_error', 'pmra_pmdec_corr', 'evlos', 'survey')})
                    .assign(k=ks[f'giants {lab}']))
    real = pd.concat(real, ignore_index=True)   # all real kinematic giants, every [Fe/H] bin
    real['cell'] = sky_cell(frame, real.ra.to_numpy(), real.dec.to_numpy())
    o = samples.overlap()   # stars in both APOGEE and BOSS
    dfeh = (o.feh_b - o.feh_a).to_numpy()   # (M,) BOSS - APOGEE [Fe/H] [dex]
    knots = lmc_knots()
    g = pd.read_csv(ROOT / 'results/giant_conversion.csv')
    nbox = g[g.survey == 'BOSS']   # giants in the BOSS selection box per unit formed mass, on an ([M/H], log age) grid
    # Surviving fraction of the formed mass on the PARSEC grid, used by lmcold.sims.load for the formed masses.
    survival = surviving_mass.survival_grid(pd.read_parquet(path('parsec_gaiaEDR3')))
    return frame, real, dfeh, knots, nbox, survival
```

#### One analogue: placing it and weighting its particles (steps 1-3)
The particles are rotated into the frame of their spin, placed at the LMC with the fitted frame, given their LMC-scale [Fe/H]
(only inside the 12-deg cone of the LMC's mass measurement) and weighted by their expected number of BOSS giants $E$.

```python
def particles(suite, name, frame, knots, nbox, survival):
    """Particles of one analogue (steps 1-3): disc-frame X, V (N, 3), the sky observables of lmcold.sims.place, rho [deg] from the
    vdMK14 centre, LMC-scale [Fe/H] (NaN outside RHO_MASS), the expected BOSS giants per particle E, and the loaded particles
    (lmcold.sims.load: native feh, age, m_formed).

    X [kpc], V [km/s]; the sky observables are a dict of (N,) arrays ra, dec [deg], dist [kpc], pmra, pmdec [mas/yr], vlos [km/s];
    E (N,) is in giants (m_formed [Msun] times giants per Msun)."""
    st = sims.load(suite, name, survival)   # dict of particle arrays: x, v, m, m_formed, feh, age
    X, V, _ = sims.disc_frame(st['x'], st['v'], st['m'])   # (N, 3) [kpc], [km/s] in the frame of the spin
    s = sims.place(X, V, frame)   # sky observables with the LMC's fitted frame and distance
    rho = angular_distance(s['ra'], s['dec'], LMC['ra0'], LMC['dec0'])   # deg from the reference centre (lmcold.constants)
    cone = rho < RHO_MASS   # (N,) bool, inside the 12-deg radius of the LMC mass measurement
    feh = np.full(len(rho), np.nan)
    feh[cone] = sims.lmc_scale(st['feh'][cone], st['m_formed'][cone], *knots)   # matched to the LMC's mass inside the same cone
    # E = formed mass [Msun] x BOSS giants per Msun interpolated at each particle's native [M/H] and age.
    return X, V, s, rho, feh, st['m_formed'] * sims.on_grid(nbox, 'n_box', st['feh'], st['age']), st
```

#### Drawing the mock giants (step 4)
For each real giant, one particle of the same cell, drawn with probability proportional to $E$; the mock giant then gets the real
giant's errors (proper motions scaled by $k$) and, for a BOSS giant, a BOSS − APOGEE [Fe/H] difference.

```python
def observe(suite, name, frame, real, dfeh, knots, nbox, survival, rng):
    """Mock giant catalogue of one analogue (module docstring, steps 1-4): dict of arrays with the kinematic columns (COLS), feh
    (measured, LMC scale), particle index, and the disc-frame R, v_R, v_phi, v_z of the particle; and the number of real giants
    without a mock.

    The inputs after suite, name are those of setup(); rng is a numpy Generator. Units: ra, dec [deg], pmra, pmdec and their errors
    [mas/yr], vlos, evlos, vR, vphi, vz [km/s], R [kpc], feh [dex]; one entry per real giant that has a mock."""
    X, V, s, rho, feh, E, _ = particles(suite, name, frame, knots, nbox, survival)
    # Particles that could be giants of the sample: LMC-scale [Fe/H] defined, 2 < rho < 9 deg, inside the velocity window.
    ok = np.isfinite(feh) & (rho > 2) & (rho < 9) & (s['vlos'] > VLOS_WINDOW[0]) & (s['vlos'] < VLOS_WINDOW[1])
    E = np.where(ok, E, 0.0)
    idx = np.flatnonzero(E > 0)   # indices of the candidate particles
    cell = sky_cell(frame, s['ra'][idx], s['dec'][idx])
    pick = np.full(len(real), -1)   # particle drawn for each real giant (-1: none in its cell)
    # Particles sorted by cell, so that the members of each cell are one slice.
    order = np.argsort(cell, kind='stable')
    cells, start = np.unique(cell[order], return_index=True)
    # Cell -> (first, last + 1) position of its members in `order`.
    bounds = dict(zip(cells, zip(start, np.r_[start[1:], len(order)])))
    # For each cell holding real giants, draw as many particles as there are real giants in it.
    for c, rows in real.groupby('cell').indices.items():
        if c in bounds:
            members = idx[order[bounds[c][0]:bounds[c][1]]]
            pick[rows] = rng.choice(members, size=len(rows), p=E[members] / E[members].sum())   # with replacement, weight E
    have = pick >= 0
    r, i = real[have], pick[have]   # the real giants with a mock, and their particles
    # Proper-motion noise: the real giant's 2 x 2 covariance [(mas/yr)^2], drawn through its Cholesky factor, times k of its bin.
    cov = np.stack([np.stack([r.pmra_error ** 2, r.pmra_pmdec_corr * r.pmra_error * r.pmdec_error], -1),
                    np.stack([r.pmra_pmdec_corr * r.pmra_error * r.pmdec_error, r.pmdec_error ** 2], -1)], -2)
    noise = r.k.to_numpy()[:, None] * np.einsum('nij,nj->ni', np.linalg.cholesky(cov), rng.normal(size=(len(r), 2)))
    err = np.where(r.survey == 'boss', rng.choice(dfeh, len(r)), 0.0)   # [Fe/H] error [dex]: none for an APOGEE donor
    R = np.hypot(X[i, 0], X[i, 1])   # kpc, in-plane radius of the particle in its own disc frame
    # Mock catalogue: true sky observables plus noise, the donor's errors, measured [Fe/H], and the particle's disc-frame
    # velocities (v_R, v_phi from the in-plane position and velocity; v_z) for the truth.
    # dmu (RR Lyrae distance modulus) is NaN: no distance term for giants.
    cat = dict(ra=s['ra'][i], dec=s['dec'][i], pmra=s['pmra'][i] + noise[:, 0], pmdec=s['pmdec'][i] + noise[:, 1],
               pmra_error=r.pmra_error.to_numpy(), pmdec_error=r.pmdec_error.to_numpy(), pmra_pmdec_corr=r.pmra_pmdec_corr.to_numpy(),
               vlos=s['vlos'][i] + r.evlos.to_numpy() * rng.normal(size=len(r)), evlos=r.evlos.to_numpy(),
               dmu=np.full(len(r), np.nan), feh=feh[i] + err, particle=i, R=R,
               vR=(X[i, 0] * V[i, 0] + X[i, 1] * V[i, 1]) / R, vphi=(X[i, 0] * V[i, 1] - X[i, 1] * V[i, 0]) / R, vz=V[i, 2])
    return cat, int((~have).sum())
```

#### Output files
Each analogue goes to its own file; `merge` joins them into `results/sims_observed.csv` once all are present.

```python
def part(suite, name):
    """Output file of one analogue (PARTS/<suite>_<name>.csv)."""
    return PARTS / f'{suite}_{name}.csv'


def merge():
    """Join the per-analogue files into OUT, in the order of sims.names(); exit with an error, leaving OUT unchanged, if an analogue
    is missing or lacks one of its [Fe/H] bins."""
    names = sims.names()   # list of (suite, name) of every analogue
    # Stop if any analogue has no file yet.
    missing = [f'{s} {n}' for s, n in names if not part(s, n).exists()]
    if missing:
        sys.exit(f'merge: {len(missing)} of {len(names)} analogues have no output (e.g. {", ".join(missing[:3])})')
    # Join in the order of sims.names(); names read as strings (TNG50 names are subhalo ids).
    t = pd.concat([pd.read_csv(part(s, n), dtype={'name': str}) for s, n in names], ignore_index=True)
    # Every analogue must have exactly one row per [Fe/H] bin of lmcold.samples.FEH_BINS.
    rows = t.groupby(['suite', 'name']).population.nunique()
    if (rows != len(samples.FEH_BINS)).any() or t.duplicated(['suite', 'name', 'population']).any():
        sys.exit(f'merge: incomplete or duplicated bins in {list(rows[rows != len(samples.FEH_BINS)].index)[:3]}')
    # Write to a temporary file and rename, so that OUT is never half-written.
    tmp = OUT.with_suffix('.tmp')
    t.to_csv(tmp, index=False, float_format='%.4f')
    os.replace(tmp, OUT)
    print(f'merge: {len(names)} analogues, {len(t)} rows; bins with a non-converged fit: {int((t.fit_warnings > 0).sum())}')
```

#### Observe and fit each analogue (steps 4-6)
For each analogue: draw the mock giants (seeded by the analogue's name, so a rerun gives the same sample), split them into the
five measured-[Fe/H] bins, take the particle truth of each bin near $R$ = 4 kpc, and fit each bin of at least 50 mock giants
(at most `NMAX` of them) by maximum likelihood with the ellipsoid model. One file per analogue.

```python
for suite, name in todo:
    # Skip an analogue already done (by this or another worker).
    if part(suite, name).exists():
        continue
    t0 = time.time()   # s, for the run-time printout
    # Random generator seeded by the CRC32 checksum of 'suite|name': the same mock sample on every rerun.
    rng = np.random.default_rng(zlib.crc32(f'{suite}|{name}'.encode()))
    cat, lost = observe(suite, name, *inputs, rng)
    rows = []   # one dict per [Fe/H] bin
    for j, (lo, hi) in enumerate(samples.FEH_BINS):
        b = (cat['feh'] >= lo) & (cat['feh'] < hi)   # mock giants in bin j of measured [Fe/H]
        c = {k: v[b] for k, v in cat.items()}   # the mock catalogue of this bin
        # Bin size, distinct particles behind its mock giants, and real giants without a mock in this analogue.
        row = dict(suite=suite, name=name, population=j, n=int(b.sum()), n_part=len(np.unique(c['particle'])), lost=lost)
        near = np.abs(c['R'] - kinfit.R_SIG) < TRUTH_DR   # particles within TRUTH_DR of R = 4 kpc (kinfit.R_SIG)
        # Truth [km/s]: mean v_phi and the dispersions of v_R, v_phi, v_z of those particles.
        row |= dict(t_vphi=c['vphi'][near].mean(), t_sigma_R=c['vR'][near].std(), t_sigma_phi=c['vphi'][near].std(),
                    t_sigma_z=c['vz'][near].std(), t_n=int(near.sum()))
        # Fit only bins of at least 50 mock giants (others keep NaN fit columns), a random subset of at most NMAX.
        if b.sum() >= 50:
            sub = rng.permutation(int(b.sum()))[:NMAX]
            d = kinfit.data(*(c[k][sub] for k in COLS))
            # Record RuntimeWarnings of the fit, to count the non-converged maximisations.
            with warnings.catch_warnings(record=True) as caught:
                warnings.simplefilter('always', RuntimeWarning)
                p = ml(d, 'ellipsoid')   # stage-1 maximum likelihood, as for the data
            row['fit_warnings'] = sum('did not converge' in str(w.message) for w in caught)
            # Rotation speed at R = 4 kpc [km/s] and the negative log likelihood of the best fit.
            row |= dict(vrot4=float(kinfit.rotation(p, np.array([kinfit.R_SIG]))[0]), nll=float(kinfit.neg_loglike(p, d, True)))
            # Ellipsoid dispersions, k, outlier fraction (logit), inc and theta [deg], then the whole vector as p_<name>.
            row |= {n: p[P(n)] for n in kinfit.ELLIPSOID + ('k', 'logit_fout', 'inc', 'theta')} | {f'p_{n}': v for n, v in zip(kinfit.PARAMS, p)}
        rows.append(row)
    # Write the analogue's file whole (temporary file, then rename), columns in the fixed order FIELDS.
    tmp = part(suite, name).with_suffix('.tmp')
    pd.DataFrame(rows).reindex(columns=FIELDS).to_csv(tmp, index=False, float_format='%.4f')
    os.replace(tmp, part(suite, name))
    # Progress line: giants per bin, lost giants, fitted and true rotation and sigma_R per bin, time taken.
    print(f'{suite} {name}: {[r["n"] for r in rows]} giants, lost {lost}, '
          f'vrot4 {[round(r.get("vrot4", np.nan), 1) for r in rows]} truth {[round(r["t_vphi"], 1) for r in rows]}, '
          f'sigma_R {[round(r.get("sigma_R", np.nan), 1) for r in rows]} truth {[round(r["t_sigma_R"], 1) for r in rows]} '
          f'({time.time() - t0:.0f} s)', flush=True)
```

## Rotation and dispersion against metallicity

Figure 9.2 compares the rotation, dispersion and rotational support of the observed analogues with those of the LMC
giants. Three choices make the comparison local and like-for-like. The kinematics are read at one radius, $R$ = 4 kpc, inside the
footprint of the real giants, rather than averaged over each galaxy's own density profile: the analogues' metal-rich stars are more
centrally concentrated than the LMC's, and a galaxy-wide average would mix their kinematics with their concentration. The
simulations are matched to the LMC in the cumulative fraction of the formed mass below each population in metallicity (step 3), so the
most metal-poor bin is the first 2.1 per cent of the mass. Figure 9.2 shows the populations against the stellar
mass formed before them, that fraction times the galaxy's total formed mass, the quantity in which disc settling is usually discussed
([Kassin et al. 2012](https://doi.org/10.1088/0004-637X/758/2/106); Pillepich et al. 2019); the analogues formed 1.7 times more stars
than the LMC (median $4.1\times10^{9}$ against $2.4\times10^{9}$ M☉), so each of their bins lies at 1.7 times the LMC's mass. The same comparison
against the mass fraction, where every bin of every galaxy is at the same position, is Figure 2. And
rotational support is $v_{\rm rot}/\sigma_z$: at the LMC's inclination $\sigma_z$ rests mostly on the line-of-sight velocities, so it
is the dispersion least sensitive to the proper-motion error scale and to in-plane streaming, and, because the analogues are more
anisotropic than the LMC ($\sigma_R/\sigma_z$ = 1.27 against 1.19 for [Fe/H] < -1.5), it is the definition most favourable to
them. Other definitions are compared in Table 9.1. The mock giants are drawn at the real giants' positions with all bins
together, so within a bin their spatial distribution is the analogue's own.

At the metal-rich end, the TNG50 analogues bracket the LMC ($v_{\rm rot}$ = 55–87 km/s for the 16th–84th percentiles in the
bin [Fe/H] > -0.5, against 78 km/s), and two of the five Auriga analogues rotate as fast (81 and 88 km/s); all Auriga and
FIRE-2 analogues are hotter than the LMC at every metallicity.

At the metal-poor end no analogue resembles the LMC. In the bin [Fe/H] < -1.5 the analogues rotate at $v_{\rm rot}$ = 15 km/s
(TNG50 median; 16th–84th percentiles 6–29 km/s; Auriga 4–32, FIRE-2 15–34 km/s, lowest to highest) against
the LMC's 57 km/s, and are hotter, $\sigma_z$ = 49 km/s (TNG50; 43–56; Auriga 49, FIRE-2 46) against 32 km/s; the
coldest has 33 km/s. Their rotational support is $v_{\rm rot}/\sigma_z$ = 0.32 (TNG50; 0.11–0.58), 0.30 (Auriga) and
0.67 (FIRE-2); the highest of all 689 analogues is 1.03, against the LMC's 1.76 ± 0.05, and the highest of their
particle values is 1.18. In the next bin, -1.5 < [Fe/H] < -1.1 (2.1–7.7 per cent of the mass), 115 analogues
have $v_{\rm rot}/\sigma_z$ > 1 and 1 reaches the LMC's 1.90 (1.98). At the LMC's own formed mass below that bin,
$1.2\times10^{8}$ M☉, none does: TNG50 populations with the same mass formed before them have $v_{\rm rot}$ = 24 km/s,
$\sigma_z$ = 46 km/s and $v_{\rm rot}/\sigma_z$ = 0.53 (median), against the LMC's 58 km/s, 31 km/s and 1.90. The LMC's most
metal-poor giants ($2.2\times10^{7}$ M☉ formed before them) lie below the mass of the first bin of all but 11 analogues, and
0 of those 11 reach the LMC there. From the third bin on, which begins at 7.7 per cent of the mass and ends at 34 per
cent, the analogues reach the LMC: 58, 125 and 37 of the 689, 689 and 678 analogues whose bins span the LMC's mass in the
three metal-richer bins lie at or above its band there (166, 335 and 138 bin by bin). In the simulations rotational
support rises steeply with metallicity, from a hot, slowly rotating metal-poor component to a disc; in the LMC it changes by a factor
of 3–4 over the same range (to 4.9 for APOGEE+BOSS and 7.7 for APOGEE alone at [Fe/H] > -0.5), and the
metal-poor giants are already rotation-dominated. The shape of their velocity ellipsoid, on the other hand, is ordinary: the LMC's
$\sigma_z/\sigma_R$ = 0.84 for [Fe/H] < -1.5 lies inside the analogues' range (0.71–0.88, 16th–84th percentiles).

The gap does not depend on how the old population is defined. Particle values at $R$ = 4 kpc, without fits or the resampling of
step 4, give a median $v_\phi/\sigma_z$ of 0.24 for the mapped bin, 0.26 for the simulations' own [Fe/H] < -1.5, and 0.28
for the oldest stars holding the same fraction of the formed mass, with maxima of 1.00, 1.19 and 1.38 among the
689 analogues, all below the LMC's 1.76 (`results/sims_populations.csv`, from `scripts/fig_sims_populations.py`). Nor does
it depend on the dispersion used (Table 9.1): with $\sigma_R$ or the root-mean-square of the three components, fitted or
from the particles, no analogue reaches the LMC in the most metal-poor bin, and in the next bin 20 reach its lowest value under
at least one of the six definitions, 1 to 4 under any single one. The choice matters only where the simulated discs begin: in
the third bin 96 to 186 analogues reach the LMC, depending on the definition.

````python
import matplotlib

matplotlib.use('Agg')
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from sims_observe import lmc_knots
from lmcold.layout import panel_grid
from lmcold.sources import ROOT

NMIN_SIM = 300  # mock giants: smaller bins give unstable fits
# (column, axis label) of the three panels: v_rot(4 kpc) [km/s], sigma_z(4 kpc) [km/s], v_rot / sigma_z.
QUANT = [('vrot4', r'$v_{\rm rot}(4\,{\rm kpc})$ [km s$^{-1}$]'), ('sigma_z', r'$\sigma_z(4\,{\rm kpc})$ [km s$^{-1}$]'),
         ('ratio_z', r'$v_{\rm rot}/\sigma_z$')]
LMC_COLOUR = 'k'   # LMC measurement bands black (PI, letter Fig. 2)
RRL_LABEL = r'LMC RR Lyrae ($\sigma_R$ in b, c)'
RRL_QUANT = {'vrot4': 'vrot4', 'sigma_z': 'sigma_R', 'ratio_z': 'ratio_R'}  # the RR Lyrae column shown in each panel
PGRID = np.linspace(0.002, 0.998, 400)  # probability levels of the band
BAND_ALPHA = 0.7  # opacity at the peak density
CONTOURS = {(0.16, 0.84): '-', (0.025, 0.975): '--'}  # quantile pairs of the band drawn as lines (68 and 95 per cent)
YMAX_RATIO = 6.0   # upper limit of the v_rot/sigma_z axis (panel c): room for the top of the LMC band (APOGEE alone, 7.7, is cut)
SUITE_COLOURS = {'TNG50': '#4c72b0', 'Auriga': '#dd8452', 'FIRE-2': '#55a868'}  # one colour per simulation suite


def lmc_band(t, q):
    """Lower and upper edge of the LMC band of quantity q per bin: the giants' 16th-84th percentiles and APOGEE +- 1 sigma."""
    lo, hi = t[f'{q}_p16'].to_numpy(), t[f'{q}_p84'].to_numpy()  # APOGEE + BOSS 16th-84th percentiles
    # APOGEE alone (value and error; columns present only for the bins where it was fitted) widens the band.
    a, e = t.get(f'{q}_apogee'), t.get(f'{q}_apogee_err')
    if a is not None:
        lo, hi = np.fmin(lo, (a - e).to_numpy()), np.fmax(hi, (a + e).to_numpy())
    return lo, hi


def rms(r, prefix='', suffix=''):
    """sigma_RMS of a table with columns <prefix>sigma_R<suffix>, ..._phi, ..._z."""
    return np.sqrt(sum(r[f'{prefix}sigma_{c}{suffix}'] ** 2 for c in ('R', 'phi', 'z')) / 3)


def smooth(x, y, xs):
    """y (per bin at x; along the first axis) at xs, joined by a monotone cubic (PCHIP, no overshoot) in log x."""
    from scipy.interpolate import PchipInterpolator
    return PchipInterpolator(np.log10(x), y, axis=0)(np.log10(xs))


def density_band(a, x, v, colour, label):
    """Draw the distribution of the analogues' values v (DataFrame: analogue x bin) as a band whose opacity is proportional to the
    probability density: per bin a Gaussian KDE gives the quantile function on PGRID; between bins each quantile is joined by
    smooth() (so the distribution shifts rather than splitting into two; quantiles kept non-decreasing in p); the density dp/dy on a fine y grid sets
    the opacity, scaled to its maximum in each column (BAND_ALPHA at the peak); the quantiles of CONTOURS are drawn as thin lines."""
    from matplotlib.colors import LinearSegmentedColormap, to_rgb
    from scipy.stats import gaussian_kde
    Q = []  # quantile function of each bin on PGRID
    for j in range(len(x)):
        u = v[j].dropna().to_numpy()  # values of the analogues in bin j
        # KDE evaluated on a y grid reaching 3 sd beyond the extreme values; its normalised cumulative sum is the CDF.
        ygrid = np.linspace(u.min() - 3 * u.std(), u.max() + 3 * u.std(), 2000)
        cdf = np.cumsum(gaussian_kde(u)(ygrid))
        Q.append(np.interp(PGRID, cdf / cdf[-1], ygrid))
    Q = np.array(Q)  # shape (bins, len(PGRID))
    lx = np.log10(x)
    xs = np.linspace(lx[0], lx[-1], 400)  # fine grid in log x between the first and last bin
    Qx = np.maximum.accumulate(smooth(x, Q, 10 ** xs).T, axis=0)  # (p, x)
    lo, hi = a.get_ylim()
    ys = np.linspace(lo, hi, 600)  # y grid over the current axis range
    # Density dp/dy per x column: derivative of the inverted quantile function; normalised to 1 at each column's peak.
    dens = np.array([np.gradient(np.interp(ys, Qx[:, i], PGRID), ys) for i in range(len(xs))]).T
    dens /= dens.max(axis=0, keepdims=True)
    # Colour map from transparent to the suite colour at opacity BAND_ALPHA.
    cmap = LinearSegmentedColormap.from_list('band', [(*to_rgb(colour), 0.0), (*to_rgb(colour), BAND_ALPHA)])
    a.pcolormesh(10 ** xs, ys, np.where(dens < 0.02, 0, dens), cmap=cmap, shading='nearest', lw=0,
                  edgecolors='none', antialiased=False, zorder=1, rasterized=True)  # tails below 2% of the peak blank
    # The quantiles of CONTOURS as thin lines (solid 68, dashed 95 per cent).
    for levels, ls in CONTOURS.items():
        for pk in levels:
            y = np.array([np.interp(pk, PGRID, Qx[:, i]) for i in range(len(xs))])
            a.plot(10 ** xs, y, ls, color=colour, lw=0.5, zorder=1.2)
    a.fill_between([], [], color=colour, alpha=BAND_ALPHA / 2, lw=0, label=label)  # legend entry


def formed_masses():
    """Total formed stellar mass [Msun] of every analogue (results/sims_formed_mass.csv, scripts/build_sims_formed_mass.py), as a
    Series indexed by (suite, name)."""
    return pd.read_csv(ROOT / 'results/sims_formed_mass.csv', dtype={'name': str}).set_index(['suite', 'name']).m_formed


def kinematics(ax, t, s, x_lmc, X, edges, xlabel, letters, counts, rrl, x_rrl):
    """One row of the figure (module docstring) on the three axes ax: the LMC bins at x_lmc, the RR Lyrae (row rrl of the kinematics
    table) at x_rrl (x of their 16th, 50th, 84th [Fe/H] percentiles), each analogue's bins at X (DataFrame
    indexed by (suite, name), columns = bins), bin edges = (x positions, LMC [Fe/H] labels). The TNG50 median and band are drawn at
    the median TNG50 x of each bin; counts: the numbers written above the LMC band in the v_rot / sigma_z panel. Returns the summary
    rows."""
    rows = []
    for a, (q, lab), letter in zip(ax, QUANT, letters):
        lo, hi = lmc_band(t, q)  # LMC band edges per bin
        for suite, c in SUITE_COLOURS.items():
            # v: values of quantity q, analogue x bin; xv: their x positions in the same layout.
            v = s[s.suite == suite].pivot(index='name', columns='population', values=q).reindex(columns=range(len(x_lmc)))
            xv = X.loc[suite].reindex(v.index)
            # TNG50: median over the analogues at the median x of each bin; Auriga, FIRE-2: one line per analogue.
            if suite == 'TNG50':
                x_tng = xv.median().to_numpy()
                xs = np.geomspace(x_tng[0], x_tng[-1], 200)
                a.plot(xs, smooth(x_tng, v.median().to_numpy(), xs), '-', color=c, lw=1.2, zorder=2)
            else:
                for i, (name, r) in enumerate(v.iterrows()):
                    xa = xv.loc[name].to_numpy()
                    xs = np.geomspace(xa[0], xa[-1], 200)
                    a.plot(xs, smooth(xa, r.to_numpy(), xs), '-', color=c, lw=0.8, label=suite if i == 0 else None)
            # Summary per bin over the analogues of this suite (results/sims_kinematics_summary.csv).
            for k in range(len(x_lmc)):
                col = v[k].dropna()
                rows.append(dict(suite=suite, population=k, quantity=q, median=col.median(), p16=col.quantile(.16),
                                 p84=col.quantile(.84), max=col.max(), n_analogues=len(col), n_above_lmc=int((col >= lo[k]).sum())))
        # Thin vertical lines at the LMC bin edges; log x axis; y from 0.
        for e in edges[0]:
            a.axvline(e, color='0.85', lw=0.5, zorder=0)
        a.set(xlabel=xlabel, ylabel=lab, xscale='log')
        a.set_ylim(0, YMAX_RATIO if q == 'ratio_z' else 1.08 * max(a.get_ylim()[1], np.nanmax(hi)))
        if q == 'ratio_z':
            for k in range(len(x_lmc)):  # above-left of the LMC band at each bin, further left where the band then rises steeply
                above = hi[k] + 0.5 < YMAX_RATIO   # (> 1 to the next bin); where the band leaves the axis, left of it under the top edge
                steep = k + 1 < len(x_lmc) and hi[k + 1] - hi[k] > 1
                a.text(x_lmc[k] * (0.8 if steep else 1.04) if above else x_lmc[k] / 1.6, hi[k] + 0.2 if above else YMAX_RATIO - 0.25,
                       counts[k], ha='right', va='bottom' if above else 'center', fontsize=6)
        # TNG50 probability band (drawn after the y range is set: it fills the axis range).
        density_band(a, x_tng, s[s.suite == 'TNG50'].pivot(index='name', columns='population', values=q),
                     SUITE_COLOURS['TNG50'], 'TNG50 (median, 68, 95 per cent)')
        # one slightly transparent colour, no edge; at least 2 per cent of the (linear) axis so the narrow metal-poor part shows
        pad = np.clip(0.01 * a.get_ylim()[1] - (hi - lo) / 2, 0, None)
        a.fill_between(x_lmc, lo - pad, hi + pad, color=LMC_COLOUR, lw=0, alpha=0.75, label='LMC giants (APOGEE + BOSS)', zorder=1.5)
        # RR Lyrae: open diamond at their median [Fe/H] x, bars over 16th-84th percentiles in x and y.
        r = RRL_QUANT[q]
        a.errorbar(x_rrl[1], rrl[r], xerr=[[x_rrl[1] - x_rrl[0]], [x_rrl[2] - x_rrl[1]]],
                   yerr=[[rrl[r] - rrl[f'{r}_p16']], [rrl[f'{r}_p84'] - rrl[r]]], fmt='D', ms=3.5, mfc='w', mec=LMC_COLOUR,
                   ecolor=LMC_COLOUR, elinewidth=0.7, capsize=0, zorder=3, label=RRL_LABEL)
        a.tick_params(axis='x', which='both', top=False)   # the top edge carries only the [Fe/H] bin edges
        # Top axis: the LMC [Fe/H] of the bin edges.
        top = a.secondary_xaxis('top')
        top.set_xticks(edges[0], [f'{f:g}' for f in edges[1]], fontsize=7)
        top.tick_params(direction='in')
        top.minorticks_off()
        top.set_xlabel('[Fe/H] (LMC)', fontsize=7)
        # Panel letter; x range padded by factors 1.6 and 1.25 beyond the outermost points.
        a.text(0.04, 0.92, f'({letter})', transform=a.transAxes)
        a.set_xlim(min(x_lmc[0], np.nanmin(X.to_numpy())) / 1.6, 1.25 * max(x_lmc[-1], np.nanmax(X.to_numpy())))
    return rows


def percent(p):
    """Fraction p as the text written in panel (c): '0%', one decimal below 1 per cent, else whole per cent."""
    return '0%' if p == 0 else (f'{100 * p:.1f}%' if p < 0.01 else f'{100 * p:.0f}%')


def counts_at_mass(t, s, x_lmc, X, q='ratio_z'):
    """Analogues at or above the lower edge of the LMC band at the LMC's own formed mass below each bin: each analogue's values,
    joined between its bins linearly in log x, read at x_lmc[k]; analogues whose bins do not span x_lmc[k] are not counted.
    Returns a DataFrame (population, n_covering, n_above_lmc)."""
    lo, _ = lmc_band(t, q)
    # Values (analogue x bin) and the x of each analogue's bins in the same order.
    v = s.pivot_table(index=['suite', 'name'], columns='population', values=q).reindex(columns=range(len(x_lmc)))
    xv = X.reindex(v.index).to_numpy()
    rows = []
    for k, xk in enumerate(x_lmc):
        n = above = 0
        for xa, ya in zip(xv, v.to_numpy()):
            ok = np.isfinite(ya) & np.isfinite(xa)  # bins shown (>= NMIN_SIM mock giants)
            # Count the analogue only if its shown bins span xk; interpolate linearly in log x.
            if ok.sum() >= 2 and xa[ok].min() <= xk <= xa[ok].max():
                n += 1
                above += np.interp(np.log10(xk), np.log10(xa[ok]), ya[ok]) >= lo[k]
        rows.append(dict(population=t.population[k], n_covering=n, n_above_lmc=int(above)))
    return pd.DataFrame(rows)


def definitions(t, s):
    """Number of analogues at or above the LMC per bin and definition (module docstring)."""
    # LMC value per definition and bin: the lower of APOGEE + BOSS and APOGEE alone.
    lmc = {'z': np.fmin(t.ratio_z_p16, t.ratio_z_apogee - t.ratio_z_apogee_err),
           'R': np.fmin(t.ratio_R_p16, t.ratio_R_apogee - t.ratio_R_apogee_err),
           'RMS': np.fmin(t.vrot4 / rms(t), t.vrot4_apogee / rms(t, suffix='_apogee'))}
    # Analogue values: fitted (first three) and from the mock giants' particles (t_* columns).
    sim = {'z': s.vrot4 / s.sigma_z, 'R': s.vrot4 / s.sigma_R, 'RMS': s.vrot4 / rms(s),
           'z particles': s.t_vphi / s.t_sigma_z, 'R particles': s.t_vphi / s.t_sigma_R, 'RMS particles': s.t_vphi / rms(s, 't_')}
    floor = pd.concat(lmc, axis=1).min(axis=1)  # lowest LMC value of the three definitions, per bin ('any')
    rows = []
    for j in range(len(t)):
        b = (s.population == j) & (s.n >= NMIN_SIM)  # analogues shown in bin j
        r = dict(population=t.population[j], n_analogues=int(b.sum()))
        for k, v in sim.items():
            r[k] = int((v[b] >= lmc[k.split()[0]][j]).sum())
        r['any'] = int((pd.concat(sim, axis=1)[b].max(axis=1) >= floor[j]).sum())
        r |= {f'lmc_{k}': float(v[j]) for k, v in lmc.items()}
        rows.append(r)
    return pd.DataFrame(rows)
````

#### Helper
`ratio` is the rotational support of a set of particles, $v_\phi/\sigma_z$, each particle weighted by its expected number of giants.

```python
def ratio(sel, X, V, E):
    """E-weighted mean v_phi / standard deviation of v_z of the selected particles.

    sel: (N,) bool; X [kpc], V [km/s]: (N, 3) disc-frame positions and velocities; E: (N,) expected giants per particle."""
    # Cylindrical radius [kpc] and v_phi = (x v_y - y v_x) / R [km/s] of the selected particles; weights w = E.
    R = np.hypot(X[sel, 0], X[sel, 1])
    vphi = (X[sel, 0] * V[sel, 1] - X[sel, 1] * V[sel, 0]) / R
    w = E[sel]
    vz = V[sel, 2]
    return np.average(vphi, weights=w) / np.sqrt(np.average((vz - np.average(vz, weights=w)) ** 2, weights=w))
```

## Details of the mock observation

- *Selection* (`scripts/fetch_tng50.py`, `lmcold.sims`): TNG50-1 centrals (`primary_flag = 1`) in the mass ranges above; the Auriga
  level-3 dwarfs and the FIRE-2 runs named above.
- *Placement* (`scripts/sims_observe.py`, `lmcold.sims.place`): star particles within `sims.R_MAX` = 50 kpc, rotated to the angular
  momentum of the stars within `sims.ALIGN_R` = 10 kpc, put at the LMC's centre, distance, orientation and systemic motion of
  the joint fit, each particle at its own 3D position.
- *Tracer weight*: BOSS-selected red giants per particle from its formed mass, age and metallicity (`results/giant_conversion.csv`),
  from the same PARSEC isochrones and Kroupa initial mass function as the LMC's stellar mass.
- *Metallicity*: the LMC [Fe/H] of the same rank in cumulative formed mass (`lmcold.sims.lmc_scale`).
- *Sampling*: one mock giant per real giant, from the particles in the same disc cell (`CELL_R` × `CELL_PHI`) and line-of-sight
  velocity window (`VLOS_WINDOW`); errors as in step 4.
- *Fit and display*: each bin on its own, maximum likelihood, at most `NMAX` stars; bins of fewer than `NMIN_SIM` mock giants are not
  shown. The alternative definitions of rotational support are in `results/sims_definitions.csv`.

TNG50 and Auriga record the position and velocity of every star particle at its birth. `scripts/build_sims_birth.py` groups the

0.1 Gyr of age. The galaxy moves through the simulation volume while it forms stars, so the birth positions and
velocities are taken relative to a track of the galaxy's centre and velocity against formation time, built from the stars
themselves. Each cohort is rotated to its own spin, and its mass-weighted $v_\phi$ and $\sigma_z$ are measured at
0.5 < $R$ < 3 kpc; the same stars are measured today in the frame of the galaxy's present disc.
The table, `results/sims_birth.csv`, is the input of the Letter's Fig. 3 and of the birth points of the next chapter's figure of the
progenitors.

#### Helpers
`mean_vphi_sigz` measures the rotation and vertical dispersion of a set of stars; `cohorts` does the whole measurement for one
analogue: it reads its stars, builds the track of the galaxy they formed in, and measures every cohort at birth and today.

```python
def mean_vphi_sigz(X, V, w):
    """Mass-weighted mean v_phi and sigma_z [km/s] of particles at X, V (N, 3) in a frame with the spin along +z.

    X [kpc], V [km/s], w (N,) weights [Msun]; sigma_z is about the weighted mean v_z."""
    # v_phi = (x v_y - y v_x) / R: azimuthal velocity about +z [km/s]
    vphi = (X[:, 0] * V[:, 1] - X[:, 1] * V[:, 0]) / np.hypot(X[:, 0], X[:, 1])
    # v_z about the weighted mean, so sigma_z is the dispersion
    vz = V[:, 2] - np.average(V[:, 2], weights=w)
    return np.average(vphi, weights=w), np.sqrt(np.average(vz ** 2, weights=w))


def cohorts(job):
    """One row per birth cohort of analogue job = (suite, name) (columns as results/sims_birth.csv)."""
    suite, name = job
    d = sims.birth(suite, name)   # z = 0 phase space, formation time and birth phase space of every star particle
    X, V, _ = sims.disc_frame(d['x'], d['v'], d['m'])   # today, in the z = 0 disc frame
    # stars within R_TODAY of the centre today, in situ only (exsitu is all False for TNG50)
    use = (np.linalg.norm(d['x'], axis=1) < R_TODAY) & ~d['exsitu']
    track = sims.progenitor_track(d['birth_x'][use], d['a_form'][use], d['birth_v'][use])   # centre and velocity against scale factor
    rows = []
    for lo in np.arange(0, 13.5, COHORT):   # cohort ages lo to lo + COHORT [Gyr]
        # indices of the used stars with ages in [lo, lo + COHORT) [Gyr]
        i = np.flatnonzero(use & (d['age'] >= lo) & (d['age'] < lo + COHORT))
        if len(i) < MIN_COHORT:
            continue
        # age: cohort centre [Gyr]; n: stars of the cohort within R_TODAY
        row = dict(suite=suite, halo=name, age=lo + COHORT / 2, n=len(i))
        # P, B: birth positions [kpc] and velocities [km/s] relative to the track, about the cohort's spin; k: stars at 0.5 < R < R_BIRTH
        P, B, k = sims.birth_phase_space(d['birth_x'][i], d['a_form'][i], d['birth_v'][i], d['m'][i], track, R_BIRTH)
        if k.sum() >= MIN_IN:
            row['vphi_birth'], row['sz_birth'] = mean_vphi_sigz(P[k], B[k], d['m'][i][k])
            row['vphi_same'], row['sz_same'] = mean_vphi_sigz(X[i[k]], V[i[k]], d['m'][i[k]])   # the same stars today
            row['n_birth'] = int(k.sum())
        rows.append(row)
    return rows
```

#### Helpers
`in_annulus` selects the points in one of the two annuli; `radii` counts, for one analogue, the old stars in each annulus at birth
and where the old stars now at 3–5 kpc were born.

```python
def in_annulus(P, ann):
    """Mask of the points P (N, 3) [kpc] with ann[0] < R < ann[1] (cylindrical) and |z| < ann[1]."""
    R = np.hypot(P[:, 0], P[:, 1])
    return (R > ann[0]) & (R < ann[1]) & (np.abs(P[:, 2]) < ann[1])


def radii(job):
    """One row of results/sims_birth_radii.csv for analogue job = (suite, name)."""
    suite, name = job
    d = sims.birth(suite, name)
    X, _, _ = sims.disc_frame(d['x'], d['v'], d['m'])   # positions today [kpc], z = 0 disc frame
    # Stars within R_TODAY [kpc] today and not ex situ; the progenitor's track (centre, velocity against time) from their birth
    # data.
    use = (np.linalg.norm(d['x'], axis=1) < R_TODAY) & ~d['exsitu']
    track = sims.progenitor_track(d['birth_x'][use], d['a_form'][use], d['birth_v'][use])
    # Counters of the output row; per cohort, the annulus masks at birth, the OUTER mask today and the masses [Msun] are collected.
    row = dict(suite=suite, halo=name, n_inner=0, n_outer=0, coh=0, coh_inner=0, coh_outer=0)
    born_inner, born_outer, today, m = [], [], [], []
    for lo in np.arange(OLD_AGES[0], OLD_AGES[1], COHORT):
        # Indices of the stars of this cohort (age lo to lo + COHORT [Gyr]); cohorts with fewer than 20 stars are skipped.
        i = np.flatnonzero(use & (d['age'] >= lo) & (d['age'] < lo + COHORT))
        if len(i) < 20:
            continue
        P, _, _ = sims.birth_phase_space(d['birth_x'][i], d['a_form'][i], d['birth_v'][i], d['m'][i], track, R_BIRTH)   # birth positions [kpc]
        # Masks (n,) of the cohort's stars born in INNER (a) and OUTER (b); counts and cohorts with at least MIN_IN stars in each.
        a, b = in_annulus(P, INNER), in_annulus(P, OUTER)
        row['coh'] += 1
        row['n_inner'] += int(a.sum()); row['n_outer'] += int(b.sum())
        row['coh_inner'] += int(a.sum() >= MIN_IN); row['coh_outer'] += int(b.sum() >= MIN_IN)
        born_inner.append(a); born_outer.append(b); today.append(in_annulus(X[i], OUTER)); m.append(d['m'][i])
    row['n_today'], row['f_born_inner'], row['f_born_outer'] = 0, np.nan, np.nan
    if today:
        # All old cohorts together: t marks the stars in OUTER today; fractions only with at least MIN_TODAY of them.
        a, b, t, m = map(np.concatenate, (born_inner, born_outer, today, m))
        row['n_today'] = int(t.sum())
        if t.sum() >= MIN_TODAY:
            row['f_born_inner'] = m[t & a].sum() / m[t].sum()   # mass fraction of today's 3-5 kpc old stars born at 0.5-3 kpc
            row['f_born_outer'] = m[t & b].sum() / m[t].sum()
    return row
```

```python
import h5py
import numpy as np
import pandas as pd

from lmcold import sims
from lmcold.sources import ROOT

WIDTHS = (0.5, 0.1)          # Gyr, cohort widths (0.1 is the one quoted; 0.5 as the early TNG50 cohorts)
LB_MIN, LB_MAX = 7.5, 13.5   # Gyr, range of lookback times at formation
R_SPIN, MIN_SPIN = 3.0, 30   # kpc and stars: the spin axis is that of the cohort's stars within R_SPIN at birth
R_IN, R_OUT = 0.5, 3.0       # kpc, cylindrical annulus of the measurement
```

#### Helpers
`spin_frame` rotates to the spin of a set of stars, `vphi_sigma_z` measures rotation and vertical dispersion in the annulus, and
`cohorts` reads one analogue and measures each of its cohorts at birth.

```python
def spin_frame(x, v, m):
    """Rotation matrix (rows: new x, y, z axes) with z along the total angular momentum of particles x, v [kpc, km/s], m.

    x, v: (N, 3); m: (N,) [Msun]. Returns a (3, 3) matrix R; positions in the new frame are x @ R.T."""
    L = (m[:, None] * np.cross(x, v)).sum(0)   # total angular momentum [Msun kpc km/s]
    ez = L / np.linalg.norm(L)   # new z: along L
    # New x: perpendicular to both the old z and the new z (along the line of nodes); new y completes a right-handed frame.
    ex = np.cross([0., 0., 1.], ez)
    ex /= np.linalg.norm(ex)
    return np.stack([ex, np.cross(ez, ex), ez])


def vphi_sigma_z(X, V, m):
    """Mass-weighted mean v_phi and sigma_z [km/s] and the number of particles at R_IN < R < R_OUT.

    X [kpc], V [km/s]: (N, 3) in a frame with the spin along +z; m: (N,) [Msun]."""
    R = np.hypot(X[:, 0], X[:, 1])   # kpc, cylindrical radius
    k = (R > R_IN) & (R < R_OUT)   # particles in the annulus
    # Azimuthal velocity (x v_y - y v_x) / R, and v_z about its mass-weighted mean [km/s].
    vphi = (X[k, 0] * V[k, 1] - X[k, 1] * V[k, 0]) / R[k]
    vz = V[k, 2] - np.average(V[k, 2], weights=m[k])
    return np.average(vphi, weights=m[k]), np.sqrt(np.average(vz ** 2, weights=m[k])), int(k.sum())


def cohorts(name):
    """Rows of the table for FIRE-2 analogue name (a key of lmcold.sims.FIRE)."""
    d = sims.FIRE[name]   # folder of the analogue
    # z = 0 snapshot: cosmology from the header, then the star particles (PartType4).
    with h5py.File(d / 'output/snapshot_600.hdf5') as f:
        h, om = f['Header'].attrs['HubbleParam'], f['Header'].attrs['Omega0']
        af, m = f['PartType4/StellarFormationTime'][:], f['PartType4/Masses'][:] * 1e10 / h   # scale factor at formation; mass [Msun]
    with h5py.File(d / 'track/host_coordinates.hdf5') as f:
        x, v = f['star.form.host.distance'][:], f['star.form.host.velocity'][:]   # (N, 3) at formation [kpc, km/s], host frame
    lb = sims.lookback(af, h, om)   # Gyr, lookback time of formation
    near = np.linalg.norm(x, axis=1) < R_SPIN   # stars born within R_SPIN [kpc, 3D distance] of the host centre
    rows = []   # one dict per (width, cohort)
    for width in WIDTHS:
        # Cohorts [lo, lo + width) in lookback time [Gyr].
        for lo in np.arange(LB_MIN, LB_MAX, width):
            c = (lb >= lo) & (lb < lo + width)
            # Skip a cohort with fewer than MIN_SPIN stars to define the spin axis.
            if (c & near).sum() < MIN_SPIN:
                continue
            # Spin axis of the cohort's central stars; all cohort stars rotated to it and measured in the annulus.
            rot = spin_frame(x[c & near], v[c & near], m[c & near])
            vphi, sz, n = vphi_sigma_z(x[c] @ rot.T, v[c] @ rot.T, m[c])
            # tilt [deg]: angle between the new z axis (row 2 of rot) and the host frame's z axis.
            rows.append(dict(halo=d.name, width=width, lb=lo + width / 2, n=n, vphi=vphi, sz=sz,
                             tilt=np.degrees(np.arccos(np.clip(rot[2, 2], -1, 1)))))
    return rows
```
