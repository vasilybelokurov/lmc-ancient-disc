# How the simulated old stars lose their rotation

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/vasilybelokurov/lmc-ancient-disc/blob/main/kb/notebooks/chapters/sims_evolution.ipynb)

The previous chapter showed that no simulated LMC analogue, observed and fitted like the LMC, has a rotating metal-poor population:
their most metal-poor stars have a rotational support $v_{\rm rot}/\sigma_z\approx$ 0.3 against the LMC's 1.8. Here $v_\phi$ (or
$v_{\rm rot}$ for a fitted rotation curve) is the mean azimuthal velocity of a group of stars about the spin axis of the galaxy, $\sigma_z$
the dispersion of their velocities along that axis, and $R$ the cylindrical radius from it. Times are given as redshift $z$ ($z=0$
today, larger in the past) or in Gyr. This chapter asks whether the simulated
old stars were born hot or became hot, and what heated them. A cohort is a group of star particles born within a short interval of
time; the look-back time is the time before the present at which an event happened, and the age of a star today equals the
look-back time of its birth.

## Born rotating

Both TNG50 ([Pillepich et al. 2019](https://doi.org/10.1093/mnras/stz2338)) and Auriga
([Grand et al. 2024](https://doi.org/10.1093/mnras/stae1598)) record each star particle's birth position and velocity, so we can ask
whether the old stars formed hot or were heated later (Figure 3). They formed rotating. Measured relative to the galaxy
they formed in, the stars now 9–12 Gyr old had $v_\phi$ = 36 km/s, $\sigma_z$ = 19 km/s and $v_\phi/\sigma_z$ = 1.91 in the median
TNG50 analogue (16th–84th percentiles over the analogues 1.36–2.56; Auriga 1.61), and 406 of the 666 TNG50 analogues
with such stars reached the LMC's 1.76. Today the same stars have $v_\phi$ = 12 km/s and $\sigma_z$ = 40 km/s ($v_\phi/\sigma_z$ = 0.30,
at most 0.63 at the 84th percentile; Auriga 0.26), and none reaches it. The LMC's three metal-poor bins lie below the TNG50
median at birth, but within its 16th–84th percentiles, at the ages that their metallicity implies ($v_{\rm rot}/\sigma_z$ = 1.76, 1.90
and 2.28 against 1.99, 2.20 and 2.62 at 10.7, 9.9 and 8.4 Gyr; Figure 3c): the LMC's old population still has most of
the rotational support with which the simulated old stars were born. Their vertical dispersion at birth, 19 km/s, is lower than
the LMC's old giants have today (31–32 km/s), so if the LMC's old stars were born as the simulated ones were, they too were
heated, but less: $\sigma_z^2$ grew by about 600 km² s⁻², against 1300 in the simulations. TNG50 forms its young stars as cold as the
LMC's metal-rich giants ($\sigma_z\approx$ 15 km/s at birth); Auriga forms stars of every age with $\sigma_z\approx$ 23–27 km/s. Rotation at
birth is measured less well than dispersion: for stars younger than 1.5 Gyr, which have had little time to change, TNG50 gives
$v_\phi$ = 49 km/s at birth against 57 km/s today, so the birth rotation may be low by up to that difference.

````python
import matplotlib

matplotlib.use('Agg')
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy.stats import norm

from lmcold.constants import LMC_AMR_POVICK24
from lmcold.layout import panel_grid
from lmcold.samples import FEH_BINS
from lmcold.sources import ROOT

PLOT_BIN = 0.5  # Gyr, plotted age bins: median of the cohort values in each bin, per galaxy (display only)
YTOP = dict(vphi=90, sz=65, vs=6)  # upper limits of panels (a)-(c) (PI): room above the curves, and for the legend in (b)
AGE_MAX = 12.59  # Gyr, oldest PARSEC isochrone (lmcold.constants, LMC_AMR_POVICK24 comment)


def lmc_ages():
    """Mean age and 16th-84th percentile ages [Gyr] of each giant [Fe/H] bin under the reference AMR (formed-mass weights)."""
    # Fine age grid [Gyr] (uniform: constant star formation); mean [Fe/H] and Gaussian scatter [dex] of the AMR at each age
    a = np.linspace(0, AGE_MAX, 5000)
    mu, sd = (np.interp(a, LMC_AMR_POVICK24['age'], LMC_AMR_POVICK24[k]) for k in ('feh', 'sigma'))
    out = []
    for lo, hi in FEH_BINS:
        # w: probability that a star of each age has [Fe/H] in the bin (the age weights); c: their cumulative fraction
        w = norm.cdf(hi, mu, sd) - norm.cdf(lo, mu, sd)
        c = np.cumsum(w) / w.sum()
        # Weighted mean age and the ages at which the cumulative weight reaches 0.16 and 0.84
        out.append((np.average(a, weights=w), np.interp(0.16, c, a), np.interp(0.84, c, a)))
    return out
````

Figure 10.1 follows the main progenitors of 60 of the TNG50 analogues through time (the main progenitor is
the galaxy's most massive ancestor at each earlier time, from the merger tree), to see in what conditions the old stars were born and
when and how they lost their rotation. At birth (panel a), rotation is set by the stellar mass the galaxy had at the time: stars born
in progenitors of $10^{7}$–$10^{7.5}$ M☉ rotate at $v_\phi$ = 18 km/s, those born at $10^{9}$–$10^{9.5}$ M☉ at 52 km/s, while their
vertical dispersion stays at 15–21 km/s. The star-forming gas rotates faster than the stars born from it at every mass (26
and 68 km/s in the same two ranges) and is as cold, $\sigma_z$ = 16–21 km/s, close to its effective sound speed (21 km/s; in
TNG50 the pressure of the star-forming gas is set by an effective equation of state); its $v_\phi/\sigma_z$ rises from 1.35 at
$z$ = 4 to 3.8 at $z$ = 0.7. The LMC's metal-poor giants rotate at 57–58 km/s although the LMC had formed only
$10^{7.1}$–$10^{7.8}$ M☉ in more metal-poor stars, a mass at which TNG50's stars are born rotating at 18–22 km/s. Followed
after birth (panel b), the stars born more than 10 Gyr ago lose most of their rotation before the first tracked snapshot, a
median 0.28 Gyr later ($v_\phi/\sigma_z$ from 1.62 to 0.27, and 0.14 after 2 Gyr); those born 8–10 Gyr ago fall from
2.38 to 0.81; those born in the last 4 Gyr keep it (3.48 to 3.09) and lose it only slowly (2.0 after
3 Gyr). The 100-Myr cohorts born at $z$ = 1.5–4 in 20 progenitors followed at every snapshot (panel c)
lose most of the alignment of their orbital planes within 0.5 Gyr: the coherence

$$
C=\frac{\left|\sum m\,\mathbf{j}\right|}{\sum m\,|\mathbf{j}|}
$$

falls from 0.68 to 0.37, while the late control cohorts keep 0.93. Here the sums run over the stars of a cohort,
$m$ is a star particle's mass and $\mathbf{j}=\mathbf{x}\times\mathbf{v}$ its angular momentum per unit mass, with positions $\mathbf{x}$ from
the progenitor's centre and velocities $\mathbf{v}$ relative to the mean velocity of its stars within 10 kpc; a cohort is the
stars formed in the 100 Myr before a snapshot within 3 kpc of the centre, identified by their particle numbers and
measured at each later snapshot within 10 kpc. $C$ compares only the directions of the orbits: it is 1 when all lie in one
plane and turn the same way, does not change if the stars speed up or slow down or the frame is rotated, and is $\approx N^{-1/2}$ for
$N$ randomly oriented orbits (0.08 for the median cohort of 140 stars). Panel (c) shows, in 0.1-Gyr bins, the median over the
galaxies of each galaxy's median. The old stars are not slowed down; their orbits are turned in different directions, and the
cohort's rotation cancels. In the simulations this happens to the stars born more than 10 Gyr ago in galaxies of every mass they
then had; the LMC's old stars kept their rotation, as do the late control cohorts, born at $z\leq$ 0.35.

````python
import matplotlib

matplotlib.use('Agg')
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from lmcold.layout import panel_grid
from lmcold.sources import ROOT

MIN_GAL = 5  # galaxies per plotted bin


def gal_median(d, by, q):
    """Galaxy-weighted median and 16-84 per cent of q per bin `by` (median per galaxy first); bins with >= MIN_GAL galaxies."""
    # Median of q per (bin, galaxy), then grouped by bin.
    g = d.groupby([by, 'halo'], observed=True)[q].median().groupby(level=0, observed=True)
    # Per bin: median and 16th, 84th percentiles over galaxies, and the number of galaxies.
    out = pd.DataFrame({'m': g.median(), 'lo': g.quantile(.16), 'hi': g.quantile(.84), 'n': g.size()})
    return out[out.n >= MIN_GAL]
````

#### Helper
`kin` is the mass-weighted rotation and vertical dispersion of a set of particles in a frame with the spin along $z$.

```python
def kin(X, V, m):
    """Mass-weighted mean v_phi and sigma_z [km/s] in a frame with the spin along +z (particles at R > 0.1 kpc).

    X [kpc], V [km/s]: (N, 3); m: (N,) [Msun]."""
    R = np.hypot(X[:, 0], X[:, 1])
    # Particles on the axis (R <= 0.1 kpc) are dropped (v_phi below divides by R).
    k = R > 0.1
    # v_phi = (x v_y - y v_x) / R [km/s]; v_z about its mass-weighted mean, so sigma_z is the mass-weighted rms.
    vphi = (X[k, 0] * V[k, 1] - X[k, 1] * V[k, 0]) / R[k]
    vz = V[k, 2] - np.average(V[k, 2], weights=m[k])
    return np.average(vphi, weights=m[k]), np.sqrt(np.average(vz ** 2, weights=m[k]))
```

#### Panel (a): the star-forming gas
For each progenitor and gas snapshot: the star-forming gas within 3 kpc defines the spin axis and the mean velocity; then its
rotation, vertical dispersion and sound speed at 0.5 < $R$ < 3 kpc.

```python
def gas_rows(name):
    """Star-forming gas kinematics of one progenitor at GAS_SNAPS (panel a)."""
    # b: the SubLink main branch (SnapNum, Mstar [Msun], ...) of this progenitor.
    b, rows = sims.main_branch(name), []
    for s in GAS_SNAPS:
        p = sims.progenitor(name, s, 'gas', b)
        if p is None or 'gas' not in p:
            continue
        g = p['gas']   # x [kpc], v [km/s], m [Msun], sfr, u (internal energy per unit mass [(km/s)^2])
        # Star-forming gas within 3 kpc (at least 50 cells) defines the velocity origin and the spin axis.
        sf = (g['sfr'] > 0) & (np.linalg.norm(g['x'], axis=1) < 3)
        if sf.sum() < 50:
            continue
        v = g['v'] - np.average(g['v'][sf], axis=0, weights=g['m'][sf])
        # Total angular momentum of that gas; ez its direction; ex, ey complete a right-handed frame (rows of rot).
        L = (g['m'][sf, None] * np.cross(g['x'][sf], v[sf])).sum(0)
        ez = L / np.linalg.norm(L)
        ex = np.cross([0., 0., 1.], ez); ex /= np.linalg.norm(ex)
        rot = np.stack([ex, np.cross(ez, ex), ez])
        # All gas in that frame; the measured cells: star-forming, 0.5 < R < 3 kpc and |z| < 3 kpc (at least 50).
        X, V = g['x'] @ rot.T, v @ rot.T
        R = np.hypot(X[:, 0], X[:, 1])
        k = (g['sfr'] > 0) & (R > 0.5) & (R < 3) & (np.abs(X[:, 2]) < 3)
        if k.sum() < 50:
            continue
        vphi, sz = kin(X[k], V[k], g['m'][k])
        cs = np.sqrt(np.average(5 / 9 * 2 * g['u'][k], weights=g['m'][k]))  # gamma (gamma - 1) u, gamma = 5/3
        rows.append(dict(halo=name, snap=s, z=1 / p['a'] - 1, logMs=np.log10(b['Mstar'][b['SnapNum'] == s][0]), n=int(k.sum()),
                         vphi=vphi, sz=sz, cs=cs))
    return rows
```

#### Panels (a) and (b): the cohorts at birth and later
The birth values of the 0.1-Gyr cohorts as in `scripts/build_sims_birth.py`, with the progenitor's stellar mass at the time; then
the same stars, found by particle number, at every later tracked snapshot in the progenitor's disc frame.

```python
def history_rows(name):
    """Panel (a) birth points with the progenitor's stellar mass, and the panel (b) histories, of one analogue."""
    b = sims.main_branch(name)
    d = sims.birth('TNG50', name)
    ids = sims.progenitor(name, 99, 'track', b)['stars']['ids']  # same particles and order as sims.birth (tests/test_sims.py)
    use = np.linalg.norm(d['x'], axis=1) < 10   # kpc, stars within 10 kpc today, as build_sims_birth.R_TODAY
    # The progenitor's centre and velocity against time, from the birth positions and velocities of these stars.
    track = sims.progenitor_track(d['birth_x'][use], d['a_form'][use], d['birth_v'][use])
    snaps = {s: p for s in TRACK_SNAPS if (p := sims.progenitor(name, s, 'track', b))}
    lb = {s: p['lookback'] for s, p in snaps.items()}   # Gyr, lookback time of each tracked snapshot
    frames = {}   # snapshot: (particle ids, disc-frame X [kpc], V [km/s], m [Msun]) of all the progenitor's stars
    # Each tracked snapshot in the progenitor's disc frame.
    for s, p in snaps.items():
        st = p['stars']
        X, V, _ = sims.disc_frame(st['x'], st['v'], st['m'])
        frames[s] = (st['ids'], X, V, st['m'])
    rows = []
    # Cohorts of width COHORT [Gyr] in age, from 0 to 13.5 Gyr, with at least 100 stars of which 50 have birth phase-space
    # coordinates.
    for lo in np.arange(0, 13.5, COHORT):
        i = np.flatnonzero(use & (d['age'] >= lo) & (d['age'] < lo + COHORT))
        if len(i) < 100:
            continue
        # P, B: birth positions [kpc] and velocities [km/s] in the progenitor's frame at birth; k: stars where they are defined.
        P, B, k = sims.birth_phase_space(d['birth_x'][i], d['a_form'][i], d['birth_v'][i], d['m'][i], track)
        if k.sum() < 50:
            continue
        age = lo + COHORT / 2
        vphi, sz = kin(P[k], B[k], d['m'][i][k])
        s0 = min(lb, key=lambda s: abs(lb[s] - age))   # tracked snapshot nearest the cohort's birth
        rows.append(dict(halo=name, age=age, dt=0.0, vphi=vphi, sz=sz, logMs=np.log10(b['Mstar'][b['SnapNum'] == s0][0])))
        # The same stars at every tracked snapshot after the cohort formed (at least 30 found by particle id).
        cid = ids[i[k]]
        for s, (sid, X, V, m) in frames.items():
            if lb[s] > lo:  # the cohort is not fully formed yet
                continue
            w = np.isin(sid, cid)
            if w.sum() >= 30:
                vphi, sz = kin(X[w], V[w], m[w])
                rows.append(dict(halo=name, age=age, dt=age - lb[s], vphi=vphi, sz=sz))
    return rows
```

#### Panel (c): the coherence of the orbital planes
The stars formed in the `WINDOW` before a snapshot within 3 kpc, followed by particle number at every later snapshot (within 10
kpc), and their coherence $C=|\sum m\,\mathbf{j}|/\sum m\,|\mathbf{j}|$ at each.

```python
def coherence_rows(job):
    """Panel (c): C of 100-Myr cohorts of one progenitor (job = name, cutout kind, snapshots, dtmax), at every later snapshot up to
    dtmax [Gyr] (late control: 0.9, so that the next every-third snapshot, 0.45-0.55 Gyr later at low z, is included)."""
    name, kind, snaps, dtmax = job
    b = sims.main_branch(name)
    S = []   # (snapshot, lookback [Gyr], stars, disc-frame X, V) of each snapshot with stars
    # Every snapshot with stars in the progenitor's disc frame.
    for s in snaps:
        p = sims.progenitor(name, s, kind, b)
        if p and 'stars' in p:
            st = p['stars']
            X, V, _ = sims.disc_frame(st['x'], st['v'], st['m'])
            S.append((s, p['lookback'], st, X, V))
    rows = []
    for i, (s, lb, st, X, V) in enumerate(S):
        k = (st['age_form'] - lb >= 0) & (st['age_form'] - lb < WINDOW) & (np.linalg.norm(st['x'], axis=1) < 3)   # formed in the WINDOW before s
        # At least 30 stars in the cohort.
        if k.sum() < 30:
            continue
        cid, t_form = st['ids'][k], np.average(st['age_form'][k], weights=st['m'][k])   # ids; mean lookback of formation [Gyr]
        for s1, lb1, st1, X1, V1 in S[i:]:
            if t_form - lb1 > dtmax + 1e-6:
                break
            # The cohort's stars within 10 kpc at a later snapshot (at least 20); dt [Gyr] since the mean formation time.
            w = np.isin(st1['ids'], cid) & (np.linalg.norm(st1['x'], axis=1) < 10)
            if w.sum() >= 20:
                rows.append(dict(halo=name, set=kind, s_birth=s, n0=int(k.sum()), dt=t_form - lb1, C=sims.coherence(X1[w], V1[w], st1['m'][w])))
    return rows
```

## What turns the orbits

To find what turns the orbital planes we repeat the first 0.5 Gyr of each cohort with test particles, massless particles moved
in a given gravitational field (Figure 10.2). The 92 cohorts born at $z$ = 1.5–4 in 16 of the 20
progenitors followed at every snapshot (stars formed in the 25 Myr before a snapshot, median 50 stars) are started from the
stars' positions and velocities at the birth snapshot and integrated in the gravity of their own progenitor. The gravity is a
multipole expansion, a sum of spherical harmonics $Y_{lm}$ of degree $l$ and order $m$ with radial coefficients, to $l$ = 6, of all
the particles of the main subhalo within 60 kpc of the stellar centre (made with AGAMA;
[Vasiliev 2019](https://doi.org/10.1093/mnras/sty2672)), made at every snapshot and interpolated linearly in time; the frame follows
the stellar centre, with the inertial acceleration that this adds. The particles are the Subfind members (the particles the halo
finder assigns to the subhalo) of the main subhalo, so the gravity of satellites is not included. Removing terms of the expansion
removes parts of the gravity, and the coherence $C$ of the test particles is measured as for the stars. With all terms the test
particles follow the stars: the median difference in $C$ is -0.002 (16th–84th percentiles -0.10 to +0.09), with a correlation
coefficient of 0.87 over the cohorts and snapshots. Below we quote each variant's loss of $C$ after 0.2–0.35 Gyr, summed over the
cohorts, as a fraction of the real stars' loss, with the 16th–84th percentiles of a bootstrap over the galaxies.

The orbits are turned by the non-axisymmetric gravity of the dark-matter halo (Figure 10.2a). The potential held fixed
at the birth snapshot turns them as much as the evolving one (1.06 of the real loss, 1.01–1.13), so no change of the potential
during the 0.5 Gyr, by accretion or a merger, is needed. Its axisymmetric part about the cohort's spin ($m=0$) gives 0.06, and its
spherical part ($l=0$) 0.14. Making only the gas and stars spherical leaves 0.85 (0.79–0.91); making only the dark matter
spherical leaves 0.35 (0.28–0.41). Without the odd terms, which describe a lopsided or off-centre mass distribution, the loss is
unchanged (0.99).

Which shape of the halo matters is tested in the frozen potential, with the halo changed in its own frame, given by the iterative
inertia tensor of the dark matter within 3 kpc ([Zemp et al. 2011](https://doi.org/10.1088/0067-0049/197/2/30)), whose
eigenvectors are the halo's major, intermediate and minor axes (Figure 10.2b). Rotating the halo so that its minor
axis lies along the stars' spin does not help: the loss is 1.09 (1.02–1.18) of the real one. Making it axisymmetric about its own
minor axis, which keeps its flattening and its tilt but removes its elongation in the plane, lowers the loss to 0.76 overall; making
it axisymmetric about the stars' spin, which keeps only the flattening, removes it (0.07). How the loss divides between the two
remaining parts depends on the angle between the halo's minor axis and the stars' spin at birth. For cohorts born within 45° of
alignment, the elongation in the plane does most of it: the mean loss falls from 0.15–0.29 to 0.06–0.15 when the elongation is
removed. For cohorts born more than 45° from alignment, the tilt of the flattened halo alone gives 0.30, as much as the full halo's
0.29. Both are departures of the potential from axisymmetry about the stars' spin; the flattening alone does nothing. Rotating the
halo into alignment turns its tilt into an elongation in the plane, which turns the orbits as much.

The departure from axisymmetry about the disc is measured by $\epsilon_2=|\Phi_2|/|\Phi_0|$, the ratio of the $l=2$, $m\neq0$ terms
of the dark matter's potential, about the spin of the stars within 3 kpc, to its monopole, at 2 kpc (defined in full in the
next section). For 623 cohorts born at $z$ = 4–1.0 in all 60 progenitors (followed at every third snapshot), the fraction of
their coherence that they lose in the next 0.4–0.5 Gyr, $(C_0-C_1)/(C_0-C_{\rm random})$, where $C_0$ and $C_1$ are the coherence
at birth and one tracked snapshot later and $C_{\rm random}$ = 0.084 that of randomly oriented orbits, rises from 0.02 at
$\epsilon_2\approx$ 0.0020 to 0.39 at 0.0079 and 0.66 at 0.018 (Figure 10.2c). The Spearman rank correlation is
0.54, 0.50 at fixed redshift (partial correlation) and 0.34 within galaxies (each galaxy's mean $\epsilon_2$ and mean
fraction subtracted). How quickly $\epsilon_2$
falls, and why, is the subject of the next section. Three caveats apply. The test particles do not attract each other. The expansion
is not softened, so inside 0.6 kpc its force is up to 15 per cent stronger than the simulation's (median ratio at 300 star
positions 0.3–0.6 kpc from the centre of one progenitor at $z$ = 2.3, against a direct sum over its particles with the TNG50
softening). The early halos are nearly prolate, so their minor axis, and the angle in Figure 10.2b, are not sharply
defined.

The same happens in the two other suites, with stars formed under different models of star formation and feedback. We measured the
public LMC-mass runs of each, the 5 Auriga analogues and the 3 FIRE-2 ones (m11d, m11e and m11h), as above, in 100-Myr cohorts
followed through their high-cadence outputs (not shown). The oldest stars lose their rotation as quickly as in TNG50: those born more
than 10 Gyr ago fall from $v_\phi/\sigma_z$ = 0.78 to 0.20 within 0.25 Gyr in Auriga, and from 0.70 to 0.15
in FIRE-2, and the orbital planes of the cohorts born at $z$ > 1.5 lose their alignment ($C$ from 0.67 to 0.28 and from
0.61 to 0.36 within 0.5 Gyr). The suites differ at birth. Stars born 8–12 Gyr ago have $v_\phi/\sigma_z$ = 1.4–2.6 at
birth in TNG50, 1.2–1.8 in Auriga and 0.4–1.3 in FIRE-2, whose dwarfs never form an ordered disc (even its cohorts
born less than 9 Gyr ago start at $C$ = 0.43), in line with the rarity of stellar discs among the dwarfs of the FIREbox
volume below 10⁹ M☉ ([Benavides et al. 2025](https://doi.org/10.1093/mnras/staf1847)). In Auriga the halo tests of
Figure 10.2b, run on its 20 cohorts at $z$ = 3.6–1.65, give the TNG50 result: with the dark matter made round the loss
falls to 0.26 of the full one, axisymmetric about the stars' spin to 0.08, and with the halo's elongation in the plane removed to
0.43 for cohorts born within 45° of alignment and 0.76 for the others. What turns the orbits in FIRE-2 we did not test.

The Letter's version of this experiment follows $v_\phi/\sigma_z$ instead of $C$. On the 20 Auriga cohorts (`scripts/auriga_scramble.py`,
`results/auriga_scramble.csv`; gravity frozen at the birth snapshot, as the full Auriga snapshots are too far apart to interpolate),
the real stars go from $v_\phi/\sigma_z$ = 1.27 at birth to 0.31, -0.10 and 0.22 at 0.13, 0.26 and 0.43 Gyr; test particles in all
of the gravity go to 0.36, 0.04 and 0.06, with only the dark matter non-spherical to 0.44, 0.08 and 0.18, and in its
part axisymmetric about the cohort's spin they keep their rotation (1.41, 1.33 and 1.42). In TNG50 the real stars go from
1.9 to 0.6 over the same time (medians over the galaxies of each galaxy's median, in the time bins of the Letter's figure).

### The test-particle experiments

`scripts/sims_scramble.py` runs the experiments of Figure 10.2a and b. For each progenitor it reads the snapshots of
the first 0.5 Gyr after each cohort's birth, puts them in a frame centred on the stars, and takes the cohort's stars
at the birth snapshot as initial conditions. The gravity of the progenitor at each snapshot is a multipole expansion made with AGAMA,
written to a file of coefficients; a variant of the gravity is the same file with some coefficients set to zero. Expansions of all
the particles together, of the dark matter alone and of the gas and stars alone allow one kind of mass to be made spherical while
the other keeps its shape. A uniform acceleration is added to each potential so that the frame stays on the stellar centre. The
stars are then integrated as test particles in every variant, and their coherence $C$ and their $v_\phi/\sigma_z$ are measured at
the times of the snapshots, as for the real stars.

The Letter's Fig. 4 uses three versions of the gravity. In the first, all of the progenitor's gravity is kept but frozen at the birth
snapshot: if the test particles lose their rotation in it, no change of the potential in time is needed. In the second, the gas and
stars are made spherical (their expansion is kept only to $l=0$) and the dark matter keeps its full shape, so only the dark matter can
turn the orbits. In the third, all of the gravity is kept only in its terms with $m=0$, which makes it axisymmetric about the cohort's
spin, so that it exerts no torque about that axis. The second and third follow the evolution of the potential from snapshot to
snapshot. The test particles keep their rotation only in the third. A runnable version of this experiment for one cohort is on the
page of the Letter's Fig. 4.

#### Helpers: the frame of each snapshot and the halo's axes
`load` reads one snapshot of a progenitor and moves it to the stars' centre and mean velocity; `frame` builds axes with $z$ along a
given direction; `halo_axes` finds the principal axes of the dark matter.

```python
def shrink(x, m, r0=10., r1=0.5, nmin=20):
    """Shrinking-sphere centre [kpc] of points x (N, 3) with weights m (N,): the weighted mean within radii from r0 down to r1 (15
    steps, each about the previous centre), stopping when fewer than nmin points remain."""
    c = np.zeros(3)  # start at the origin of x
    # 15 radii, logarithmically spaced from r0 to r1 [kpc]; each step re-centres on the points inside the current sphere.
    for rad in np.geomspace(r0, r1, 15):
        w = np.linalg.norm(x - c, axis=1) < rad
        if w.sum() < nmin:
            break
        c = np.average(x[w], axis=0, weights=m[w])
    return c


def load(name, s, b):
    """Particles of snapshot s in the local frame (stellar centre, mean velocity of the stars within 2 kpc), or None.

    Returns dict s, a (scale factor), lb (lookback [Gyr]), st (stars: x [kpc], v [km/s], m [Msun], ids, age_form), dm and gas
    (x, m); positions relative to the stars' centre, star velocities relative to their mean within R_VEL."""
    p = sims.progenitor(name, s, 'full', b)  # full cutout of snapshot s along main branch b (dict per particle type)
    # Skip snapshots without a cutout, stars or dark matter.
    if not p or 'stars' not in p or 'dm' not in p:
        return None
    st = p['stars']
    c0 = shrink(st['x'], st['m'])  # stellar centre [kpc]
    w2 = np.linalg.norm(st['x'] - c0, axis=1) < R_VEL  # stars within R_VEL of it
    v0 = np.average(st['v'][w2], axis=0, weights=st['m'][w2])
    # Positions relative to c0 for every type; velocities relative to v0 [km/s] for the stars.
    out = dict(s=s, a=p['a'], lb=p['lookback'], st=dict(st, x=st['x'] - c0, v=st['v'] - v0), dm=dict(x=p['dm']['x'] - c0, m=p['dm']['m']))
    # Gas, or an empty set if the cutout has none.
    gas = p.get('gas', dict(x=np.zeros((0, 3)), m=np.zeros(0)))
    out['gas'] = dict(x=gas['x'] - c0, m=gas['m'])
    return out


def frame(n):
    """Rows (x, y, z) with z along the unit vector n and x in the simulation x-y plane."""
    ez = n / np.linalg.norm(n)
    ex = np.cross([0, 0, 1.], ez)  # perpendicular to ez and to the simulation z axis
    # Normalise; if n is along the simulation z axis, take the simulation x axis.
    ex = ex / np.linalg.norm(ex) if np.linalg.norm(ex) > 1e-8 else np.array([1., 0, 0])
    return np.stack([ex, np.cross(ez, ex), ez])


def halo_axes(x, rmax=3., tol=1e-3, itmax=50):
    """Rows (major, intermediate, minor) of the iterative reduced inertia tensor of the points x within ellipsoidal radius rmax;
    right-handed, with the largest component of the major and minor axes positive (a fixed sign, whatever the eigensolver returns).

    x: (N, 3) [kpc], centred; rmax [kpc]. Each iteration weights the points by 1 / r_ell^2 inside the ellipsoid of the previous axis
    ratios q = b/a, s = c/a (Zemp et al. 2011, ApJS 197, 30), until q and s change by less than tol."""
    q = s = 1.  # start from a sphere
    E = np.eye(3)  # current axes (rows)
    for _ in range(itmax):
        y = x @ E.T  # coordinates along the current axes
        re2 = y[:, 0] ** 2 + (y[:, 1] / q) ** 2 + (y[:, 2] / s) ** 2   # ellipsoidal radius squared [kpc^2]
        w = (re2 < rmax ** 2) & (re2 > 0)  # points inside the ellipsoid (the centre excluded: weight 1 / re2)
        # Reduced inertia tensor sum x_i x_j / r_ell^2; eigenvalues ascending, so the axes are reversed to major first.
        val, vec = np.linalg.eigh((x[w, :, None] * x[w, None, :] / re2[w, None, None]).sum(0))
        E = vec[:, ::-1].T
        qn, sn = np.sqrt(val[1] / val[2]), np.sqrt(val[0] / val[2])  # new axis ratios b/a, c/a
        if abs(qn - q) < tol and abs(sn - s) < tol:
            break
        q, s = qn, sn
    E[[0, 2]] *= np.sign(E[[0, 2], np.abs(E[[0, 2]]).argmax(1)])[:, None]   # fixed signs: largest component of major, minor positive
    E[1] = np.cross(E[2], E[0])                                              # right-handed (eigh may return either sign)
    return E
```

#### Gravity: multipole expansions and their parts
`expansion` writes the AGAMA multipole expansion of a set of particles to a file; `keep_terms` copies that file with some
coefficients set to zero; `accel` adds the inertial acceleration of the frame that follows the stellar centre; `evolving` joins
the expansions of successive snapshots into one potential interpolated linearly in time.

```python
def expansion(x, m, path):
    """Write the AGAMA Multipole expansion (LMAX, GRID nodes from RMIN to RMAX, no symmetry) of particles x (N, 3) [kpc], m (N,)
    [Msun] to the file path; returns path."""
    agama.Potential(type='Multipole', particles=(x, m), lmax=LMAX, symmetry='none', gridSizeR=GRID, rmin=RMIN, rmax=RMAX).export(path)
    return path


def keep_terms(src, keep, tag):
    """Copy of the Multipole INI src with the coefficients of the (l, m) terms for which keep(l, m) is False set to zero."""
    out, cols = [], None  # output lines; cols: keep flag per column of the current coefficient table
    for line in open(src):
        s = line.rstrip('\n')
        if s.startswith('#radius'):   # header of a coefficient table: columns 'l=..,m=..'
            cols = [True] + [keep(*map(int, c.replace('l=', '').replace('m=', '').split(','))) for c in s.split('\t')[1:]]
        # Data row of a coefficient table: zero the columns of the terms not kept (the radius column is kept).
        elif cols and s and not s.startswith('#'):
            s = '\t'.join(v if k else '0' for v, k in zip(s.split('\t'), cols))
        out.append(s)
    dst = src.replace('.ini', f'_{tag}.ini')  # new file name with the variant tag
    open(dst, 'w').write('\n'.join(out) + '\n')
    return dst


def accel(pots, centres, t):
    """UniformAcceleration potential -a0(t) for the potentials (one per time t) and the frame points of each.

    a0 at each time is the mean force of that time's potential on its frame points (centres: one (K, 3) array [kpc] per time);
    t in AGAMA time units. Subtracting it keeps the frame's origin, the stellar centre, at rest."""
    acc = np.array([p.force(c).mean(axis=0) for p, c in zip(pots, centres)])  # (times, 3) mean force per unit mass
    return agama.Potential(type='UniformAcceleration', file=np.column_stack([t, -acc]))


def evolving(files, t, path):
    """AGAMA Evolving potential from the expansion files (one per time t, AGAMA units), linearly interpolated in time; its
    description is written to path."""
    # AGAMA INI: one line 'time file' per snapshot.
    open(path, 'w').write('[Potential]\ntype=Evolving\ninterpLinear=true\nTimestamps\n' + ''.join(f'{tk:.8f} {f}\n' for tk, f in zip(t, files)))
    return agama.Potential(path)


def particles(c, parts, R):
    """Positions (rotated by R, rows: axes) and masses of the particle types parts ('dm', 'gas', 'st') of snapshot c, within RMAX."""
    X = np.concatenate([c[p]['x'] for p in parts]) @ R.T  # (N, 3) [kpc] in the frame with axes R
    M = np.concatenate([c[p]['m'] for p in parts])  # (N,) [Msun]
    k = np.linalg.norm(X, axis=1) < RMAX  # inside the outer node of the expansion
    return X[k], M[k]
```

#### Rotational support of a set of stars
$v_\phi/\sigma_z$ about a given axis (`support`), the spin of a set (`spin`), the stars at 0.5 < $R$ < 3 kpc that are followed
(`in_ring`), and $v_\phi/\sigma_z$ about the three axes of the output (`supports`).

```python
def support(x, v, m, ez):
    """Mass-weighted v_phi / sigma_z of particles x, v (N, 3) about the unit axis ez (v_phi positive along ez's sense); NaN with fewer
    than 5 particles."""
    if len(m) < 5:
        return np.nan
    F = frame(ez)
    X, V = x @ F.T, v @ F.T  # positions and velocities in a frame with z along ez
    # Azimuthal velocity (x v_y - y v_x) / R [km/s]; vertical velocity relative to the mass-weighted mean.
    vphi = (X[:, 0] * V[:, 1] - X[:, 1] * V[:, 0]) / np.hypot(X[:, 0], X[:, 1])
    vz = V[:, 2] - np.average(V[:, 2], weights=m)
    # Mass-weighted mean v_phi over the mass-weighted rms of v_z.
    return np.average(vphi, weights=m) / np.sqrt(np.average(vz ** 2, weights=m))


def spin(x, v, m):
    """Unit vector of the total angular momentum of particles x, v (N, 3), m (N,)."""
    L = (m[:, None] * np.cross(x, v)).sum(0)  # total angular momentum [Msun kpc km/s]
    return L / np.linalg.norm(L)


def in_ring(x, v, m):
    """The particles at r < R_SUP[1] kpc and cylindrical R > R_SUP[0] kpc about the spin of those within R_SUP[1] (as the stars at
    formation, lmcold.sims.birth_phase_space)."""
    r = np.linalg.norm(x, axis=1)  # spherical radius [kpc]
    # Spin of the particles within R_SUP[1]; cylindrical R about it is |x cross ez|.
    ez = spin(x[r < R_SUP[1]], v[r < R_SUP[1]], m[r < R_SUP[1]])
    return (r < R_SUP[1]) & (np.linalg.norm(np.cross(x, ez), axis=1) > R_SUP[0])


def supports(x, v, m, disc, birth):
    """v_phi / sigma_z about the set's own spin, the galaxy's disc spin and the cohort's spin at birth (keys own, disc, birth); NaN
    when the set is empty (all particles beyond 10 kpc), as C."""
    # Empty set: NaN for all three axes.
    if m.sum() == 0:
        return dict(own=np.nan, disc=np.nan, birth=np.nan)
    return dict(own=support(x, v, m, spin(x, v, m)), disc=support(x, v, m, disc), birth=support(x, v, m, birth))
```

#### One cohort: the real stars, the variants of the gravity and the orbits
`run_cohort` measures the real stars at each later snapshot, builds every variant of the gravity from the expansions of the dark
matter, the baryons and both together, integrates the stars as test particles in each, and measures $C$ and $v_\phi/\sigma_z$ of
the test particles at the snapshot times.

```python
def run_cohort(snaps, i0, ids, t_form, tmp):
    """Rows of results/sims_scramble.csv (without halo) for one cohort.

    snaps: the loaded snapshots of a progenitor (load()); i0: index of the birth snapshot in snaps; ids: particle ids of the cohort;
    t_form: its mass-weighted mean lookback time of formation [Gyr]; tmp: directory for the expansion files. Empty if fewer than
    three snapshots fall within DTMAX of formation."""
    c0 = snaps[i0]  # the birth snapshot
    # Cohort members present at birth within R_MEMBER of the centre.
    w0 = np.isin(c0['st']['ids'], ids) & (np.linalg.norm(c0['st']['x'], axis=1) < R_MEMBER)
    xs, vs, ms = c0['st']['x'][w0], c0['st']['v'][w0], c0['st']['m'][w0]   # the cohort at its birth snapshot
    sel = in_ring(xs, vs, ms)        # the stars whose v_phi/sigma_z is followed: chosen once, at the birth snapshot
    ids_sel = c0['st']['ids'][w0][sel]
    n = (ms[:, None] * np.cross(xs, vs)).sum(0); n /= np.linalg.norm(n)   # the cohort's spin at birth
    R = frame(n)   # integration frame: z along n, fixed
    # Snapshots from birth up to DTMAX after the mean formation time.
    later = [i for i in range(i0, len(snaps)) if t_form - snaps[i]['lb'] <= DTMAX + 1e-6]
    if len(later) < 3:
        return []
    t = np.array([(c0['lb'] - snaps[i]['lb']) / KPC_KMS_GYR for i in later])   # snapshot times since birth, AGAMA units
    # 500 random stars within 2 kpc of each snapshot (rotated by P): the points on which the frame acceleration is averaged.
    pick = lambda c, P: (lambda xc: xc[np.random.default_rng(0).choice(len(xc), min(500, len(xc)), replace=False)])(
        c['st']['x'][np.linalg.norm(c['st']['x'], axis=1) < 2] @ P.T)
    cen = [pick(snaps[i], R) for i in later]
    # The real stars at each later snapshot: C of the members, the galaxy's disc spin, v_phi/sigma_z of the followed stars.
    meas, vs_meas, disc = [], [], []
    for i in later:
        st = snaps[i]['st']
        w = np.isin(st['ids'], c0['st']['ids'][w0]) & (np.linalg.norm(st['x'], axis=1) < R_MEMBER)
        meas.append(sims.coherence(st['x'][w], st['v'][w], st['m'][w]))
        disc.append(sims.disc_frame(st['x'], st['v'], st['m'])[2][2])   # the galaxy's disc spin (simulation frame)
        f = np.isin(st['ids'], ids_sel)
        vs_meas.append(supports(st['x'][f], st['v'][f], st['m'][f], disc[-1], n))
    # Expansions at every later snapshot, in the integration frame: all particles, dark matter only, baryons (gas + stars) only.
    f_all = [expansion(*particles(snaps[i], ('dm', 'gas', 'st'), R), f'{tmp}/s{i}_all.ini') for i in later]
    f_dm = [expansion(*particles(snaps[i], ('dm',), R), f'{tmp}/s{i}_dm.ini') for i in later]
    f_bar = [expansion(*particles(snaps[i], ('gas', 'st'), R), f'{tmp}/s{i}_bar.ini') for i in later]
    pots = {}                                                     # variant: (potential, rotation cohort frame -> integration frame)
    # Evolving, one expansion of all particles: A, B, C, C1, D, D0.
    for v, keep in EVOLVING.items():
        fs = [keep_terms(f, keep, v) for f in f_all]
        pots[v] = (agama.Potential(evolving(fs, t, f'{tmp}/evo_{v}.ini'), accel([agama.Potential(f) for f in fs], cen, t)), np.eye(3))
    # E: all of the gravity, frozen at the birth snapshot.
    p0 = agama.Potential(f_all[0])
    pots['E'] = (agama.Potential(p0, accel([p0, p0], [cen[0], cen[0]], t[[0, -1]])), np.eye(3))
    # Evolving, dark matter and baryons expanded separately: (terms kept of the dark matter, of the baryons).
    for v, (kd, kb) in {'A_split': (ALL, ALL), 'G_DM': (ALL, L0), 'G_bar': (L0, ALL)}.items():
        fd, fb = [keep_terms(f, kd, v) for f in f_dm], [keep_terms(f, kb, v) for f in f_bar]
        both = [agama.Potential(agama.Potential(d), agama.Potential(b)) for d, b in zip(fd, fb)]
        pots[v] = (agama.Potential(evolving(fd, t, f'{tmp}/evo_{v}_dm.ini'), evolving(fb, t, f'{tmp}/evo_{v}_bar.ini'),
                                   accel(both, cen, t)), np.eye(3))

    def frozen(pd_, pb_, c):
        """Dark matter pd_ plus baryons pb_, frozen, with the frame acceleration on the points c."""
        return agama.Potential(pd_, pb_, accel([agama.Potential(pd_, pb_)] * 2, [c, c], t[[0, -1]]))

    # Frozen at birth, separate expansions, the same terms kept in both.
    for v, keep in {'E_split': ALL, 'C_fr': EVEN, 'D0_fr': M0}.items():
        pots[v] = (frozen(agama.Potential(keep_terms(f_dm[0], keep, v)), agama.Potential(keep_terms(f_bar[0], keep, v)), cen[0]), np.eye(3))
    # The halo's shape: H1, H1p (dark matter axisymmetric about its own minor or major axis) and H2 (halo rotated onto the cohort).
    E = halo_axes(c0['dm']['x'])                                  # simulation frame (centred)
    misalign = np.degrees(np.arccos(min(1., abs(E[2] @ n))))      # deg, halo minor axis to the cohort's spin
    for v, P in (('H1', E), ('H1p', E[[1, 2, 0]])):               # integration frame: z' = minor (H1) or major (H1p) axis
        fd = expansion(*particles(c0, ('dm',), P), f'{tmp}/{v}_dm.ini')
        fb = expansion(*particles(c0, ('gas', 'st'), P), f'{tmp}/{v}_bar.ini')
        pots[v] = (frozen(agama.Potential(keep_terms(fd, AXI, 'axi')), agama.Potential(fb), pick(c0, P)), P @ R.T)
    t1 = E[0] - (E[0] @ n) * n; t1 /= np.linalg.norm(t1)   # projection of the halo's major axis on the cohort plane
    T = np.stack([t1, np.cross(n, t1), n])
    c2 = dict(c0, dm=dict(c0['dm'], x=c0['dm']['x'] @ E.T @ T))  # x -> T^T E x: halo axes onto (t1, t2, n)
    fd = expansion(*particles(c2, ('dm',), R), f'{tmp}/H2_dm.ini')
    pots['H2'] = (frozen(agama.Potential(fd), agama.Potential(f_bar[0]), cen[0]), np.eye(3))
    # Orbits of the cohort's stars in each variant, and C and v_phi/sigma_z of the test particles at the snapshot times.
    rows = []
    x, v = xs @ R.T, vs @ R.T
    for var, (pot, Q) in pots.items():
        orbs = agama.orbit(potential=pot, ic=np.column_stack([x @ Q.T, v @ Q.T]), time=t[-1], dtype=object, verbose=False)
        w6 = np.stack([o(t) for o in orbs], axis=1)   # (times, stars, 6): positions and velocities
        Cs = [sims.coherence(s[w, :3], s[w, 3:], ms[w]) for s in w6 for w in [np.linalg.norm(s[:, :3], axis=1) < R_MEMBER]]
        vs_sim = [supports(s[sel, :3] @ Q, s[sel, 3:] @ Q, ms[sel], R @ d, np.array([0., 0., 1.]))   # in the cohort frame R
                  for s, d in zip(w6, disc)]
        rows += [dict(s_birth=c0['s'], z=1 / c0['a'] - 1, n0=len(ms), misalign=misalign, variant=var, t=tk * KPC_KMS_GYR, C_meas=cm,
                      C_sim=cs, **{f'vs_meas_{k}': q for k, q in vm.items()}, **{f'vs_sim_{k}': q for k, q in vsim.items()})
                 for tk, cm, cs, vm, vsim in zip(t, meas, Cs, vs_meas, vs_sim)]
    return rows
```

#### One progenitor
`one` loads every snapshot of a progenitor, finds the cohorts born before each odd snapshot, and runs each.

```python
def one(name):
    """Rows of results/sims_scramble.csv for all cohorts of the FULL progenitor name."""
    b = sims.main_branch(name)  # main-progenitor branch of the galaxy
    # Every snapshot of FULL_SNAPS in the local frame (load), missing ones dropped.
    snaps = [d for d in (load(name, s, b) for s in FULL_SNAPS) if d is not None]
    rows = []
    with tempfile.TemporaryDirectory() as tmp:
        for i, c in enumerate(snaps):
            # Birth snapshots: odd snapshot numbers only.
            if c['s'] % BIRTH_EVERY == 0:
                continue
            st = c['st']
            ar = st['age_form'] - c['lb']   # Gyr before this snapshot at which each star formed
            # Cohort: stars formed within WINDOW before this snapshot, within R_COHORT of the centre.
            k = (ar >= 0) & (ar < WINDOW) & (np.linalg.norm(st['x'], axis=1) < R_COHORT)
            if k.sum() >= MIN_STARS:
                # Mass-weighted mean formation lookback time [Gyr]; run the experiment and tag the rows with the galaxy.
                t_form = float(np.average(st['age_form'][k], weights=st['m'][k]))
                rows += [dict(halo=name, **r) for r in run_cohort(snaps, i, st['ids'][k], t_form, tmp)]
    print(name, 'done', flush=True)
    return rows
```

### The same experiment in Auriga

`scripts/auriga_scramble.py` repeats the experiment of the Letter's Fig. 4 on the Auriga analogues, with the functions of
`scripts/sims_scramble.py`. The progenitor is traced back by the particle numbers of the analogue's stars today. The full Auriga
snapshots, with gas and dark matter, are too far apart to interpolate the gravity over the 0.5 Gyr of the
experiment, so all three versions are frozen at the birth snapshot; the real stars are followed through the star-only outputs written
between the full snapshots.

#### Reading Auriga: the progenitor and its centre
`host_ids` takes the analogue's stars at $z=0$; `centre` and `local` put each snapshot in the frame of the progenitor those stars
trace; `read_full` and `read_stars` read a full snapshot (gas, dark matter, stars) and a star-only output in that frame.

```python
def host_ids(d):
    """IDs of the star particles within R_HOST of the analogue at z = 0 (subhalo 0 of groups_251)."""
    # Star IDs and positions [comoving kpc/h x 1e-3] from every file of the z = 0 snapshot 251.
    ids, x = [], []
    for fn in sorted(glob.glob(str(d / 'snapdir_251/snapshot_251.*.hdf5'))):
        with h5py.File(fn) as f:
            h = f['Header'].attrs['HubbleParam']
            if 'PartType4' in f:
                ids.append(f['PartType4/ParticleIDs'][:]); x.append(f['PartType4/Coordinates'][:])
    # Centre of subhalo 0 (the analogue) in the same comoving units.
    with h5py.File(sorted(glob.glob(str(d / 'groups_251/fof_subhalo_tab_251.*.hdf5')))[0]) as f:
        c = f['Subhalo/SubhaloPos'][0]
    ids, x = np.concatenate(ids), np.concatenate(x)
    # Distance converted to physical kpc at z = 0 (a = 1): x 1e3 / h.
    return ids[np.linalg.norm(x - c, axis=1) * 1e3 / h < R_HOST]


def centre(ids, x, m, host):
    """Shrinking-sphere centre [kpc] started at the median position of the progenitor's stars, refined on all stars."""
    def shrink(c, x, r):
        # Iterate: mean position of the stars inside r, then shrink r by 20 per cent, down to 0.5 kpc or fewer than 30 stars.
        while r > 0.5:
            k = np.linalg.norm(x - c, axis=1) < r
            if k.sum() < 30:
                break
            c, r = x[k].mean(0), 0.8 * r
        return c
    # First pass on the progenitor's stars from 50 kpc, second pass on all stars from 20 kpc.
    k = np.isin(ids, host)
    return shrink(shrink(np.median(x[k], axis=0), x[k], 50.), x, 20.)


def local(st, parts, host):
    """Shift st (stars with ids) and the other parts to the progenitor centre and the velocity origin of the stars within R_VEL."""
    # Centre [kpc], then the mass-weighted mean velocity [km/s] of the stars within R_VEL of it, subtracted from every part.
    c = centre(st['ids'], st['x'], st['m'], host)
    w = np.linalg.norm(st['x'] - c, axis=1) < R_VEL
    v0 = np.average(st['v'][w], axis=0, weights=st['m'][w])
    for p in [st] + parts:
        p['x'], p['v'] = p['x'] - c, p['v'] - v0


def read_full(d, s, host):
    """Full snapshot s: gas, dm (types 1-3) and st (stars with ids and age [Gyr]) about the progenitor; a, lookback [Gyr]."""
    # Collect (x, v, m[, ids, formation scale factor]) per part from every file of the snapshot; types 1-3 are dark matter.
    P = {p: [] for p in ('gas', 'dm', 'st')}
    for fn in sorted(glob.glob(str(d / f'snapdir_{s:03d}/snapshot_{s:03d}.*.hdf5'))):
        with h5py.File(fn) as f:
            hdr, mt = dict(f['Header'].attrs), f['Header'].attrs['MassTable']
            for t, p in (('PartType0', 'gas'), ('PartType1', 'dm'), ('PartType2', 'dm'), ('PartType3', 'dm'), ('PartType4', 'st')):
                if t in f:
                    g = f[t]
                    x = g['Coordinates'][:]
                    # particles without a Masses field take the common mass of their type from the header MassTable
                    m = g['Masses'][:] if 'Masses' in g else np.full(len(x), mt[int(t[-1])])
                    extra = (g['ParticleIDs'][:], g['GFM_StellarFormationTime'][:]) if p == 'st' else ()
                    P[p].append((x, g['Velocities'][:], m) + extra)
    h, a, om = hdr['HubbleParam'], hdr['Time'], hdr['Omega0']
    # Lookback time of the snapshot [Gyr]; then each part converted to physical kpc, km/s and Msun (see the module docstring).
    lb = float(sims.lookback(a, h, om))
    out = dict(a=a, lb=lb)
    for p, L in P.items():
        c = [np.concatenate(q) for q in zip(*L)]
        out[p] = dict(x=c[0] * 1e3 * a / h, v=c[1] * np.sqrt(a), m=c[2] * 1e10 / h)
        if p == 'st':
            ok = c[4] > 0   # wind particles have formation time <= 0
            out[p] = {k: v[ok] for k, v in out[p].items()} | dict(ids=c[3][ok], age=sims.lookback(c[4][ok], h, om) - lb)
    # Move everything to the progenitor's centre and velocity origin.
    local(out['st'], [out['gas'], out['dm']], host)
    return out


def read_stars(d, i, host):
    """Star-only output i: stars (ids, x, v, m) about the progenitor and lookback [Gyr]. Heavy low-resolution particles dropped."""
    # Star particles of every file of the star-only output: IDs, formation scale factor, mass, position, velocity (code units).
    L = []
    for fn in sorted(glob.glob(str(d / f'snapdir_highfreqstars_{i:03d}/snapshot_highfreqstars_{i:03d}.*.hdf5'))):
        with h5py.File(fn) as f:
            hdr = dict(f['Header'].attrs)
            if 'PartType4' in f:
                g = f['PartType4']
                L.append((g['ParticleIDs'][:], g['GFM_StellarFormationTime'][:], g['Masses'][:], g['Coordinates'][:], g['Velocities'][:]))
    ids, af, m, x, v = (np.concatenate(q) for q in zip(*L))
    # Real stars (formation time > 0, not wind particles) lighter than 100x the median mass (the heavy low-resolution particles).
    ok = (af > 0) & (m < 100 * np.median(m))
    h, a, om = hdr['HubbleParam'], hdr['Time'], hdr['Omega0']
    # Physical units: Msun, kpc, km/s; then centred on the progenitor.
    st = dict(ids=ids[ok], m=m[ok] * 1e10 / h, x=x[ok] * 1e3 * a / h, v=v[ok] * np.sqrt(a))
    local(st, [], host)
    return st, float(sims.lookback(a, h, om))
```

#### Gravity and one cohort
`split` separates the dark matter and the baryons for their expansions; `cohort` follows one cohort's real stars through the
high-cadence outputs, builds the three frozen variants of the gravity at the birth snapshot, integrates the stars as test particles
in each and measures $v_\phi/\sigma_z$ and $C$ at the times of the outputs.

```python
def split(F, R):
    """Dark matter and baryons within RMAX in the frame R (rows: axes): ((X, M), (X, M))."""
    out = []
    # Dark matter first, then gas + stars; positions rotated into frame R, kept within RMAX (60 kpc) of the centre.
    for parts in (('dm',), ('gas', 'st')):
        X = np.concatenate([F[p]['x'] for p in parts]) @ R.T
        M = np.concatenate([F[p]['m'] for p in parts])
        k = np.linalg.norm(X, axis=1) < RMAX
        out.append((X[k], M[k]))
    return out


def cohort(name, s, window=WINDOW):
    """Rows of results/auriga_scramble.csv for the cohort born before full snapshot s of analogue name (window: the cohort's age
    range [Gyr]; the earlier measurement used 0.1)."""
    d = sims.AURIGA[name]
    # The progenitor's star IDs and the birth snapshot in its frame.
    host = host_ids(d)
    F = read_full(d, s, host)
    st = F['st']
    # The cohort: stars younger than window [Gyr] at the birth snapshot, within R_COHORT (3 kpc); none if fewer than MIN_STARS (30).
    k = (st['age'] >= 0) & (st['age'] < window) & (np.linalg.norm(st['x'], axis=1) < R_COHORT)
    if k.sum() < MIN_STARS:
        return []
    xs, vs, ms, ids = st['x'][k], st['v'][k], st['m'][k], st['ids'][k]   # the cohort at its birth snapshot [kpc, km/s, Msun]
    n = spin(xs, vs, ms)   # its spin at birth: the z axis of the integration frame
    # Rotation matrix (rows: axes) whose z axis is n; sel: the followed stars of the ring R_SUP (boolean, (N,)).
    R = frame(n)
    sel = in_ring(xs, vs, ms)
    # Columns common to every row: redshift, numbers of cohort and followed stars, angle [deg] between the spin and the dark-matter
    # halo's minor axis (third row of halo_axes).
    meta = dict(halo=name, s_birth=s, z=1 / F['a'] - 1, n0=int(k.sum()), n_sel=int(sel.sum()),
                misalign=float(np.degrees(np.arccos(min(1., abs(halo_axes(F['dm']['x'])[2] @ n))))))
    # First row: the real cohort at t = 0 (v_phi/sigma_z of the followed stars, coherence C of all members).
    rows = [dict(meta, variant='real', t=0.0, vs_meas=support(xs[sel], vs[sel], ms[sel], n), C_meas=sims.coherence(xs, vs, ms))]
    # The real stars at each high-cadence output within DTMAX of the birth snapshot.
    t_out = []
    for i in OUTS:
        with h5py.File(sorted(glob.glob(str(d / f'snapdir_highfreqstars_{i:03d}/*.hdf5')))[0]) as f:
            hd = dict(f['Header'].attrs)
        # Time of this output after the birth snapshot [Gyr], from the header only; outputs outside (0, DTMAX] are skipped.
        dt = F['lb'] - sims.lookback(hd['Time'], hd['HubbleParam'], hd['Omega0'])
        if not 0 < dt <= DTMAX + 1e-6:
            continue
        S, _ = read_stars(d, i, host)
        # The same followed stars (by ID) and the cohort members within R_MEMBER (10 kpc) at this output.
        f_sel = np.isin(S['ids'], ids[sel])
        f_mem = np.isin(S['ids'], ids) & (np.linalg.norm(S['x'], axis=1) < R_MEMBER)
        rows.append(dict(meta, variant='real', t=dt, vs_meas=support(S['x'][f_sel], S['v'][f_sel], S['m'][f_sel], n),
                         C_meas=sims.coherence(S['x'][f_mem], S['v'][f_mem], S['m'][f_mem])))
        t_out.append(dt)
    if not t_out:
        return rows
    t = np.r_[0.0, t_out] / KPC_KMS_GYR   # output times since the birth snapshot, AGAMA units
    # 500 random stars within 2 kpc: the points on which the frame acceleration is averaged.
    inner = st['x'][np.linalg.norm(st['x'], axis=1) < 2] @ R.T
    inner = inner[np.random.default_rng(0).choice(len(inner), min(500, len(inner)), replace=False)]
    (Xd, Md), (Xb, Mb) = split(F, R)
    x, v = xs @ R.T, vs @ R.T   # initial conditions in the integration frame
    # The three variants, all frozen at the birth snapshot; orbits, then v_phi/sigma_z and C of the test particles at each time.
    with tempfile.TemporaryDirectory() as tmp:
        fd, fb = expansion(Xd, Md, f'{tmp}/dm.ini'), expansion(Xb, Mb, f'{tmp}/bar.ini')
        for var, (kd, kb) in VARIANTS.items():
            # Dark-matter and baryon potentials with the terms of this variant; the frame acceleration of their sum is added
            # (accel).
            pd_ = agama.Potential(keep_terms(fd, kd, var) if kd is not ALL else fd)
            pb_ = agama.Potential(keep_terms(fb, kb, var) if kb is not ALL else fb)
            both = agama.Potential(pd_, pb_)
            pot = agama.Potential(pd_, pb_, accel([both, both], [inner, inner], t[[0, -1]]))
            orbs = agama.orbit(potential=pot, ic=np.column_stack([x, v]), time=t[-1] * (1 + 1e-6), dtype=object,
                               verbose=False)   # a little past the last time, where the interpolated trajectory is NaN
            # Phase-space positions at the output times, shape (n_times, N, 6); the integration frame's z is the spin, hence ez =
            # (0, 0, 1).
            w6 = np.stack([o(t) for o in orbs], axis=1)
            for tk, q in zip(t, w6):
                mem = np.linalg.norm(q[:, :3], axis=1) < R_MEMBER
                rows.append(dict(meta, variant=var, t=tk * KPC_KMS_GYR, vs_sim=support(q[sel, :3], q[sel, 3:], ms[sel], np.array([0., 0., 1.])),
                                 C_sim=sims.coherence(q[mem, :3], q[mem, 3:], ms[mem])))
    print(name, s, 'done', meta['n0'], meta['n_sel'], flush=True)
    return rows
```

## How the halo settles

Figure 10.3 follows the centre of the dark-matter halo of the 60 progenitors from 12.5 to 7.5 Gyr ago ($z$ = 4–1),
at every third snapshot (every snapshot from $z$ = 4 to 1.5 for 20 of them; 917 measurements, 10–24 per galaxy).
At each snapshot the origin is the position of the main subhalo, $\mathbf{x}_k$ are the physical positions of the particles $k$
relative to it, $\mathbf{v}_k$ their velocities and $m_k$ their masses, and the disc axis is the direction of the angular momentum
of the star particles within 3 kpc,

$$
\hat{\mathbf{e}}_z=\frac{\mathbf{L}}{|\mathbf{L}|},\qquad \mathbf{L}=\sum_{|\mathbf{x}_k|<r_{\rm in}} m_k\,\mathbf{x}_k\times(\mathbf{v}_k-\bar{\mathbf{v}}),
$$

with $r_{\rm in}$ = 3 kpc and $\bar{\mathbf{v}}$ their mass-weighted mean velocity. Only the dark matter is used below.

The shape of the halo is given by the second-moment tensor of the $N$ dark-matter particles within $r_{\rm in}$,

$$
I_{ij}=\frac{1}{N}\sum_{|\mathbf{x}_k|<r_{\rm in}} x_{k,i}\,x_{k,j},
$$

whose eigenvalues $a^2\geq b^2\geq c^2$ give the axis ratios $b/a$ (intermediate to major) and $c/a$ (minor to major), and whose
eigenvector $\hat{\mathbf{e}}_c$ of the smallest eigenvalue is the minor axis. The misalignment is the angle between the minor axis
and the disc axis,

$$
\psi=\arccos|\hat{\mathbf{e}}_c\cdot\hat{\mathbf{e}}_z|,\qquad 0\leq\psi\leq90^\circ .
$$

Unlike the iterative tensor of the previous section, this one is measured in a sphere, which biases $b/a$ and $c/a$ towards 1
([Zemp et al. 2011](https://doi.org/10.1088/0067-0049/197/2/30)); the true halo is somewhat flatter, and the trends with time are not
affected.

The asymmetry of the potential comes from a multipole expansion of the dark-matter particles within 30 kpc (AGAMA, to
$l$ = 4, no symmetry imposed), in the frame with $\hat{\mathbf{e}}_z$ along the disc axis,

$$
\Phi(r,\theta,\phi)=\sum_{l=0}^{l_{\max}}\sum_{m=-l}^{l}\Phi_{lm}(r)\,Y_{lm}(\theta,\phi),
$$

where $r$, $\theta$ and $\phi$ are the spherical radius, polar angle from $\hat{\mathbf{e}}_z$ and azimuth, $l_{\max}$ = 4, and $Y_{lm}$ are
real spherical harmonics, $\propto\cos m\phi$ for $m>0$ and $\propto\sin|m|\phi$ for $m<0$, normalised to unit mean square over the sphere
(so $\Phi_{00}$ is the spherical average of the potential). The quadrupole ($l=2$) has three parts. The $m=0$ term is a flattening along
the disc axis; it is axisymmetric about the disc, exerts no torque about that axis and does not turn the orbital planes
(Figure 10.2b). The $m=\pm1$ terms ($\propto xz$, $yz$) appear when the halo's principal axes are tilted to the disc
axis, and the $m=\pm2$ terms ($\propto x^2-y^2$, $xy$) when the halo is elongated in the plane of the disc. Their amplitudes at
$r$ = 2 kpc, relative to the monopole, are

$$
\epsilon_{\rm tilt}=\frac{\sqrt{\Phi_{2,1}^2+\Phi_{2,-1}^2}}{|\Phi_{00}|},\qquad
\epsilon_{\rm elong}=\frac{\sqrt{\Phi_{2,2}^2+\Phi_{2,-2}^2}}{|\Phi_{00}|},\qquad
\epsilon_2=\sqrt{\epsilon_{\rm tilt}^2+\epsilon_{\rm elong}^2},
$$

which do not depend on the direction chosen for $\hat{\mathbf{e}}_x$ in the plane. $\epsilon_2$ is the quantity of
Figure 10.2c: with this normalisation it is the root-mean-square over the sphere of the non-axisymmetric quadrupole,
relative to the monopole. With all five $l=2$ terms, $m=0$ included, the same ratio is $\epsilon_{2,\rm all}$, which does not depend
on the orientation of the frame and so bounds $\epsilon_2$ about any axis. For each galaxy the values are binned in 0.5-Gyr steps of
look-back time and their median is taken; Figure 10.3 shows the median of these over the galaxies and its 16th–84th
percentiles (bins with at least 20 galaxies).

The halo lines up with the disc (Figure 10.3a). The misalignment is $\psi\approx$ 39°–47° until 11.5 Gyr ago and falls
to 23° 10 Gyr ago and 9° 7.5 Gyr ago; the spread between galaxies is large early (16th–84th percentiles 17°–69°
12.5 Gyr ago) and smaller later (4°–39° 7.5 Gyr ago). The halo becomes rounder only slowly (Figure 10.3b):
over the same 5 Gyr $c/a$ rises from 0.75 to 0.84 and $b/a$ from 0.87 to 0.92. The quadrupole falls by a factor of three,
$\epsilon_2$ = 0.017 → 0.006 (Figure 10.3c), unequally in its two parts: $\epsilon_{\rm tilt}$ = 0.012 → 0.0025 and
$\epsilon_{\rm elong}$ = 0.011 → 0.004. The tilt term is the one that follows the alignment. For a halo axisymmetric about its
minor axis, with $l=2$, $m=0$ amplitude $\epsilon_{\rm flat}$ in its own frame, a rotation by $\psi$ gives

$$
\epsilon_{\rm tilt}\propto\epsilon_{\rm flat}\,|\sin2\psi|,\qquad \epsilon_{\rm elong}\propto\epsilon_{\rm flat}\sin^2\psi ,
$$

from the $l=2$ rotation matrix. With $\epsilon_{\rm flat}$ scaling as $1-c/a$, the medians predict a fall of $\epsilon_{\rm tilt}$ by
$[\sin 18^\circ/\sin 78^\circ]\times[(1-0.84)/(1-0.75)]$ = 0.31 × 0.64 = 0.20, as measured
(0.0025/0.0122 = 0.21); over the 917 measurements $\epsilon_{\rm tilt}$ correlates with $(1-c/a)|\sin2\psi|$ with a
Spearman rank coefficient of 0.80. The elongation in the plane falls much less than the $\sin^2\psi$ of the last equation would give
(a factor 0.06 between these $\psi$): it comes mostly from the halo's own triaxiality, $b/a<1$ (rank correlation with $1-b/a$ of 0.68),
which changes little.

The tilt therefore disappears as the halo aligns with the disc, by ≈9 Gyr ago, and the remaining asymmetry is the elongation
of the halo in the plane. The total stays above the $\epsilon_2\approx$ 0.003 below which cohorts keep their orbital planes
(Figure 10.2c): 7.5 Gyr ago the median is 0.006, where cohorts lose 10–40 per cent of the coherence they have left
in the next tracked step (medians 0.11 at $\epsilon_2\approx$ 0.0045 and 0.39 at 0.0079), against 0.57–0.66 at the 0.012–0.018 of
cohorts born 10.5–11.9 Gyr ago (median look-back times of those bins). The same holds for the shape and the alignment taken one at a
time. The 623 cohorts lose a median 0.02 of their available coherence if born at $\epsilon_2$ < 0.003, 0.09 if born at
$\psi$ < 10°, 0.03 at $c/a$ > 0.9 and 0.03 at $b/a$ > 0.95, against 0.42–0.48 otherwise (dotted lines in Figure 10.3). Each
level is set by the same rule: it is the edge of the first bin, in the binned cohorts, whose median loss is below 10 per cent. These
thresholds are not independent: the four quantities are correlated with each other and with time. The median halo reaches the
alignment threshold 8 Gyr ago, but reaches neither of the shape thresholds ($c/a$ = 0.84, $b/a$ = 0.92 7.5 Gyr ago). Stars born
early are born in a flattened, misaligned and elongated halo and lose their orbital planes; those born later are born in a halo
aligned with the disc and lose less, but not nothing, because the halo stays elongated in the plane of the disc.

#### Helpers
`disc_axes` gives the frame of the stellar disc; `coeffs` reads the coefficients $\Phi_{lm}$ of an expansion at one radius.

```python
def disc_axes(st):
    """Rows (x, y, z) of the frame with z along the angular momentum of the stars within R_IN, x along z_sim x z."""
    # Stars within R_IN [kpc] of the centre; their velocities relative to the mass-weighted mean [km/s]
    k = np.linalg.norm(st['x'], axis=1) < R_IN
    v = st['v'][k] - np.average(st['v'][k], axis=0, weights=st['m'][k])
    # Total angular momentum (3,) [Msun kpc km/s] and its unit vector, the disc axis
    L = (st['m'][k, None] * np.cross(st['x'][k], v)).sum(0)
    ez = L / np.linalg.norm(L)
    # x in the disc plane along z_sim x z; y completes the right-handed frame; returned as a (3, 3) rotation matrix
    ex = np.cross([0., 0., 1.], ez); ex /= np.linalg.norm(ex)
    return np.stack([ex, np.cross(ez, ex), ez])


def coeffs(pot, r):
    """Phi_lm of an AGAMA Multipole potential at radius r, from its exported table (columns 'l=0,m=0' ...).

    Returns {'l=..,m=..': value [(km/s)^2]}, interpolated linearly in ln r between the table's radial nodes."""
    # AGAMA has no direct accessor for the coefficients: export the potential to a text file and read it back
    with tempfile.NamedTemporaryFile(suffix='.ini') as f:
        pot.export(f.name)
        lines = open(f.name).read().split('\n')
    # The '#Phi' section: a header line of column names (radius, then 'l=..,m=..'), then one row per radial node
    i = lines.index('#Phi') + 1
    head = lines[i].lstrip('#').split('\t')
    # Read the numeric rows until the section ends (blank line, comment or the next '[...]' section)
    rows = []
    for ln in lines[i + 1:]:
        if not ln.strip() or ln.startswith('#') or ln.startswith('['):
            break
        rows.append([float(q) for q in ln.split()])
    # T: (n_nodes, 1 + n_coeffs); column 0 is the radius [kpc]
    T = np.array(rows)
    return {c: np.interp(np.log(r), np.log(T[:, 0]), T[:, j + 1]) for j, c in enumerate(head[1:])}
```

#### The halo of one progenitor at every snapshot
In the frame of the stellar disc: the quadrupole of the dark matter's expansion at `R0` (tilt and elongation parts) and the axis
ratios and misalignment of the dark matter within `R_IN`.

```python
def halo_rows(name):
    """Rows of results/sims_halo_settle.csv for progenitor name: one per snapshot of SNAPS with dark matter and stars."""
    b = sims.main_branch(name)   # main branch of the SubLink tree (centres and velocities per snapshot)
    out = []
    for s in SNAPS:
        # Full cutout if it exists (stars + dark matter); otherwise dark matter and stars from the separate cutouts
        full = sims.progenitor(name, s, 'full', b)
        p, st = (full, full) if full else (sims.progenitor(name, s, 'dm', b), sims.progenitor(name, s, 'track', b))
        if not (p and st and 'dm' in p and 'stars' in st):
            continue
        # Dark-matter positions (N, 3) [kpc] rotated into the disc frame; r their radii [kpc]
        X = p['dm']['x'] @ disc_axes(st['stars']).T
        r = np.linalg.norm(X, axis=1)
        k = (r < R_POT) & (r > 1e-4)   # SubhaloPos is the most bound particle, often a DM one: the expansion needs r > 0
        # Multipole expansion of the dark matter within R_POT (settings of the Settings cell)
        pot = agama.Potential(type='Multipole', particles=(X[k], p['dm']['m'][k]), lmax=LMAX, symmetry='none', rmin=RMIN, rmax=R_POT,
                              gridSizeR=GRID)
        c = coeffs(pot, R0)   # Phi_lm at R0 [(km/s)^2]
        p00 = abs(c['l=0,m=0'])   # the monopole, the spherical average of the potential at R0
        # eps_tilt from the m = +-1 and eps_elong from the m = +-2 quadrupole terms (eq. settle_eps)
        tilt, elong = np.hypot(c['l=2,m=-1'], c['l=2,m=1']) / p00, np.hypot(c['l=2,m=-2'], c['l=2,m=2']) / p00
        w = r < R_IN   # dark matter within the sphere of the shape tensor
        val, vec = np.linalg.eigh((X[w, :, None] * X[w, None, :]).sum(0) / w.sum())   # ascending: minor axis first
        cc, bb, aa = np.sqrt(val)   # minor, intermediate, major axis lengths [kpc] (rms extents)
        # One row: redshift from the scale factor a; misalign = angle [deg] of the minor axis (vec[:, 0]) to the disc axis z
        out.append(dict(halo=name, snap=s, z=1 / p['a'] - 1, lookback=p['lookback'], eps_tilt=tilt, eps_elong=elong,
                        eps2=np.hypot(tilt, elong), misalign=np.degrees(np.arccos(min(1., abs(vec[2, 0])))), b_a=bb / aa, c_a=cc / aa,
                        n_in=int(w.sum())))
    return out
```

## A round halo: self-interacting dark matter

If the asymmetry of the young halo turns the orbits, a halo that is round at its centre would let the old stars keep their rotation.
Self-interacting dark matter (SIDM; [Spergel & Steinhardt 2000](https://doi.org/10.1103/PhysRevLett.84.3760);
[Tulin & Yu 2018](https://doi.org/10.1016/j.physrep.2017.11.004)), in which dark-matter particles scatter off each other with a
cross-section per unit mass $\sigma/m$, gives such a halo: scattering makes the velocities of the dark-matter particles isotropic and
the halo round inside the radius where each particle has scattered about once
([Peter et al. 2013](https://doi.org/10.1093/mnras/sts535)). Cold dark matter (CDM) is the standard, collisionless case. We measured
this in the LMC-mass zoom Halo104 of the SIDM Concerto suite ([Nadler et al. 2025](https://doi.org/10.3847/1538-4357/adf553)), run
from the same initial conditions with CDM and with a velocity-dependent cross-section, characterised by its value at low
relative velocity, $\sigma_0/m$ = 147 cm² g⁻¹, and the velocity scale above which it falls, $w$ = 120 km/s (dark matter only, particle mass
$6.3\times10^{3}$ M☉), at $z$ = 4, 3, 2, 1.5 and 0 (Figure 5). The host's virial mass grows from ≈$8\times10^{9}$ M☉ at $z$ = 4 to
≈$4\times10^{10}$ M☉ at $z$ = 1.5 and ≈$9\times10^{10}$ M☉ today in both runs. Its shape is measured with the
same iterative tensor, in ellipsoidal shells around its densest core, and $\epsilon_2$ from a multipole expansion of its particles.

The CDM halo is elongated at its centre at every redshift: $c/a$ = 0.33–0.50 within 5 kpc and $\epsilon_{2,\rm all}$(2 kpc) = 0.012–0.021.
The SIDM halo is round at its centre already at $z$ = 4 ($c/a$ = 0.87–0.92 within 3 kpc; here 'round' means that every shell out to
that radius has $c/a$ > 0.85), and the round region grows with time, to 11–13 kpc ($c/a$ = 0.89–0.97) from $z$ = 2 on, as
expected as more of the halo has had time to scatter (Peter et al. 2013). Its quadrupole at 2 kpc is 5 to 40 times smaller than in
CDM: $\epsilon_{2,\rm all}$ = 0.0042 at $z$ = 4, 0.0011 at $z$ = 2 and 0.0004 at $z$ = 1.5 and today. The exception is $z$ = 3, during a merger of two
cores of almost equal mass, when the SIDM halo is elongated too ($c/a\approx$ 0.6, $\epsilon_{2,\rm all}$ = 0.0065); it is round again by $z$ = 2.
The CDM values are as large as, or up to 2 times, the TNG50 medians at the same redshifts, as expected for a halo without baryons:
in TNG50 the baryons make the inner halo rounder ([Kazantzidis et al. 2004](https://doi.org/10.1086/423992);
[Despali et al. 2019](https://doi.org/10.1093/mnras/stz273)). The SIDM halo lies below $\epsilon_2$ = 0.003 from $z$ = 2 on and just
above it at $z$ = 4; in TNG50, cohorts born in halos below this value keep their orbital planes. A CDM halo of the LMC's mass is
therefore asymmetric enough at its centre to turn the orbits of its first stars, and an SIDM halo with this cross-section, outside
mergers, is not. This is one halo, simulated without baryons and without a disc; whether a disc forming in an SIDM halo keeps its old
stars rotating needs a hydrodynamical SIDM simulation.

```python
def path(snap):
    """Directory of snapshot snap: the z = 0 or the high-redshift download."""
    return Z0_DIR if snap == 235 else HIGHZ_DIR


def read(run, snap, a):
    """Positions [physical kpc] and masses [Msun] of types 1 + 2, and positions of types 3-5, over the 16 Gadget format-1 files."""
    hi, m, lo = [], [], []   # per file: type 1-2 positions, their masses, type 3-5 positions
    for k in range(16):
        with open(path(snap) / run / f'snapshot_{snap}.{k}', 'rb') as f:
            # Header record (256 bytes between 4-byte markers): six particle counts (int32) and six masses (float64, 1e10 Msun/h).
            f.read(4); h = f.read(256); f.read(4)
            n, mass = np.frombuffer(h[:24], 'i4'), np.frombuffer(h[24:72], 'f8')
            # Position record: 3 float32 per particle, all types in order [comoving Mpc/h], shape (sum n, 3).
            f.read(4); x = np.frombuffer(f.read(12 * n.sum()), 'f4').reshape(-1, 3)
        # Offsets of each type in the position block; types 1-2 with masses [Msun] from the mass table, types 3-5 positions only.
        c = np.r_[0, np.cumsum(n)]
        hi.append(x[c[1]:c[3]]); m.append(np.repeat(mass[1:3] * 1e10 / H, n[1:3])); lo.append(x[c[3]:])
    s = a * 1e3 / H   # comoving Mpc/h -> physical kpc
    return np.concatenate(hi).astype(float) * s, np.concatenate(m), np.concatenate(lo).astype(float) * s


def halos(run, snap, a):
    """Rockstar halos: id, upid, Mvir [Msun], Rvir [physical kpc], centre [physical kpc]."""

    # Msun/h -> Msun; comoving kpc/h -> physical kpc (Rvir); comoving Mpc/h -> physical kpc (centre).
    t['Mvir'] /= H; t['Rvir'] *= a / H
```

#### Centre, shape and quadrupole
`densest` finds the host's densest core, `shell_shape` the axis ratios of an ellipsoidal shell, `eps2_at` the quadrupole of the
expansion relative to its monopole at given radii.

```python
def shrink(x, c, r0, r1, nmin=200):
    """Shrinking-sphere centre of the points x from c, radius r0 down to r1 (20 steps) or until fewer than nmin points."""
    for r in np.geomspace(r0, r1, 20):
        # Points within r of the current centre; stop below nmin, else move to their mean position.
        w = np.linalg.norm(x - c, axis=1) < r
        if w.sum() < nmin:
            break
        c = x[w].mean(0)
    return c


def densest(x, m, c0, rvir):
    """Centre [physical kpc] of the densest core near c0: the N_PEAKS highest peaks of the smoothed counts of the points x (N, 3)
    within CORE_BOX x rvir [kpc] of c0, each refined by a shrinking sphere (CORE_SHRINK), and the one with the most mass m within
    CORE_R kept."""
    h = CORE_BOX * rvir   # half-size of the box [kpc]
    # Counts on a 100^3 grid over the box about c0 (bin edges e per axis), smoothed with a Gaussian of 1 bin.
    D, e = np.histogramdd(x[np.linalg.norm(x - c0, axis=1) < h] - c0, bins=100, range=[(-h, h)] * 3)
    D = gaussian_filter(D, 1)
    # Local maxima (5^3-bin neighbourhood), sorted by height; the N_PEAKS highest.
    pk = np.argwhere(D == maximum_filter(D, 5))
    pk = pk[np.argsort(-D[tuple(pk.T)], kind='stable')][:N_PEAKS]
    # Each peak's bin centre refined by a shrinking sphere CORE_SHRINK (2 -> 0.2 kpc); keep the centre with most mass within CORE_R.
    cs = [shrink(x, c0 + np.array([(e[k][p[k]] + e[k][p[k] + 1]) / 2 for k in range(3)]), *CORE_SHRINK) for p in pk]
    return max(cs, key=lambda c: m[np.linalg.norm(x - c, axis=1) < CORE_R].sum())


def shell_shape(x, m, r1, r2, tol=1e-3, itmax=100):
    """q, s, axes (rows: major, intermediate, minor) and number of points of the ellipsoidal shell r1 <= r_ell < r2.

    x (N, 3) [kpc, centred], m (N,) [Msun], r1, r2 [kpc, major-axis radius]. Iterative reduced inertia tensor (weights m / r_ell^2;
    Zemp et al. 2011, ApJS 197, 30) until q = b/a and s = c/a change by less than tol; NaN for fewer than 50 points."""
    q = s = 1.        # start from a sphere
    E = np.eye(3)     # principal axes (rows) in the input frame
    for _ in range(itmax):
        # Coordinates along the current axes and squared ellipsoidal radius [kpc^2]; points in the shell.
        y = x @ E.T
        re2 = y[:, 0] ** 2 + (y[:, 1] / q) ** 2 + (y[:, 2] / s) ** 2
        w = (re2 >= r1 ** 2) & (re2 < r2 ** 2)
        if w.sum() < 50:
            return np.nan, np.nan, E, int(w.sum())
        # Eigen-decomposition of the reduced inertia tensor sum m x x^T / r_ell^2; axes ordered major, intermediate, minor.
        val, vec = np.linalg.eigh((m[w, None, None] * x[w, :, None] * x[w, None, :] / re2[w, None, None]).sum(0))
        E = vec[:, ::-1].T
        a, b, c = np.sqrt(val[::-1])
        # Converged when q and s change by less than tol.
        if abs(b / a - q) < tol and abs(c / a - s) < tol:
            return b / a, c / a, E, int(w.sum())
        q, s = b / a, c / a
    return q, s, E, int(w.sum())


def eps2_at(x, m, frame, rmax, radii):
    """eps2 and its part without m = 0, at radii, of the Multipole expansion of x (centred) in the frame (rows).

    eps2 = sqrt(sum over m of Phi_2m^2) / |Phi_00| (all five l = 2 terms) and eps2_na the same without Phi_20, each interpolated
    linearly in ln r at radii [kpc]; x (N, 3) [kpc], m (N,) [Msun], particles within rmax [kpc]."""
    k = np.linalg.norm(x, axis=1) < rmax   # particles within rmax
    # Multipole expansion of the particles rotated into the frame, exported to a temporary INI file and read back as text.
    with tempfile.NamedTemporaryFile(suffix='.ini') as f:
        agama.Potential(type='Multipole', particles=(x[k] @ frame.T, m[k]), lmax=LMAX, symmetry='none', gridSizeR=GRID, rmin=RMIN,
                        rmax=rmax).export(f.name)
        lines = open(f.name).read().split('#dPhi/dr')[0].splitlines()
    # Table of Phi_lm(r): header line '#radius', columns 'l=..,m=..'; r [kpc] and Phi_lm per column.
    i = next(j for j, s in enumerate(lines) if s.startswith('#radius'))
    cols = lines[i].split('\t')[1:]
    tab = np.array([[float(v) for v in s.split('\t')] for s in lines[i + 1:] if s.strip()])
    r, P = tab[:, 0], dict(zip(cols, tab[:, 1:].T))
    l2 = [c for c in cols if c.startswith('l=2,')]   # the five l = 2 terms
    tot = np.sqrt(sum(P[c] ** 2 for c in l2)) / np.abs(P['l=0,m=0'])
    non = np.sqrt(sum(P[c] ** 2 for c in l2 if c != 'l=2,m=0')) / np.abs(P['l=0,m=0'])
    return np.interp(np.log(radii), np.log(r), tot), np.interp(np.log(radii), np.log(r), non)
```

## What the comparison shows

Observed as the giants were, simulated LMC-mass galaxies from three suites reproduce, in TNG50, the LMC's rotating metal-rich disc, but
none reproduces its metal-poor population. Their lowest-metallicity 2.1 per cent of stars, by formed mass, are
dispersion-supported ($v_{\rm rot}/\sigma_z\approx$ 0.3, at most 1.0); the LMC's are rotation-supported ($v_{\rm rot}/\sigma_z$ = 1.8). The gap
is not a measurement effect: the same selection, errors and fit recover the analogues' own $v_{\rm rot}/\sigma_z$ (in the most
metal-poor bin to +0.07, 16th–84th percentiles -0.11 to +0.24), and their particle values without any fitting are as low; nor does it
depend on whether the old population is defined by metallicity rank, by the simulations' own metallicity or by age.

Two limitations remain. The TNG50 analogues may be heated by their dark-matter particles, which are about 5 times more massive than
their star particles (4.5 × 10⁵ against 8.5 × 10⁴ M☉; [Ludlow et al. 2021](https://doi.org/10.1093/mnras/stab2770)); the 12 times
better resolved Auriga and FIRE-2 zooms give the same result, but they are 5 + 3 galaxies. The analogues are not selected for the
LMC's history: its first passage, its interaction with the SMC, or a quiet early assembly. Within these limits, the simulations do not
form their earliest, most metal-poor stars in a hot spheroid: they form them rotating, like the LMC's, and then heat them, early stars
within a few hundred Myr of their birth (Figure 3 and Figure 10.1). The heating is the turning of their
orbital planes by the gravity of the young dark-matter halo, which is not axisymmetric about the disc: elongated in its plane or tilted
to it (Figure 10.2); the tilt goes as the halo lines up with the disc, by ≈9 Gyr ago, but the elongation in the
plane stays (Figure 10.3). In the LMC this heating was much weaker, or the old stars have since been spun up. A halo whose
centre was nearly axisymmetric when the first stars formed would explain the first possibility; an SIDM halo of the LMC's mass is round
enough at its centre from $z\approx2$ on, outside mergers, but a CDM halo is not (Figure 5).

## Details

### Stars at birth

`scripts/build_sims_birth.py` measures the TNG50 analogues with birth data (all stars within 10 kpc today) and the Auriga analogues
(in-situ stars only), in cohorts of 0.1 Gyr of age with at least 100 stars. The galaxy in which the stars formed is followed
through time from their recorded birth positions and velocities: the stars are sorted by formation time and taken in groups of
60; each group gives a centre (shrinking sphere) and a velocity (median of the inner half), median-filtered over 5 groups and
interpolated in the cosmological scale factor $a=1/(1+z)$. Birth positions are in physical kpc relative to this track at each star's formation time. The frame has
$z$ along the angular momentum of the cohort within 3 kpc; $v_\phi$ and $\sigma_z$ are mass-weighted at 0.5 < $R$ < 3 kpc (the inner cut
guards against errors in the track's centre), with at least 50 stars; today they are measured in the $z=0$ disc frame.

### Numerical heating

Collisional heating of the stars by the more massive dark-matter particles (Ludlow et al. 2021) raises velocity dispersions; it cannot
turn orbital planes, which is what removes the rotation here. Two results of this chapter argue against it as the cause. Test
particles in a smooth multipole expansion of the progenitor's gravity, frozen at birth and free of particle noise, lose their
coherence as fast as the real stars (Figure 10.2a). And the Auriga and FIRE-2 zooms, with lighter dark-matter
particles, give the same result.

### Normalisation of the multipole coefficients

AGAMA's multipole expansion writes $\Phi(r,\theta,\phi)=\sum\Phi_{lm}(r)Y_{lm}$ with real $Y_{lm}$ of unit mean square over the sphere
($Y_{00}=1$, $Y_{l0}=\sqrt{2l+1}\,P_l(\cos\theta)$, with $P_l$ the Legendre polynomials), so $\Phi_{00}$ is the spherical average of the
potential, and the same normalisation holds for every $m$; $\epsilon_2$ is therefore the root-mean-square over the sphere of the
non-axisymmetric quadrupole relative to the monopole, and $\sum_m\Phi_{2m}^2$ does not change when the frame is rotated, so
$\epsilon_2\leq\epsilon_{2,\rm all}$. We checked this on a [Miyamoto & Nagai (1975)](https://doi.org/10.1093/pasj/27.4.533) disc sampled with
$5\times10^{5}$ particles and expanded as in the scrambling experiment: at 1.9 kpc, $\Phi_{00}$ equals the spherical average of the
expansion's potential and $\Phi_{20}$ equals $\sqrt5\langle\Phi P_2\rangle$, each to a fraction better than $10^{-6}$, and rotating the disc
changes $\sum_m\Phi_{2m}^2$ by a fraction smaller than $10^{-8}$. In `scripts/sims_halo_settle.py` each $\Phi_{lm}(r)$ is a spline in $\ln r$ on 40 radii from
0.2 to 30 kpc fitted to the particles (AGAMA Multipole, $l_{\max}$ = 4), not a fixed radial basis.

### The SIDM measurement

`scripts/fetch_concerto.py` and `scripts/sims_sidm_shapes.py` use the CDM and SIDM runs of Halo104, dark matter only, the two finest
particle types. The host is the distinct halo of highest virial mass of the halo catalogue with no particle of the coarser types within
its virial radius $R_{\rm vir}$ (the radius within which the mean density reaches the virial overdensity of the halo finder). Its centre is the densest core: the five highest peaks of the smoothed particle counts within
0.5 $R_{\rm vir}$, each refined by a shrinking sphere from 2 to 0.2 kpc, and the one with most mass within 2 kpc kept (a single
shrinking sphere from $R_{\rm vir}$ lands between the two cores during the $z$ = 3 merger). The shape is the iterative reduced inertia
tensor (Zemp et al. 2011) in logarithmically spaced ellipsoidal shells from 0.4 kpc to $R_{\rm vir}$, shells with at least
$10^{4}$ particles. The quadrupole is an AGAMA multipole expansion of the particles within 2 $R_{\rm vir}$ ($l_{\max}$ = 4, 50 radial
nodes from 0.1 kpc), with the polar axis along the minor axis of the 1.5–2.5 kpc shell; $\epsilon_{2,\rm all}$ (all $m$) and $\epsilon_2$
($m\neq0$) are read at 2 kpc.
