## The quantities

The old population of the Large Magellanic Cloud (LMC) is traced by two kinds of star. The red giants have proper motions from
*Gaia* DR3 ([Gaia Collaboration, Vallenari et al. 2023](https://doi.org/10.1051/0004-6361/202243940)) and line-of-sight velocities
and iron abundances from two spectroscopic surveys: APOGEE ([Majewski et al. 2017](https://doi.org/10.3847/1538-3881/aa784d); data
release 17, [Abdurro'uf et al. 2022](https://doi.org/10.3847/1538-4365/ac4414)) and the BOSS spectrograph of SDSS-V
([Kollmeier et al. 2026](https://doi.org/10.3847/1538-3881/ae0576)), whose spectra are analysed by the BOSSNet code
([Sizemore et al. 2024](https://doi.org/10.3847/1538-3881/ad291d)). The iron abundance [Fe/H] is the logarithm of the
iron-to-hydrogen ratio relative to the Sun's; the giants are divided into five bins of measured [Fe/H], with edges at
-1.5, -1.1, -0.8 and -0.5. The RR Lyrae stars of the OGLE-IV survey
([Udalski et al. 2015](https://doi.org/10.48550/arXiv.1504.05966); [Soszyński et al. 2016](https://doi.org/10.48550/arXiv.1606.02727))
have *Gaia* proper motions and photometric [Fe/H], but no line-of-sight velocities. A 'group' below is one giant bin or the RR Lyrae.
Stars are selected by their angular distance $\rho$ from the LMC centre of
[van der Marel & Kallivayalil (2014)](https://doi.org/10.1088/0004-637X/781/2/121) (vdMK14 below): the default sample has
2 < $\rho$ < 9°.

The kinematic model, described with the velocity field in chapter 6, places each group on a thin disc seen in perspective. Its
geometry is the kinematic centre (the centre of the model disc), offset by $\Delta x$ (east) and $\Delta y$ (north) from the vdMK14 centre, in degrees; the
inclination $i$ of the disc ($i=0$ face-on); and the position angle $\theta$ (north through east) of the line of nodes, the line along
which the disc plane crosses the plane of the sky. Each star then has a cylindrical radius $R$ in the disc plane and an azimuth $\phi$
measured from the line of nodes. The mean rotation speed is

$$
v_{\rm rot}(R)=v_0\tanh(R/R_0)\,[1+s\,(R-R_\sigma)],
$$

with a level $v_0$, a turnover radius $R_0$, an outer slope $s$ and a reference radius $R_\sigma$ = 4 kpc. Because the
three parameters are correlated, we quote the rotation speed at $R_\sigma$, $v_{\rm rot}(4\,{\rm kpc})=v_0\tanh(4\,{\rm kpc}/R_0)$. In the symmetric disc (Level 1) the stars
move on circles; Level 2 adds non-circular streaming with a two-fold ($m=2$) pattern in azimuth. The random motions about the mean form
a velocity ellipsoid aligned with the disc, with dispersions $\sigma_R$ (radial), $\sigma_\phi$ (azimuthal) and $\sigma_z$
(vertical) at $R_\sigma$; all three change with radius by the same factor $e^{\gamma(R-R_\sigma)}$, so $\gamma$ (kpc$^{-1}$) is the
logarithmic radial gradient of the dispersions and $1-e^\gamma$ their fractional decline per kpc. Rotational support is the ratio of
the rotation speed to a dispersion, $v_{\rm rot}/\sigma_R$ or $v_{\rm rot}/\sigma_z$. The *Gaia* proper-motion errors of a group are
multiplied by one scale factor $k$, and a fraction $f$ of its stars is assigned to a broad outlier component. In stage 1 each group is
fitted on its own, with its own centre, systemic motion (the proper motion and line-of-sight velocity of the centre), orientation
and streaming; in stage 2 all groups are fitted jointly and share
these, each keeping its own rotation curve and ellipsoid. The fits sample the posterior probability of the parameters with a Markov
chain; we quote its median and its 16th–84th percentiles. Where a fit is instead a maximum of the likelihood $L$, its errors come from
the inverse Hessian (the curvature of $\ln L$), and two models of the same stars are compared by the difference of their maximum
log-likelihoods, $\Delta\ln L$. A mock is a simulated data set drawn from a known model at the positions and with the
errors of the real stars, and fitted like the data.

## Rotational support against metallicity

Figure 7.1 and Table 7.1 give the result. From the most metal-poor to the most metal-rich giants the rotation
speed at 4 kpc rises from 57.1 to 77.6 km/s, and all three dispersions fall: $\sigma_R$ from
38.7 to 20.9 km/s, $\sigma_\phi$ from 32.0 to 17.3 km/s and
$\sigma_z$ from 32.4 to 15.8 km/s. The rotational support rises accordingly, $v_{\rm rot}/\sigma_R$
from 1.48 to 3.72 and $v_{\rm rot}/\sigma_z$ from 1.76 to 4.91. Most of
the change happens above [Fe/H] = -1.1: the two metal-poor bins differ by 1.3 km/s in $v_{\rm rot}$
and by at most 2.1 km/s in each dispersion, while from -1.1 to the metal-rich end $v_{\rm rot}$ grows
by 15 km/s and $\sigma_R$ and $\sigma_z$ drop by 13 and 12 km/s. Rotation
and dispersion both change, by comparable amounts.

The ellipsoid has the same ordering of axes in every giant bin: $\sigma_R$ is the largest, and $\sigma_\phi$ and $\sigma_z$ are
smaller and nearly equal, $\sigma_\phi/\sigma_R$ = 0.77–0.84 and $\sigma_z/\sigma_R$ =
0.76–0.84. The ratios differ between bins by more than their errors ($\sigma_z/\sigma_R$ =
0.83 ± 0.01 and 0.76 ± 0.01 in the two metal-rich bins, and 0.52 for the APOGEE giants alone above
-0.5), so the shape is similar, not identical, and the measured bins mix populations. The vertical support is therefore
everywhere higher than the radial one, $v_{\rm rot}/\sigma_z$ exceeding $v_{\rm rot}/\sigma_R$ by
19–32 per cent. All dispersions decline outwards, by
4.5–5.1 per cent per kpc in the two metal-poor bins and
6.4–8.5 per cent per kpc above [Fe/H] = -1.1, and the outer slope is
negative in every group ($s$ = -0.05 to -0.11 kpc$^{-1}$): the rotation curves of the giants peak at
$R$ = 4.6–5.0 kpc and decline beyond. The values depend little on the radial range of the sample:
fitted at 4 < $\rho$ < 9°, the giants' $\sigma_R$ and $\sigma_z$ change by at most
1.3 km/s, $\sigma_\phi$ falls by 0.8–3.1 km/s, and
$v_{\rm rot}(4\,{\rm kpc})$ rises by 0.2–3.2 km/s, so the support ratios change by at most
0.2. We take these shifts as the systematic error of the values in Table 7.1.

The RR Lyrae have proper motions only, and their errors, 64 km/s per component (median), are larger than their dispersions. Fitted at 2 < $\rho$ <
9° with one error scale, the crowded inner disc, where *Gaia* underestimates its errors most, sets $k$ =
1.27 and the fit removes the excess scatter by a dispersion that falls by 43 per cent per kpc
($\gamma$ = -0.57 kpc$^{-1}$), which no other group shows; beyond 4° one $k$ =
1.13 describes the errors (Figure 7.7) and the gradient is $\gamma$ = -0.05
kpc$^{-1}$, as for the giants. We therefore quote the RR Lyrae at 4 < $\rho$ < 9°:
$v_{\rm rot}(4\,{\rm kpc})$ = 63.1 ± 2.3 km/s, $\sigma_R$ = 38.1 (+2.9/−3.1) km/s and $\sigma_\phi$ =
23.9 (+4.0/−4.8) km/s, and $v_{\rm rot}/\sigma_R$ = 1.66 (+0.17/−0.14) (16th–84th percentiles about the median). Their
$\sigma_z$, constrained by the proper motions alone, is not measured. The RR Lyrae thus have the radial dispersion and radial support
of the two metal-poor giant bins ($\sigma_R$ = 38.0–38.7 km/s, $v_{\rm rot}/\sigma_R$ =
1.48–1.54), close to their median [Fe/H] of -1.38 (16th–84th percentiles
-1.70 to -1.08), and, compared over the same radii (4 < $\rho$ <
9°, where the two metal-poor giant bins give $v_{\rm rot}(4\,{\rm kpc})$ = 60.2 and
61.5 km/s), the same rotation to within 3 km/s. Their $\sigma_\phi$ is less certain than its
error suggests: it is correlated with $k$ (correlation coefficient -0.7) and rises to about
28 km/s in the posterior draws where $\sigma_z$ takes values like those of the giants
(20–35 km/s), against 29.9–32.0 km/s in the
metal-poor giant bins, so we do not claim a difference in $\sigma_\phi$.

````python
import pandas as pd
from lmcold.tex import fout_percent
# Rotational support per population (s) and the stage-1 fits: giants from the 2-9 deg run, RR Lyrae from the 'outer' run (4 < rho < 9 deg)
s = pd.read_csv('results/kin_support_ellipsoid.csv').set_index('population')
fits = pd.concat([pd.read_csv('results/kin_mcmc_data_ellipsoid.csv').query('stage == 1 and population != "RR Lyrae"'),
                  pd.read_csv('results/kin_mcmc_data_ellipsoid_outer.csv').query('stage == 1 and population == "RR Lyrae"')])
# par: one value (median, p16 or p84) of a fitted parameter; pm: median with its 16th-84th percentile offsets
par = lambda pop, n, c='median': fits[(fits.population == pop) & (fits.param == n)][c].iloc[0]
def pm(m, lo, hi, d):
    return f'{m:.{d}f} (+{hi - m:.{d}f}/−{m - lo:.{d}f})'
# One row per population: rotation at 4 kpc, dispersions, their radial gradient γ = d ln σ / dR, ratios, error scale k, outlier fraction f
rows = {}
for pop, r in s.iterrows():
    q = lambda c, d: pm(r[c], r[f'{c}_p16'], r[f'{c}_p84'], d)
    rows[pop] = {'N': int(r.n), 'v_rot(4 kpc) [km/s]': q('vrot4', 1), 'σ_R [km/s]': q('sigma_R', 1), 'σ_φ [km/s]': q('sigma_phi', 1),
                 'σ_z [km/s]': '–' if pop == 'RR Lyrae' else q('sigma_z', 1),
                 'γ [1/kpc]': pm(par(pop, 'dlnsig_dR'), par(pop, 'dlnsig_dR', 'p16'), par(pop, 'dlnsig_dR', 'p84'), 3),
                 'v_rot/σ_R': q('ratio_R', 2), 'v_rot/σ_z': '–' if pop == 'RR Lyrae' else q('ratio_z', 2),
                 'k': pm(par(pop, 'k'), par(pop, 'k', 'p16'), par(pop, 'k', 'p84'), 2),
                 'f [%]': f"{fout_percent(par(pop, 'logit_fout')):.1f}"}
# Rows from the APOGEE line-of-sight velocities alone, where measured (value ± error)
for pop, r in s[s.vrot4_apogee.notna()].iterrows():
    e = lambda c, d: f"{r[f'{c}_apogee']:.{d}f} ± {r[f'{c}_apogee_err']:.{d}f}"
    rows['APOGEE ' + pop] = {'N': int(r.n_apogee), 'v_rot(4 kpc) [km/s]': e('vrot4', 1), 'σ_R [km/s]': e('sigma_R', 1),
                             'σ_φ [km/s]': e('sigma_phi', 1), 'σ_z [km/s]': e('sigma_z', 1), 'γ [1/kpc]': '',
                             'v_rot/σ_R': e('ratio_R', 2), 'v_rot/σ_z': e('ratio_z', 2), 'k': '', 'f [%]': ''}
pd.DataFrame(rows).T
````

````python
import matplotlib

matplotlib.use('Agg')
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from fit_kinematics_mcmc import OUTER_INNER, ring
from lmcold import kinfit, samples
from lmcold.layout import panel_grid
from lmcold.sources import DATA, ROOT

# Columns passed to kinfit.data: position [deg], proper motion and errors [mas/yr], correlation, velocity and error [km/s]
COLS = ('ra', 'dec', 'pmra', 'pmdec', 'pmra_error', 'pmdec_error', 'pmra_pmdec_corr', 'vlos', 'evlos')
# Parameters free in the APOGEE-alone fit (module docstring): v0 [km/s], the three dispersions [km/s], k, outlier fraction
OWN = ('v0', 'sigma_R', 'sigma_phi', 'sigma_z', 'k', 'logit_fout')
APOGEE_FROM = -1.1  # [Fe/H]: bins with this lower edge or above get the APOGEE reference
# Quantities summarised per population: v_rot(4 kpc), the dispersions [km/s] and the two ratios
QUANT = ('vrot4', 'sigma_R', 'sigma_phi', 'sigma_z', 'ratio_R', 'ratio_z')
# The three dispersions (keys used to name the APOGEE columns) and their labels
SIG_COLOURS = dict(sigma_R='C0', sigma_phi='C1', sigma_z='C2')
SIG_LABELS = dict(sigma_R=r'$\sigma_R$', sigma_phi=r'$\sigma_\phi$', sigma_z=r'$\sigma_z$')


def draws(run, pop):
    """Paired stage-1 draws of population pop in run: dict of QUANT arrays, and the posterior-median parameter vector."""
    # Draw arrays keyed 'stage|population index|parameter'; i: index of pop in the run's population list
    z = np.load(DATA / f'kin_mcmc_data_{run}_samples.npz')
    i = list(z['names']).index(pop)
    g = lambda n: z[f'1|{i}|{n}_0'].reshape(-1)  # noqa: E731  population's own parameter
    # Per draw: v_rot(4 kpc) = v0 tanh(R_SIG / R0) [km/s], the dispersions [km/s] and the ratios
    d = dict(vrot4=g('v0') * np.tanh(kinfit.R_SIG / g('R0')), sigma_R=g('sigma_R'), sigma_phi=g('sigma_phi'), sigma_z=g('sigma_z'))
    d |= dict(ratio_R=d['vrot4'] / d['sigma_R'], ratio_z=d['vrot4'] / d['sigma_z'])
    # Parameter vector: start values, overwritten by the median of every stage-1 parameter of pop (suffix '_0' dropped)
    p = kinfit.start()
    for key in z.files:
        if key.startswith(f'1|{i}|'):
            n = key.split('|')[2]
            p[kinfit.PARAMS.index(n[:-2] if n.endswith('_0') else n)] = np.median(z[key])
    return d, p


def summary(d):
    """Median and 16th, 84th percentiles of each draw array, as flat columns."""
    out = {}
    for n, v in d.items():
        out |= dict(zip((n, f'{n}_p16', f'{n}_p84'), np.percentile(v, [50, 16, 84])))
    return out
````

### Load the data
The giant catalogues of Zenodo bundle 1 through `lmcold.samples.giants()`, merged and split into the five [Fe/H] bins; the RR Lyrae
members through `lmcold.samples.rrab()` and, for their photometric [Fe/H], `data/rrab.parquet` directly. The posterior draws are read
where they are used, by `draws()`: `data/kin_mcmc_data_ellipsoid_samples.npz` (giants) and
`data/kin_mcmc_data_ellipsoid_outer_samples.npz` (RR Lyrae).

````python
giants = samples.giants()  # (label, catalogue dict) per [Fe/H] bin, from data/giants_apogee.parquet and giants_boss.parquet
rrab = samples.rrab()  # RR Lyrae members with Gaia astrometry (data/rrab.parquet)
rr = pd.read_parquet(DATA / 'rrab.parquet')  # the full RR Lyrae table, with the photometric [Fe/H] column feh
````

### The giant bins
For each [Fe/H] bin: the stars at $2<\rho<9^\circ$, the medians and 16th-84th percentiles of the stage-1 posterior draws of the
velocity-ellipsoid model, and the [Fe/H] spread of the stars. From [Fe/H] = `APOGEE_FROM` up, the APOGEE giants of the bin are also fitted
alone by maximum likelihood ($v_0$, the three dispersions, $k$ and the outlier fraction free; the rest held at the bin's posterior
medians), with errors from the inverse Hessian.

````python
rows = []
for (lo, _), (label, c) in zip(samples.FEH_BINS, giants):
    # Stars at 2 < rho < 9 deg; draws and the median parameter vector of the bin's stage-1 fit
    c = ring(c)
    pop = f'giants {label}'
    d, p = draws('ellipsoid', pop)
    # Row: number of stars, median and 16th-84th percentiles of [Fe/H] [dex], posterior summaries
    r = dict(population=pop, n=len(c['ra']), feh_med=np.median(c['feh']), feh_p16=np.percentile(c['feh'], 16),
             feh_p84=np.percentile(c['feh'], 84)) | summary(d)
    if lo >= APOGEE_FROM:
        # APOGEE giants of the bin; maximum likelihood of OWN, the rest held at p; v, e: values and errors
        a = c['survey'] == 'apogee'
        pa, ea, _ = kinfit.fit(kinfit.data(*(c[k][a] for k in COLS)), free=OWN, p0=p, ellipsoid=True)
        v = {n: pa[kinfit.PARAMS.index(n)] for n in OWN}
        e = {n: ea[kinfit.PARAMS.index(n)] for n in OWN}
        # tanh(R_SIG / R0) with R0 held, to turn v0 into v_rot(4 kpc) [km/s]
        t4 = np.tanh(kinfit.R_SIG / p[kinfit.PARAMS.index('R0')])
        r |= dict(n_apogee=int(a.sum()), feh_med_apogee=np.median(c['feh'][a]), vrot4_apogee=v['v0'] * t4, vrot4_apogee_err=e['v0'] * t4)
        r |= {f'{n}_apogee': v[n] for n in SIG_COLOURS} | {f'{n}_apogee_err': e[n] for n in SIG_COLOURS}
        # Ratios v_rot/sigma with relative errors added in quadrature
        for q, s in (('ratio_R', 'sigma_R'), ('ratio_z', 'sigma_z')):
            r[f'{q}_apogee'] = v['v0'] * t4 / v[s]
            r[f'{q}_apogee_err'] = r[f'{q}_apogee'] * np.hypot(e['v0'] / v['v0'], e[s] / v[s])
    rows.append(r)
````

### The RR Lyrae and the table

## Metallicity errors and the two surveys

A star's bin is set by its measured [Fe/H], and BOSS measurements scatter by about 0.09 dex about APOGEE's (robust
standard deviation on stars observed by both and not used in the calibration of the BOSS scale). The bins above
-1.1 are 0.3 dex wide or open-ended and lie near the peak of the metallicity distribution, so this scatter moves many stars
between neighbouring bins: of the BOSS giants with effective temperature $T_{\rm eff}$ < 4300 K measured above
-0.5 and also observed by APOGEE, 67 per cent lie between -0.8 and
-0.5 on the APOGEE scale. Because the
dispersions change steeply with [Fe/H] at the metal-rich end, the mixing raises the dispersions of the metal-rich bins. Fitted alone,
the APOGEE giants give $\sigma_z$ = 27.3, 18.4 and 10.4 km/s and $\sigma_R$ =
33.6, 23.2 and 19.8 km/s in the three bins above -1.1, against
27.6, 21.0 and 15.8 km/s and 34.1, 25.4 and
20.9 km/s for all giants (Table 7.1); the largest change is the vertical dispersion of the most metal-rich bin.

Two checks support this reading (Figure 7.2). They use the sky-aligned dispersion model of chapter 6: instead
of the disc-frame ellipsoid, each group has one dispersion $\sigma_{\rm los}$ of the line-of-sight velocities and one, $\sigma_{\rm PM}$
(also written $\sigma$), of the proper motions, both constant over the disc, with a pure $\tanh$ rotation curve of level $v_0$. The
BOSS velocities are not at fault: on the 1 926 stars observed by both surveys, the velocity difference divided by its
quoted error has a robust standard deviation of 0.80–0.99 in every bin of signal-to-noise ratio, *Gaia*
$G$ magnitude and $T_{\rm eff}$ (`scripts/fig_boss_velocity_errors.py`), where 1 means correct errors. And mixing the APOGEE dispersions of the bins in the proportions in
which the common stars of each BOSS bin fall into the APOGEE bins,

$$
\sigma_{\rm pred}^2=\sum_tP(t\,|\,j)\,\sigma_{{\rm APOGEE},t}^2,
$$

where $P(t\,|\,j)$ is the fraction of the common stars measured by BOSS in bin $j$ whose APOGEE [Fe/H] lies in bin $t$, predicts the
dispersions of the cool BOSS giants in the sky cells shared with APOGEE (cells of the HEALPix grid,
[Górski et al. 2005](https://doi.org/10.1086/427976), about 1.8° across, holding at least
5 APOGEE stars of the bin): 18.3, 20.1 and 24.3 km/s against the
measured 18.4, 20.8 and 26.2 km/s, from metal-rich to metal-poor. The prediction
neglects the differences in mean velocity between the bins and uses overlap stars brighter than the BOSS sample, which may explain the
2 km/s shortfall in the third bin. The hot BOSS giants ($T_{\rm eff}$ ≥ 4300 K), whose [Fe/H] BOSSNet overestimates, are no hotter
kinematically than the cool ones above -0.5 (17.8 against 18.4 km/s).

We therefore quote all giants in their measured bins as the default sample, which covers the whole metallicity range with one
selection, and the APOGEE giants alone as the reference above [Fe/H] = -1.1, where their [Fe/H] places nearly every star
in its true bin. With them the rise of rotational support is steeper: $v_{\rm rot}/\sigma_R$ = 1.90,
3.19 and 4.05 and $v_{\rm rot}/\sigma_z$ = 2.34, 4.03 and
7.71 (Figure 7.1, dashed). Below -1.1 the APOGEE samples are too small for
this test (43 stars below -1.5 in the shared cells); there the bins are wider and the dispersions
change slowly with [Fe/H], so mixing changes little. We also tried to deconvolve the metallicity errors with a mixture model in which
each BOSS star belongs to each true bin with a probability set by the response of the BOSS [Fe/H] to the true one. It recovered the
kinematics of mock data, but on the real data the dispersions of the middle bins moved with the details of the response model, so we do not use it.

````python
import healpy as hp
import matplotlib

matplotlib.use('Agg')
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from fit_kinematics_mcmc import ring
from lmcold import kinfit, samples
from lmcold.layout import panel_grid
from lmcold.sources import ROOT, path

# Columns passed to lmcold.kinfit.data: degrees, mas/yr, km/s.
COLS = ('ra', 'dec', 'pmra', 'pmdec', 'pmra_error', 'pmdec_error', 'pmra_pmdec_corr', 'vlos', 'evlos')
FREE = kinfit.BASE + ('vlos0', 'sigma_los')  # proper-motion model plus systemic velocity and line-of-sight dispersion
OWN = ('v0', 'sigma', 'sigma_los', 'k', 'logit_fout')  # refitted per subset; the rest held at the bin's L2 fit
T_HOT = 4300.  # K: BOSS [Fe/H] unbiased below, biased high above (paper App. B)
NSIDE = 32  # HEALPix cells of ~1.8 deg
MIN_APOGEE = 5  # APOGEE stars of the bin needed for a cell to count as shared
EDGES = [e for e, _ in samples.FEH_BINS] + [samples.FEH_BINS[-1][1]]   # dex, the six edges of the five [Fe/H] bins


def sub(c, keep):
    """Catalogue dict c restricted to the boolean mask keep."""
    return {k: v[keep] for k, v in c.items()}


def l2(c):
    """Maximum-likelihood L2 fit (symmetric disc, then + STREAM) of catalogue dict c."""
    d = kinfit.data(*(c[k] for k in COLS))
    # Symmetric disc with the centre (dx, dy, the first two of FREE) fixed, then free, then + STREAM; returns the parameter vector.
    p, _, _ = kinfit.fit(d, free=FREE[2:])
    p, _, _ = kinfit.fit(d, free=FREE, p0=p)
    return kinfit.fit(d, free=FREE + kinfit.STREAM, p0=p)[0]


def own(c, p):
    """v0, sigma, sigma_los, k, outliers of catalogue dict c refitted with the rest held at p: (p, err)."""
    p, e, _ = kinfit.fit(kinfit.data(*(c[k] for k in COLS)), free=OWN, p0=p)
    return p, e
````

### Load the data
The giant catalogues of Zenodo bundle 1: through `lmcold.samples.giants()`, merged and split into the five [Fe/H] bins, and, for
the stars observed by both surveys, the APOGEE and BOSS catalogues (`data/giants_apogee.parquet`, `data/giants_boss.parquet`) joined
on the Gaia source_id.

````python
giants = samples.giants()  # (label, catalogue dict) per [Fe/H] bin, from data/giants_apogee.parquet and giants_boss.parquet
# Stars in both surveys: APOGEE columns with suffix _a, BOSS columns (calibrated [Fe/H]) with suffix _b.
m = pd.read_parquet(path('release_giants_apogee')).merge(pd.read_parquet(path('release_giants_boss')), on='source_id', suffixes=('_a', '_b'))
````

### Fits of the survey subsets
For each [Fe/H] bin: the Level-2 maximum-likelihood fit of the whole bin, then $v_0$, $\sigma$, $\sigma_{\rm los}$, $k$ and the
outlier fraction refitted for APOGEE, BOSS cool ($T_{\rm eff}$ < `T_HOT`) and BOSS hot giants in the HEALPix cells shared with
APOGEE, and for the two BOSS subsets on the whole ring.

````python
rows = []   # one dict per (bin, subset)
for label, c in giants:
    c = ring(c)   # 2 < rho < 9 deg
    p = l2(c)   # L2 fit of the whole bin: the held parameters
    pix = hp.ang2pix(NSIDE, c['ra'], c['dec'], lonlat=True)   # HEALPix cell (ring ordering) of each star
    apo = c['survey'] == 'apogee'
    # Shared cells: those with >= MIN_APOGEE APOGEE stars of the bin.
    cells, counts = np.unique(pix[apo], return_counts=True)
    shared = np.isin(pix, cells[counts >= MIN_APOGEE])
    # Masks of the five subsets: three in the shared cells, two BOSS subsets on the whole ring.
    subsets = {'APOGEE': apo & shared, 'BOSS cool': ~apo & (c['teff'] < T_HOT) & shared, 'BOSS hot': ~apo & (c['teff'] >= T_HOT) & shared,
               'BOSS cool, ring': ~apo & (c['teff'] < T_HOT), 'BOSS hot, ring': ~apo & (c['teff'] >= T_HOT)}
    for name, keep in subsets.items():
        q, e = own(sub(c, keep), p)
        # Subset size, the bin's median [Fe/H] [dex] and hot fraction of its BOSS stars, refitted parameters and errors.
        r = dict(bin=label, subset=name, n=int(keep.sum()), feh_med=np.median(c['feh']), hot_frac=np.mean(c['teff'][~apo] >= T_HOT))
        r |= {n: q[kinfit.PARAMS.index(n)] for n in OWN} | {f'{n}_err': e[kinfit.PARAMS.index(n)] for n in OWN}
        rows.append(r)
        print(f"{label:22s} {name:16s} N {r['n']:6d}  v0 {r['v0']:6.1f} +- {r['v0_err']:4.1f}  sigma_los {r['sigma_los']:5.1f} +- "
              f"{r['sigma_los_err']:4.1f}", flush=True)
t = pd.DataFrame(rows)   # one row per (bin, subset)
````

### Leakage and forward prediction
On the stars observed by both surveys: the fraction of each measured (BOSS) bin whose APOGEE [Fe/H] lies in another bin, and the
BOSS dispersion and rotation predicted by mixing the APOGEE values in those proportions,
$\sigma_{\rm pred}^2=\sum_tP(t\,|\,j)\,\sigma_{{\rm APOGEE},t}^2$.

````python
# Leakage from the [Fe/H] errors, on the stars in both surveys (BOSS calibrated value vs APOGEE).
# mb: measured (BOSS) bin, tb: 'true' (APOGEE) bin of each star (-1 or 5: outside the bins).
mb, tb = np.digitize(m.feh_b, EDGES) - 1, np.digitize(m.feh_a, EDGES) - 1
leak = []   # one dict per (bin, class)
for j, (label, _) in enumerate(zip(t.bin.unique(), samples.FEH_BINS)):
    # Class by BOSS Teff: cool below T_HOT, hot at or above.
    for cls, k in (('cool', m.teff_b < T_HOT), ('hot', m.teff_b >= T_HOT)):
        n = int((k & (mb == j)).sum())   # overlap stars of the class measured in bin j
        # Fraction whose APOGEE [Fe/H] is in another bin, with its binomial error (max(n, 1) avoids division by zero).
        f = ((tb != j) & k & (mb == j)).sum() / max(n, 1)
        leak.append(dict(bin=label, cls=cls, n=n, frac=f, err=np.sqrt(f * (1 - f) / max(n, 1))))
lk = pd.DataFrame(leak)
# Forward prediction of the BOSS values from APOGEE's (see the module docstring).
apo = t[t.subset == 'APOGEE']   # APOGEE values per bin (shared cells), in bin order
pred = []
for cls, k in (('cool', m.teff_b < T_HOT), ('hot', m.teff_b >= T_HOT)):
    for j, label in enumerate(apo.bin):
        # P(t | j): fractions of the class's overlap stars of measured bin j in each APOGEE bin t; then the mixed sigma_los, v0.
        f = np.bincount(tb[k & (mb == j)], minlength=len(apo)) / max((k & (mb == j)).sum(), 1)
        pred.append(dict(bin=label, cls=cls, sigma_los_pred=np.sqrt(f @ apo.sigma_los.to_numpy() ** 2), v0_pred=f @ apo.v0.to_numpy()))
pr = pd.DataFrame(pred)   # one row per (bin, class): predicted sigma_los, v0 [km/s]
pr.to_csv(ROOT / 'results/kin_feh_forward.csv', index=False, float_format='%.5f')
# Print the predictions next to the measured BOSS cool and hot values ('BOSS cool'[5:] = 'cool').
print(pr.merge(t[t.subset.isin(['BOSS cool', 'BOSS hot'])].assign(cls=lambda x: x.subset.str[5:])[['bin', 'cls', 'sigma_los', 'v0']],
               on=['bin', 'cls']).to_string(index=False, float_format='%.2f'))
t.to_csv(ROOT / 'results/kin_feh_contamination.csv', index=False, float_format='%.5f')
lk.to_csv(ROOT / 'results/kin_feh_leakage.csv', index=False, float_format='%.5f')
print(lk.to_string(index=False, float_format='%.2f'))
````

### The BOSS velocity errors

The test of the BOSS line-of-sight velocity errors quoted above uses the 1 926 giants of both release catalogues. For
each star the difference between the BOSS and APOGEE velocities is divided by the two quoted errors added in quadrature; if the BOSS

errors are much smaller. The script repeats the test in bins of BOSS signal-to-noise ratio, *Gaia* $G$ magnitude and BOSS
$T_{\rm eff}$ and draws a diagnostic figure; the numbers on this page are recomputed from the same stars and bins by
`scripts/paper_text_kin_support.py`.

## Centre and geometry

The kinematic centre of the joint (stage-2) fit lies 0.47° east of the vdMK14 centre, at right ascension and
declination $(\alpha,\delta)$ = (81.23°, -69.61°); the Level-1 joint fit puts it at
0.72° east and 0.13° south, and the giant bins fitted on their own lie
0.32–0.54° east and up to 0.30° south, the metal-poor ones furthest south
(Figure 7.4). The RR Lyrae fitted at 4 < $\rho$ < 9° put the centre at
0.93 ± 0.16° east, with no stars within 4° to pin it down; fitted at 2 < $\rho$ <
9° they give 0.45°, close to the giants. The joint fit with sky-aligned dispersions puts the
centre 0.06° further east (0.53° east and 0.02° south, with $i$ =
28.0°, $\theta$ = 142.7° and the $m=2$ coefficient of the azimuthal velocity $m_{2c}$ =
-0.12). The density centres of the RR Lyrae and of the *Gaia* red giants, iterated means within apertures of
1.5, 2.5, 3.5 and 5° radius, lie 0.10–0.23° east. The kinematic centre is thus
0.25–0.42° from the density centres, 0.22–0.37
kpc at the LMC distance; published proper-motion centres from *Gaia* ([Gaia Collaboration, Luri et al.
2021](https://doi.org/10.1051/0004-6361/202039588)) and from *Gaia* combined with the VISTA survey of the Magellanic Clouds (VMC; e.g.
[Choi et al. 2022](https://doi.org/10.3847/1538-4357/ac4e90)) lie 0.0–0.4° east and
0.2–0.3° north. We read this as the centre of a disc model fitted to a perturbed velocity field,
not as the position of the potential minimum: it moves by 0.25° east between the Level-1 and Level-2 joint fits and by up to 0.18° with the
sky coverage of a sample (chapter 6), while no
symmetric mock fitted with proper motions and line-of-sight velocities produces an offset larger than 0.1°.

The orientation of the joint fit, $i$ = 28.0° and $\theta$ = 143.9°, is likewise a parameter
of the model. Proper motions alone give $i$ = 31–36° (maximum-likelihood fits of the giant
bins with the sky-aligned model), as in earlier proper-motion studies (Gaia Collaboration, Luri et al. 2021; Choi et al. 2022), and
adding the line-of-sight velocities lowers it to 25–28°; the RR Lyrae fitted with their
distances as well (from the relation between period and Wesenheit magnitude, a dust-insensitive combination of their OGLE-IV $V$ and $I$
magnitudes) lie in a plane with $i\approx$ 19° and $\theta\approx$ 158°, the plane of
their positions rather than of their motions.

Figure 7.3 shows how the orientation and the shape of the velocity ellipsoid change with metallicity when each group
is fitted on its own, and compares them with the joint fit. The line of nodes is the same for all giant bins, $\theta$ =
143.5–146.0° against 143.9 ± 0.2° in the joint
fit. The inclination is not: it rises from 24.6 ± 1.1° and 25.2 ± 0.8° in the two metal-poor bins to
27.4–28.3° above [Fe/H] = -1.1, where the joint value
(28.0 ± 0.1°) is set by the most numerous stars; the metal-poor bins lie
2.9–3.4 standard deviations below it. Whether this is a thicker or more flared metal-poor disc seen
in projection, or a different plane, our thin-disc model cannot tell from these data alone; a mock as thick as the old population is
recovered at its true inclination (chapter 8). The RR Lyrae, with proper motions only, constrain their orientation weakly ($i$ =
22.3 (+4.5/−4.3)°, $\theta$ = 119 (+12/−9)°). The shape of the ellipsoid does not depend on the geometry:
$\sigma_\phi/\sigma_R$ and $\sigma_z/\sigma_R$ of each giant bin agree between the two stages to within 0.013, well
inside their errors (half-widths of the 16th–84th percentile ranges 0.007–0.027). For the RR
Lyrae $\sigma_\phi/\sigma_R$ = 0.63 (+0.08/−0.10) on their own and 0.66 (+0.07/−0.08) with the shared geometry of the
joint fit at 4 < $\rho$ < 9°, lower than the 0.77–0.84
of the giants, but correlated with $k$ and with the unconstrained $\sigma_z$.

````python
rows = []
for (lo, _), (label, c) in zip(samples.FEH_BINS, giants):
    # The bin's giants at 2 < rho < 9 deg; counts, and the median and 16th-84th percentiles of their [Fe/H] [dex].
    c = ring(c)
    pop = f'giants {label}'
    r = dict(population=pop, n=len(c['ra']), n_los=int(np.isfinite(c['vlos']).sum()), feh_med=np.median(c['feh']),
             feh_p16=np.percentile(c['feh'], 16), feh_p84=np.percentile(c['feh'], 84))
    # v0 and sigma_los [km/s] (2-9 deg run) and sigma (4-9 deg run), each with p16 and p84.
    for n in ('v0', 'sigma_los'):
        r |= {n: q(pop, n), f'{n}_p16': q(pop, n, 'p16'), f'{n}_p84': q(pop, n, 'p84')}
    r |= pm(pop)
    if lo >= APOGEE_FROM:
        # Start from the bin's posterior medians wherever the summary has the parameter, kinfit.start() elsewhere.
        p = kinfit.start()
        for n in kinfit.PARAMS:
            if ((post.population == pop) & (post.param == n)).any():
                p[kinfit.PARAMS.index(n)] = q(pop, n)
        # ML fit of OWN on the APOGEE giants of the bin; their values and inverse-Hessian errors stored as *_apogee, *_apogee_err.
        a = c['survey'] == 'apogee'
        pa, ea, _ = kinfit.fit(kinfit.data(*(c[k][a] for k in COLS)), free=OWN, p0=p)
        r |= dict(n_apogee=int(a.sum()), feh_med_apogee=np.median(c['feh'][a]))
        r |= {f'{n}_apogee': pa[kinfit.PARAMS.index(n)] for n in ('v0', 'sigma_los')}
        r |= {f'{n}_apogee_err': ea[kinfit.PARAMS.index(n)] for n in ('v0', 'sigma_los')}
    rows.append(r)
# RR Lyrae members: count at 2-9 deg, photometric [Fe/H] spread, v0 (stage 1) and the 4-9 deg sigma; no line-of-sight data.
rr = rr[rr.member]
rows.append(dict(population='RR Lyrae', n=len(ring(rrab)['ra']), n_los=0, feh_med=float(rr.feh.median()),
                 feh_p16=float(rr.feh.quantile(0.16)), feh_p84=float(rr.feh.quantile(0.84)),
                 **{f'v0{s}': q('RR Lyrae', 'v0', c) for s, c in (('', 'median'), ('_p16', 'p16'), ('_p84', 'p84'))}, **pm('RR Lyrae')))
````

fit; the APOGEE ratios with propagated errors. The table goes to `results/kin_support.csv`.

````python
t = pd.DataFrame(rows)
# v0 / sigma_los (2-9 deg run) and v0 / sigma (outer run) draw by draw from the stage-1 posterior samples (keys '1|<i>|<param>_0',
# populations in the run's order).
for col, n, run in (('ratio', 'sigma_los', 'stream'), ('ratio_pm', 'sigma', 'stream_outer')):
    draws = np.load(DATA / f'kin_mcmc_data_{run}_samples.npz')
    for i, pop in enumerate(draws['names']):
        if pop in set(t.population) and f'1|{i}|{n}_0' in draws:
            r = (draws[f'1|{i}|v0_0'] / draws[f'1|{i}|{n}_0']).reshape(-1)
            t.loc[t.population == pop, [col, f'{col}_lo', f'{col}_hi']] = np.percentile(r, [50, 16, 84])
# APOGEE ratio v0 / sigma_los with the relative errors added in quadrature.
t['ratio_apogee'] = t.v0_apogee / t.sigma_los_apogee
t['ratio_apogee_err'] = t.ratio_apogee * np.hypot(t.v0_apogee_err / t.v0_apogee, t.sigma_los_apogee_err / t.sigma_los_apogee)
t.to_csv(ROOT / 'results/kin_support.csv', index=False, float_format='%.4f')
print(t[['population', 'n', 'feh_med', 'feh_p16', 'feh_p84', 'v0', 'sigma', 'sigma_los', 'ratio', 'ratio_pm', 'v0_apogee', 'sigma_los_apogee', 'ratio_apogee']].round(2).to_string(index=False))
````

The $m=2$ terms are not the only way to describe the residual pattern of the velocity field (chapter 6). We refitted each giant bin (maximum
likelihood, sky-aligned dispersions) with three alternatives, each recovered from mocks in which it is the only perturbation
(Figure 7.6): (i) an inner disc ($\rho$ < 4.5°) in its own plane, inclined
3–5° less than the outer disc, which improves the fit of the two metal-rich bins
($\Delta\ln L$ = 126 and 75; ≤ 11 in the others) and changes $v_0$
by ≤ 0.5 km/s; (ii) a different scale of the internal motions in the line-of-sight velocities and in the proper
motions, the line-of-sight deviations from the mean motion being multiplied by $(1+q_{\rm los})$, with $q_{\rm los}$ =
-0.34 and -0.45 and $\Delta\ln L$ = 77 and 80 in the two metal-rich
bins ($|q_{\rm los}|$ ≤ 0.2, $\Delta\ln L$ ≤ 6 in the others), at the price of $i$ =
38–42°; (iii) a disc whose orientation changes with time
([van der Marel et al. 2002](https://doi.org/10.1086/343775)), which alone fits worse than Level 2 with sky-aligned dispersions
($\Delta\ln L$ = -33 to -472) and combined with it fits better ($\Delta\ln L$ =
9 to 322) but requires rates of change of the line of nodes of
+196 to +587° Gyr$^{-1}$: a turn of 60–180° in one rotation
period at 4 kpc (0.3 Gyr, $2\pi R/v_{\rm rot}$ for the fastest-rotating giants), which we do not
consider a physical description of the disc. $\Delta\ln L$ is measured against Level 2 of the same bin; positive values favour the
alternative. Part of what model (ii) absorbs, a different amplitude of the internal motions along the line of sight and on the sky, is
what the velocity ellipsoid describes physically. The line-of-sight dispersions are the same in every model to
0.2 km/s. The rotation amplitude is not: it changes by up to 13 km/s between Level 2 and the
alternatives (i)–(ii) (the metal-rich bin in model ii) and by 15–56 km/s in (iii) with
streaming. The ordering by rotational support is robust: $v_0/\sigma_{\rm los}$ increases monotonically from the metal-poor to the
metal-rich bins in every model except the turning disc alone, which fits worst and in which only the two most metal-poor bins swap
(1.79 and 1.72); in the velocity-ellipsoid model both $v_{\rm rot}/\sigma_R$ and
$v_{\rm rot}/\sigma_z$ increase monotonically (Table 7.1). Absolute rotation speeds are conditional on the model.

With the velocity ellipsoid we repeated tests (i) and (ii) and tested the single error scale $k$ (maximum likelihood, each giant bin on
its own). The inner disc still improves the fit of the two metal-rich bins ($\Delta\ln L$ = 89 and
71; 0.4–9 in the others), inclined
3–4° less than the outer disc, and changes $v_{\rm rot}(4\,{\rm kpc})$ by at most
0.2 km/s and the dispersions by at most 0.1 km/s. The line-of-sight scale is no longer needed in
the bin from -0.8 to -0.5 ($q_{\rm los}$ = -0.02, $\Delta\ln L$ = 0.1,
against 77 with sky-aligned dispersions): the ellipsoid describes the different amplitude of the motions along the
line of sight and on the sky. In the most metal-rich bin it still gives $\Delta\ln L$ = 21 with $q_{\rm los}$ =
-0.29, again at the price of $i$ = 36°, and in the two metal-poor bins it runs to its bound,
+0.5, for $\Delta\ln L$ = 2.5–8.1; in every bin the dispersions change by
at most 0.9 km/s and $v_{\rm rot}(4\,{\rm kpc})$ by up to 4 km/s. Fitting the stars brighter and
fainter than the median *Gaia* $G$ magnitude of each bin separately, each half with its own $k$, rotation curve and ellipsoid and the
geometry and streaming held, gives $k$ = 1.30–2.08 for the bright halves and
1.22–1.73 for the faint ones: *Gaia* underestimates the errors of the brighter stars more, as a
constant error floor would. The dispersions of the two halves differ by at most 2.3 km/s in the three bins below
[Fe/H] = -0.8, and are 2–3 km/s higher for the faint halves in the
two metal-rich bins, in $\sigma_z$ as much as in $\sigma_R$ and $\sigma_\phi$. Because $\sigma_z$ rests mostly on the line-of-sight
velocities, which do not depend on $k$, that difference comes from the stars, plausibly the stronger [Fe/H] mixing of the fainter BOSS

#### Helpers
`fits` fits the four models to one sample, `mock_of` draws a mock catalogue from a fitted parameter vector at the positions of
the real stars, and `add` stores the fitted rows.

```python
def fits(c, free):
    """Fit the four models of MODELS to one sample.

    c : dict of ndarray, the catalogue (columns COLS); free : tuple of str, the parameters free in 'sym' (kinfit.PARAMS names).
    Returns a list of (model, nll, p, err): -lnL at the optimum, and the parameter vector and its inverse-Hessian errors (each of
    length len(kinfit.PARAMS); angles in deg, velocities in km/s). Prints one line per model.
    """
    # Likelihood inputs of the sample (kinfit.data arrays)
    d = kinfit.data(*(c[k] for k in COLS))
    p, _, _ = kinfit.fit(d, free=free[2:])  # centre fixed at the adopted position first
    # Then the centre (dx, dy, the first two of free) freed too; kinfit.fit returns (p, err, nll)
    sym = kinfit.fit(d, free=free, p0=p)
    out = [('sym', sym[2], sym[0], sym[1])]
    # The other three models, each started from 'sym'; 'tilt' models tried from every DINC_STARTS, best -lnL kept
    for name, extra in list(MODELS.items())[1:]:
        best = None
        for dinc in (DINC_STARTS if 'tilt' in name else (0.0,)):
            p0 = sym[0].copy()
            p0[kinfit.PARAMS.index('dinc')] = dinc
            r = kinfit.fit(d, free=free + extra, p0=p0)
            best = r if best is None or r[2] < best[2] else best
        out.append((name, best[2], best[0], best[1]))
    for name, nll, q, _ in out:  # q indices: kinfit.PARAMS (0 dx, 1 dy, 4 inc, 5 theta, 17 m2c, 18 m2s, 22 dinc, 23 dtheta)
        print(f'  {name:14s} -dlnL {sym[2] - nll:8.1f}  dx {q[0]:.3f} dy {q[1]:.3f} i {q[4]:.1f} theta {q[5]:.1f} '
              f'dinc {q[22]:.1f} dtheta {q[23]:.1f} |m2| {np.hypot(q[17], q[18]):.3f}', flush=True)
    return out


def mock_of(c, q):
    """Mock of catalogue c: proper motions and velocities drawn from parameter vector q, with k = 1 and no outliers.

    c : dict of ndarray (columns COLS); q : ndarray, kinfit.PARAMS vector (the fitted model to draw from).
    Returns (mock catalogue dict, the parameter vector actually used: q with k = 1 and logit_fout = -8, the mock's truth).
    """
    q = q.copy()
    q[kinfit.PARAMS.index('k')], q[kinfit.PARAMS.index('logit_fout')] = 1.0, -8.0  # Gaia errors as quoted; outlier fraction ~3e-4
    # Likelihood inputs at the real stars (positions, errors)
    d = kinfit.data(*(c[k] for k in COLS))
    # Model mean proper motions [mas/yr], line-of-sight velocity [km/s] and disc distance D [kpc] of each star.
    mra, mde, mlos, D = (np.asarray(v) for v in kinfit.model(q, d['rhat'], d['ea'], d['ed']))
    # Fixed seed: the same mock on every run
    rng = np.random.default_rng(5)
    # Proper motions: Gaia covariance plus sigma (q[8], km/s) as a proper motion at D; velocities: e_vlos and sigma_los (q[12]).
    cov = np.asarray(d['cov']) + ((q[8] / (KAPPA * D)) ** 2)[:, None, None] * np.eye(2)
    # Correlated proper-motion noise (N, 2) [mas/yr] via the Cholesky factor of each star's covariance
    noise = np.einsum('nij,nj->ni', np.linalg.cholesky(cov), rng.normal(size=(len(D), 2)))
    # Velocities [km/s] drawn only where the real star has one
    vlos = np.where(np.isfinite(c['vlos']), mlos + rng.normal(size=len(D)) * np.hypot(c['evlos'], q[12]), np.nan)
    return dict(c, pmra=mra + noise[:, 0], pmdec=mde + noise[:, 1], vlos=vlos), q
```

#### Helpers
`fits` fits the four models to one data set, `mock_of` draws a mock catalogue from a fitted parameter vector at the positions of
the real stars, and `add` stores the fitted rows.

```python
def fits(c):
    """Fit the four models of MODELS to one data set (proper motions and line-of-sight velocities).

    c : dict of ndarray, the catalogue (columns COLS).
    Returns a list of (model, nll, p, err): -lnL at the optimum, and the parameter vector and its inverse-Hessian errors (each of
    length len(kinfit.PARAMS); angles in deg, velocities in km/s). Prints one line per model.
    """
    # the arrays kinfit needs (unit vectors, Gaia covariances, velocities)
    d = kinfit.data(*(c[k] for k in COLS))
    p, _, _ = kinfit.fit(d, free=FREE[2:])  # centre fixed at the adopted position first
    # 'sym': all of FREE, started from the fixed-centre fit; sym = (p, err, nll)
    sym = kinfit.fit(d, free=FREE, p0=p)
    out = [('sym', sym[2], sym[0], sym[1])]
    # the other three models, each started from 'sym'
    for name, extra in list(MODELS.items())[1:]:
        p, e, nll = kinfit.fit(d, free=FREE + extra, p0=sym[0])
        out.append((name, nll, p, e))
    for name, nll, q, _ in out:  # q indices: kinfit.PARAMS (0 dx, 1 dy, 4 inc, 5 theta, 6 v0, 17 m2c, 18 m2s, 24 qlos)
        print(f'  {name:14s} -dlnL {sym[2] - nll:8.1f}  dx {q[0]:.3f} dy {q[1]:.3f} i {q[4]:.1f} theta {q[5]:.1f} '
              f'v0 {q[6]:.1f} qlos {q[24]:+.3f} m2c {q[17]:+.3f} m2s {q[18]:+.3f}', flush=True)
    return out


def mock_of(c, q):
    """Mock of catalogue c: proper motions and velocities drawn from parameter vector q, with k = 1 and no outliers.

    c : dict of ndarray (columns COLS); q : ndarray, kinfit.PARAMS vector (the fitted model to draw from).
    Returns (mock catalogue dict, the parameter vector actually used: q with k = 1 and logit_fout = -8, the mock's truth).
    """
    q = q.copy()
    q[kinfit.PARAMS.index('k')], q[kinfit.PARAMS.index('logit_fout')] = 1.0, -8.0  # Gaia errors as quoted; outlier fraction ~3e-4
    # the catalogue's arrays, in the form kinfit.model needs
    d = kinfit.data(*(c[k] for k in COLS))
    # Model mean proper motions [mas/yr], line-of-sight velocity [km/s] and disc distance D [kpc] of each star.
    mra, mde, mlos, D = (np.asarray(v) for v in kinfit.model(q, d['rhat'], d['ea'], d['ed']))
    # fixed seed: the mocks are reproducible
    rng = np.random.default_rng(6)
    # Proper motions: Gaia covariance plus sigma (q[8], km/s) as a proper motion at D; velocities: e_vlos and sigma_los (q[12]).
    cov = np.asarray(d['cov']) + ((q[8] / (KAPPA * D)) ** 2)[:, None, None] * np.eye(2)
    noise = np.einsum('nij,nj->ni', np.linalg.cholesky(cov), rng.normal(size=(len(D), 2)))
    # velocities only where the star has a measured one; e_vlos and sigma_los added in quadrature
    vlos = np.where(np.isfinite(c['vlos']), mlos + rng.normal(size=len(D)) * np.hypot(c['evlos'], q[12]), np.nan)
    return dict(c, pmra=mra + noise[:, 0], pmdec=mde + noise[:, 1], vlos=vlos), q


rows, fitted = [], {}  # output rows; fitted[sample, velocities][model] = parameter vector (the mocks are drawn from these)


def add(name, vel, c, out, truth=None):
    """Append the rows of fits() output `out` for sample `name`, velocities `vel` (truth: the mock's parameter vector)."""
    # keep the fitted vectors: the mocks are drawn from the 'qlos' and 'stream' fits of MOCK_BIN
    fitted[name, vel] = {m: q for m, _, q, _ in out}
    for model, nll, p, e in out:
        # one row per model: counts, -lnL, every parameter and its error (and the truth for mocks)
        rows.append(dict(sample=name, velocities=vel, model=model, n_los=int(np.isfinite(c['vlos']).sum()), nll=nll,
                         **dict(zip(kinfit.PARAMS, p)), **{f'{n}_err': v for n, v in zip(kinfit.PARAMS, e)},
                         **({} if truth is None else {f'truth_{n}': v for n, v in zip(kinfit.PARAMS, truth)})))
```

```python
# model name: parameters freed in addition to FREE (STREAM: m2c, m2s, vr0, r2c, r2s; TURN: didt, dthdt [deg/Gyr]; qlos: the
# line-of-sight deviations from the mean motion scaled by 1 + qlos).
MODELS = {'sym': (), 'stream': kinfit.STREAM, 'turn': kinfit.TURN, 'turn + stream': kinfit.TURN + kinfit.STREAM,
          'qlos + stream': ('qlos',) + kinfit.STREAM, 'turn + qlos + stream': kinfit.TURN + ('qlos',) + kinfit.STREAM}
MOCK_BIN = 3  # giants -0.8 < [Fe/H] < -0.5: the most precise bin (as in scripts/fig_kin_los_scale.py)
```

#### Helpers
`fits` fits the six models to one sample, `mock_of` draws a mock catalogue from a fitted parameter vector at the positions of
the real stars, and `add` stores the fitted rows.

```python
def fits(c):
    """Fit the six models of MODELS to one sample (proper motions and line-of-sight velocities).

    c : dict of ndarray, the catalogue (columns COLS).
    Returns a list of (model, nll, p, err): -lnL at the optimum, and the parameter vector and its inverse-Hessian errors (each of
    length len(kinfit.PARAMS); angles in deg, velocities in km/s, rates in deg/Gyr). Prints one line per model.
    """
    d = kinfit.data(*(c[k] for k in COLS))   # precomputed unit vectors, covariances etc. for the likelihood
    p, _, _ = kinfit.fit(d, free=FREE[2:])  # centre fixed at the adopted position first
    sym = kinfit.fit(d, free=FREE, p0=p)   # (p, err, nll) of 'sym' with the centre (dx, dy) free
    out = [('sym', sym[2], sym[0], sym[1])]
    # Each other model starts from the 'sym' optimum with its extra parameters freed.
    for name, extra in list(MODELS.items())[1:]:
        p, e, nll = kinfit.fit(d, free=FREE + extra, p0=sym[0])
        out.append((name, nll, p, e))
        # p indices: kinfit.PARAMS (0 dx, 4 inc, 5 theta, 6 v0, 17 m2c, 18 m2s, 24 qlos, 25 didt, 26 dthdt)
        print(f'  {name:22s} -dlnL {sym[2] - nll:8.1f}  dx {p[0]:.3f} i {p[4]:.1f} theta {p[5]:.1f} v0 {p[6]:.1f} '
              f'qlos {p[24]:+.3f} |m2| {np.hypot(p[17], p[18]):.3f} didt {p[25]:+.0f} dthdt {p[26]:+.0f}', flush=True)
    return out


def mock_of(c, q):
    """Mock of catalogue c: proper motions and velocities drawn from parameter vector q, with k = 1 and no outliers.

    c : dict of ndarray (columns COLS); q : ndarray, kinfit.PARAMS vector (the fitted model to draw from).
    Returns (mock catalogue dict, the parameter vector actually used: q with k = 1 and logit_fout = -8, the mock's truth).
    """
    q = q.copy()   # do not change the fitted vector stored in `fitted`
    q[kinfit.PARAMS.index('k')], q[kinfit.PARAMS.index('logit_fout')] = 1.0, -8.0  # Gaia errors as quoted; outlier fraction ~3e-4
    d = kinfit.data(*(c[k] for k in COLS))
    # Model mean proper motions [mas/yr], line-of-sight velocity [km/s] and disc distance D [kpc] of each star.
    mra, mde, mlos, D = (np.asarray(v) for v in kinfit.model(q, d['rhat'], d['ea'], d['ed']))
    rng = np.random.default_rng(7)   # fixed seed
    # Proper motions: Gaia covariance plus sigma (q[8], km/s) as a proper motion at D; velocities: e_vlos and sigma_los (q[12]).
    cov = np.asarray(d['cov']) + ((q[8] / (KAPPA * D)) ** 2)[:, None, None] * np.eye(2)
    # (N, 2) correlated noise [mas/yr]: Cholesky factor of each 2 x 2 covariance times two unit normals.
    noise = np.einsum('nij,nj->ni', np.linalg.cholesky(cov), rng.normal(size=(len(D), 2)))
    # Velocities only for stars with a measured one (NaN elsewhere, as in the data).
    vlos = np.where(np.isfinite(c['vlos']), mlos + rng.normal(size=len(D)) * np.hypot(c['evlos'], q[12]), np.nan)
    return dict(c, pmra=mra + noise[:, 0], pmdec=mde + noise[:, 1], vlos=vlos), q


rows, fitted = [], {}  # output rows; fitted[sample][model] = parameter vector (the mocks are drawn from these)


def add(name, c, out, truth=None):
    """Append the rows of fits() output `out` for sample `name` (truth: the mock's parameter vector)."""
    fitted[name] = {m: q for m, _, q, _ in out}
    # One row per model: number of line-of-sight velocities, -lnL, parameters, errors and (mocks) the truth.
    for model, nll, p, e in out:
        rows.append(dict(sample=name, model=model, n_los=int(np.isfinite(c['vlos']).sum()), nll=nll,
                         **dict(zip(kinfit.PARAMS, p)), **{f'{n}_err': v for n, v in zip(kinfit.PARAMS, e)},
                         **({} if truth is None else {f'truth_{n}': v for n, v in zip(kinfit.PARAMS, truth)})))
```

## The inner edge of the sample

The proper-motion dispersion of a group is separated from the *Gaia* errors only through the error scale $k$: for a star with velocity
error $e$ the likelihood constrains mainly $\sigma_{\rm PM}^2+k^2e^2$. The giants have $e\approx$
12–15 km/s and the RR Lyrae $e\approx$ 64 km/s per component
(medians at 2 < $\rho$ < 9°); for the RR Lyrae at 4 < $\rho$ <
9° ($e\approx$ 60 km/s) a change of $k$ by 0.1 moves $\sigma_{\rm PM}$ by
10–20 km/s. For the giants $k$
matters little; for the RR Lyrae it sets the dispersion. *Gaia* underestimates its errors more in crowded fields, so one $k$ need not
describe a whole group. Figure 7.7 shows the maximum-likelihood $\sigma_{\rm PM}$ and $k$ of each group
(sky-aligned Level 2, all parameters free) as the inner edge of the sample moves from 2 to
6°. For the giants $\sigma_{\rm PM}$ changes by 2.1–3.1 km/s,
partly a real decline with radius. For the RR Lyrae $k$ falls from 1.32 to 1.14 and
$\sigma_{\rm PM}$ rises from the 5 km/s lower bound of these fits to 25 km/s at 3° and
28–31 km/s from 4° out. A mock with the positions and errors of
the RR Lyrae, $\sigma_{\rm PM}$ = 30 km/s and $k$ = 1.39 inside 4° and
1.13 beyond, shows the same behaviour: one $k$ fitted from 2° returns $\sigma_{\rm PM}$ =
15 km/s, and from 4° out the fit recovers
26–29 km/s and $k$ = 1.13–1.15. The
velocity-ellipsoid model, whose dispersions may vary with radius, meets the same problem in another form: at 2 <
$\rho$ < 9° it fits the RR Lyrae with a dispersion that falls by 43 per cent per kpc. We
therefore measure the RR Lyrae dispersions at 4 < $\rho$ < 9°.

````python
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from fit_kinematics_mcmc import COLS, ml, ring
from lmcold import kinfit, samples
from lmcold.constants import KAPPA, LMC
from lmcold.footprint import angular_distance
from lmcold.layout import panel_grid
from lmcold.sources import ROOT

INNER = (2., 3., 4., 5., 6.)  # deg, inner edges of the sample
MOCK_SIGMA = 30.0  # km/s   # true proper-motion dispersion of the mock
MOCK_K = (1.39, 1.13)  # Gaia error scale inside and beyond 4 deg (RR Lyrae fits of the data at 2-4 and 4-9 deg)
COLOURS = ['#d7191c', '#fdae61', '#a6d96a', '#1a9641', '#2878b5', '#7b3294']  # giant bins metal-poor first, RR Lyrae


def fits(sample, name, c):
    """Maximum-likelihood rows of catalogue dict c (2 < rho < 9 deg) for every inner edge of INNER."""
    rho = angular_distance(c['ra'], c['dec'], LMC['ra0'], LMC['dec0'])
    # rho: angular distance of each star from the adopted LMC centre [deg]; one row per inner edge.
    out = []
    for inner in INNER:
        d = kinfit.data(*(np.asarray(c[k])[rho > inner] for k in COLS))
        # Full ML fit of the stars beyond `inner`; then errors of v0, sigma, k and the outlier fraction from the inverse Hessian,
        # geometry held.
        p = ml(d, 'stream')
        _, e, _ = kinfit.fit(d, free=('v0', 'sigma', 'k', 'logit_fout'), p0=p)
        g = lambda n, v=p: v[kinfit.PARAMS.index(n)]  # noqa: E731
        out.append(dict(sample=sample, population=name, inner=inner, n=int((rho > inner).sum()), sigma=g('sigma'),
                        sigma_err=g('sigma', e), k=g('k'), k_err=g('k', e), v0=g('v0')))
        print(out[-1], flush=True)
    return out
````

`results/kin_dispersion_cut.csv`.

````python
# Mock RR Lyrae: truth from their own fit beyond 4 deg, with sigma and the position-dependent k replaced.
c = pops[-1][1]
rho = angular_distance(c['ra'], c['dec'], LMC['ra0'], LMC['dec0'])
# c: the RR Lyrae at 2 < rho < 9 deg; q: their ML vector from the stars beyond 4 deg, with sigma set to MOCK_SIGMA [km/s].
q = ml(kinfit.data(*(np.asarray(c[k])[rho > 4.] for k in COLS)), 'stream')
q[kinfit.PARAMS.index('sigma')] = MOCK_SIGMA
d = kinfit.data(*(c[k] for k in COLS))
# Model proper motions [mas/yr] and distances D [kpc] of each RR Lyrae for vector q.
mra, mde, _, D = (np.asarray(v) for v in kinfit.model(q, d['rhat'], d['ea'], d['ed']))
k = np.where(rho < 4., MOCK_K[0], MOCK_K[1])
# Per-star covariance (N, 2, 2) [(mas/yr)^2]: Gaia covariance scaled by k^2 plus the intrinsic dispersion converted to mas/yr at D.
cov = k[:, None, None] ** 2 * np.asarray(d['cov']) + ((MOCK_SIGMA / (KAPPA * D)) ** 2)[:, None, None] * np.eye(2)
# Correlated Gaussian draws (seed 1) added to the model proper motions; the mock is fitted like the data.
noise = np.einsum('nij,nj->ni', np.linalg.cholesky(cov), np.random.default_rng(1).normal(size=(len(D), 2)))
rows += fits('mock', 'RR Lyrae', c | dict(pmra=mra + noise[:, 0], pmdec=mde + noise[:, 1]))
t = pd.DataFrame(rows)
t.to_csv(ROOT / 'results/kin_dispersion_cut.csv', index=False, float_format='%.4f')
````

## Summary of the kinematics

1. Rotational support rises monotonically with metallicity. At 4 kpc, $v_{\rm rot}/\sigma_R$ rises from
   1.48 below [Fe/H] = -1.5 to 3.72 above -0.5 and
   $v_{\rm rot}/\sigma_z$ from 1.76 to 4.91 (APOGEE alone: 4.05 and
   7.71); in total the rotation rises by 21 km/s and $\sigma_R$ and $\sigma_z$ fall by
   18 and 17 km/s, most of it above [Fe/H] = -1.1.
2. The velocity ellipsoid has the same ordering of axes in every giant bin, $\sigma_R>\sigma_\phi\approx\sigma_z$ with
   $\sigma_\phi/\sigma_R$ and $\sigma_z/\sigma_R$ of 0.76–0.84, and all dispersions decline
   outwards by 4.5–8.5 per cent per kpc. The RR Lyrae have the radial dispersion, rotation
   and radial support of the giants below [Fe/H] = -1.1 ($\sigma_R$ = 38.1 (+2.9/−3.1) km/s,
   $v_{\rm rot}/\sigma_R$ = 1.66 (+0.17/−0.14)).
3. The ordering holds in both surveys and in every disc model we tried except the worst-fitting one, which swaps only the two most
   metal-poor bins; the absolute rotation speed depends on the model.
4. Fitted on their own, the giant bins put the rotation centre 0.3–0.5° east of the
   vdMK14 centre and 0.1–0.5° from the density centres (joint fit:
   0.47° east, 0.25–0.42° from them), and all groups leave an $m=2$
   residual of 5.3–8.5 km/s in $v_t$ with a common phase about a symmetric disc.
5. Fitted on their own, all giant bins share the line of nodes ($\theta$ = 143–146°),
   but the two bins below [Fe/H] = -1.1 are less inclined ($i$ = 24.6 and 25.2°)
   than those above (27.4–28.3°; Figure 7.3); the shape of the velocity
   ellipsoid is the same whether the geometry is fitted per group or shared.
