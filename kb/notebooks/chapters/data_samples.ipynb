# Data: the samples

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/vasilybelokurov/lmc-ancient-disc/blob/main/kb/notebooks/chapters/data_samples.ipynb)

Every number on this page is written by `scripts/paper_text_data.py` from the result tables and catalogues
(`results/macros_data.tex`); the code cells are the figure script `scripts/paper_maps.py`, step by step, and the scripts that
build the catalogues, shown but not run.

We use three tracers of the old population of the Large Magellanic Cloud (LMC). The first are RR Lyrae stars of type ab (RRab:
RR Lyrae pulsating in the fundamental radial mode, with asymmetric, saw-tooth light curves), old pulsating horizontal-branch stars
whose period and light-curve shape give a distance and a photometric metallicity. They come from the
fourth phase of the Optical Gravitational Lensing Experiment (OGLE-IV; [Udalski et al. 2015](https://doi.org/10.48550/arXiv.1504.05966))
and from the third data release of the *Gaia* mission (*Gaia* DR3; [Gaia Collaboration 2016](https://doi.org/10.1051/0004-6361/201629272),
[2023](https://doi.org/10.1051/0004-6361/202243940)). The other two are red giants with spectra from the 17th data release of the
Apache Point Observatory Galactic Evolution Experiment (APOGEE DR17; [Majewski et al. 2017](https://doi.org/10.3847/1538-3881/aa784d),
[Abdurro'uf et al. 2022](https://doi.org/10.3847/1538-4365/ac4414)) and from the 20th data release of the fifth Sloan Digital Sky
Survey, observed with the Baryon Oscillation Spectroscopic Survey spectrographs (SDSS-V DR20 BOSS;
[Kollmeier et al. 2026](https://doi.org/10.3847/1538-3881/ae0576), [SDSS Collaboration 2026](https://doi.org/10.48550/arXiv.2607.26149),
[Smee et al. 2013](https://doi.org/10.1088/0004-6256/146/2/32)). This chapter defines the samples; the chapter on selection
functions derives the probability that each star entered its sample, and the chapter on stellar mass converts the
selection-corrected giant counts into the stellar mass formed at each metallicity.

Metallicity here is the iron abundance [Fe/H], the logarithm of the ratio of iron to hydrogen atoms relative to the Sun's, in dex.
All positions refer to the LMC centre $(\alpha,\delta)$ = (79.88°, -69.59°) in right ascension
and declination of [van der Marel & Kallivayalil (2014)](https://doi.org/10.1088/0004-637X/781/2/121), used here only as the
origin of the angular distance $\rho$ of a star from the centre; the kinematics chapter fits the kinematic centre.

## RR Lyrae

We unite the OGLE-IV collection of LMC RRab ([Udalski et al. 2015](https://doi.org/10.48550/arXiv.1504.05966);
[Soszyński et al. 2016](https://doi.org/10.48550/arXiv.1606.02727), [2019](https://doi.org/10.32023/0001-5237/69.2.1)) with the
RRab of the *Gaia* DR3 variability catalogue ([Clementini et al. 2023](https://doi.org/10.1051/0004-6361/202243964)) within
20° of the centre, matching the two at 1″ (arcsec). The union holds 33 898
RRab, of which 24 481 are in both catalogues. Periods and light-curve parameters come from OGLE where a star has them
and from *Gaia* otherwise. OGLE-LMC-RRLYR-39960 and later identifiers were added to the OGLE collection from
*Gaia* ([Soszyński et al. 2019](https://doi.org/10.32023/0001-5237/69.2.1)); only the 28 190 stars of OGLE's own
search are independent of *Gaia*, which matters for the completeness estimate of the selection-function chapter.

**Metallicity.** Photometric [Fe/H] follows from the period $P$ and the Fourier phase $\phi_{31}=\phi_3-3\phi_1$ of the light
curve (the phase of its third harmonic relative to three times that of the first, in a Fourier series of the magnitude in time).
For stars with OGLE $I$-band light curves we transform $\phi_{31}$ to the $V$ band and apply the relation of
[Mullen et al. (2021)](https://doi.org/10.3847/1538-4357/abefd4), as in
[Martínez-Vázquez et al. (2026, their eq. A2)](https://doi.org/10.48550/arXiv.2609.27119); this places [Fe/H] on the scale of
[Crestani et al. (2021)](https://doi.org/10.3847/1538-4357/abd183). For the remaining stars we use the *Gaia* $G$-band relation of
[Muraveva et al. (2025)](https://doi.org/10.1093/mnras/stae2679), shifted by +0.22 dex, the median difference
on the 17 386 stars with both; the difference does not depend on [Fe/H] (slope of a Deming regression, which allows
for errors in both estimates, 0.978 ± 0.012). Against the high-resolution spectroscopy of
[Gratton et al. (2004)](https://doi.org/10.1051/0004-6361:20035840), transformed to the same scale
([Zinn & West 1984](https://doi.org/10.1086/190947); [Carretta et al. 2009](https://doi.org/10.1051/0004-6361/200913003)), the
adopted values are offset by +0.07 ± 0.05 dex (48 stars; see
the chapter on RR Lyrae and BOSS calibration). Of the members defined below, 25 326 have $I$-band and
449 $G$-band metallicities; 1 106 have neither.

**Distance.** We measure the distance modulus of each star relative to the LMC centre, $\Delta\mu$ (mag; $\Delta\mu=0$ at the mean
LMC distance), from a period–Wesenheit–metallicity relation fitted to the stars at $\rho$ < 3°. A Wesenheit
index is a magnitude minus a multiple of a colour, chosen so that interstellar extinction cancels. We use the OGLE index
$W_I=I-$ 1.55 $(V-I)$ where available (scatter about the relation 0.10 mag,
13 923 stars) and the *Gaia* index $W_G=G-$ 1.90 $(G_{\rm BP}-G_{\rm RP})$
([Ripepi et al. 2019](https://doi.org/10.1051/0004-6361/201834506)) otherwise (0.19 mag); $G$, $G_{\rm BP}$ and
$G_{\rm RP}$ are the *Gaia* broad, blue and red magnitudes. Only relative distances enter this analysis.

**Members.** An RRab is a member if it lies at $\rho$ < 10°, more than 13° from the Small
Magellanic Cloud (SMC), and within $|\Delta\mu|$ < 0.6 mag; the window is 3 times the
0.2-mag scatter of $\Delta\mu$ for the *Gaia* index and leaves room for the inclined disc. The SMC radius is
where the fraction of stars with SMC-like proper motions falls to the level set by LMC stars (see the chapter on RR Lyrae and BOSS
calibration). The sample holds 26 881 members, 23 356 with *Gaia* proper motions. No
proper-motion cut is applied, so the kinematics remain unbiased. Stars in the $\Delta\mu$ sidebands 0.9 <
$|\Delta\mu|$ < 2 mag measure the foreground left in the window: 1.9 per cent at $\rho$ <
2°, 0.8–1.3 per cent at
2°–8° and 2.3 per cent at
8°–10°.

### Building the RR Lyrae catalogue

`scripts/build_rrab.py` makes the catalogue `data/rrab.parquet` that every RR Lyrae figure of this site reads. It matches the
OGLE-IV collection to the *Gaia* DR3 RR Lyrae, forms their union, computes the two photometric metallicities and the adopted one,
fits the period–Wesenheit–metallicity relations on the stars at $\rho$ < 3° to obtain $\Delta\mu$, and flags
the members.

```python
import pandas as pd
from astropy.coordinates import SkyCoord

from lmcold.constants import LMC
from lmcold.footprint import DMU_WINDOW, RHO_MAX, smc_masked
from lmcold.rrlyrae import feh_gaia_G_m25, feh_gaia_G_p26, feh_ogle_I, ogle_own_discovery, read_ogle_rrab, relative_modulus
from lmcold.sources import DATA, path
```

#### Load the data
The OGLE-IV LMC RRab collection (identifications and light-curve parameters) and the Gaia DR3 RR Lyrae
within 20 deg of the LMC (`scripts/fetch_ogle4_rrab.py`, `scripts/fetch_gaia_rrlyrae.py`).

```python
ogle = read_ogle_rrab(path("ogle4_rrab").parent)
gaia = pd.read_parquet(path("gaia_rrlyrae_lmc20"))
```

#### Cross-match OGLE with Gaia
Each OGLE RRab takes the source_id of the nearest Gaia star within 1 arcsec.

```python
# Nearest Gaia star to each OGLE RRab; a match within 1 arcsec gives the OGLE star that Gaia source_id.
co, cg = SkyCoord(ogle.ra, ogle.dec, unit="deg"), SkyCoord(gaia.ra, gaia.dec, unit="deg")
j, sep, _ = co.match_to_catalog_sky(cg)
hit = sep.arcsec < 1.0
ogle["source_id"] = np.where(hit, gaia.source_id.to_numpy()[j], -1)
assert not ogle.source_id[hit].duplicated().any()  # no Gaia star matched to two OGLE stars
```

#### The union of the two catalogues
Stars in both, in OGLE only, and in Gaia only when Gaia classifies them RRab;
`ogle_own` marks OGLE's own discoveries.

```python
g = gaia.rename(columns={"ra": "ra_gaia", "dec": "dec_gaia"})
# Union: both (matched), left_only (OGLE only), right_only (Gaia only; kept only if Gaia's class is RRab).
t = ogle.merge(g, on="source_id", how="outer", indicator=True)
t = t[(t._merge == "both") | (t._merge == "left_only") | (t.best_classification == "RRab")]
t = t.assign(in_ogle=t._merge != "right_only", in_gaia=t._merge != "left_only").drop(columns="_merge")
# Identifier placeholder for Gaia-only rows (>= FIRST_GAIA_ADDED, so not an own discovery; in_ogle is False anyway).
t["ogle_own"] = t.in_ogle & ogle_own_discovery(t.ogle_id.fillna("OGLE-LMC-RRLYR-99999"))
t["ra"] = t.ra.fillna(t.ra_gaia)
t["dec"] = t.dec.fillna(t.dec_gaia)
t["P"] = t.P.fillna(t.pf)  # period [d]: OGLE, else Gaia fundamental period
```

#### Photometric metallicities
The $I$-band and $G$-band [Fe/H] from period and $\phi_{31}$, and the adopted value:
$I$ band where available, otherwise $G$ band shifted by the median difference.

```python
t["feh_I"] = feh_ogle_I(t.P, t.phi31)
t["feh_G_m25"] = feh_gaia_G_m25(t.pf, t.phi31_g)  # Gaia period: phi31_g was fitted with it
t["feh_G_p26"] = feh_gaia_G_p26(t.pf, t.phi31_g)
t["feh_sos"] = t.metallicity  # Gaia DR3 SOS photometric metallicity (Clementini et al. 2023)
# Adopted [Fe/H]: OGLE I where available (its zero point agrees with Gratton+04 spectroscopy); otherwise Gaia G (Muraveva+25)
# shifted onto the I scale by the median difference on the stars with both (the difference does not depend on [Fe/H]:
# Deming slope 0.98 +- 0.01, scripts/fig_rrab_feh_transform.py).
shift = (t.feh_I - t.feh_G_m25).median()
t["feh"] = t.feh_I.fillna(t.feh_G_m25 + shift)
t["feh_src"] = np.where(t.feh_I.notna(), "I", np.where(t.feh_G_m25.notna(), "G", ""))
```

#### Distance from the LMC centre
Angular distance $\rho$ [deg] from the reference centre; the catalogue is cut at 20 deg.

```python
c0 = SkyCoord(LMC["ra0"], LMC["dec0"], unit="deg")
t["rho"] = SkyCoord(t.ra, t.dec, unit="deg").separation(c0).deg
t = t[t.rho < 20].reset_index(drop=True)  # same radius as the Gaia query; drops OGLE stars beyond it
```

#### Relative distance modulus
Period-Wesenheit-[Fe/H] relations fitted on the core ($\rho$ < 3 deg) give
$\Delta\mu$ [mag] for each star, from the OGLE index where available and the Gaia index otherwise.

```python
# Relative distance modulus, dmu = 0 at the mean LMC distance (only relative distances are needed: the SMC is ~0.45 mag behind).
# OGLE Wesenheit W_I = I - 1.55 (V - I) (Udalski et al. 2008) where available (scatter in the core 0.10 mag), otherwise the Gaia
# Wesenheit W_G = G - 1.90 (BP - RP) (intensity averages; Ripepi et al. 2019; 0.19 mag, crowded BP/RP); each period-Wesenheit-
# [Fe/H] relation is fitted on the LMC itself (rho < 3 deg).
# Stars without [Fe/H] take the median.
feh = t.feh.fillna(t.feh.median()).to_numpy()
core = ((t.rho < 3) & t.feh.notna()).to_numpy()  # fit only on stars with measured [Fe/H]
fits = []
for name, W in (("G", t.int_average_g - 1.90 * (t.int_average_bp - t.int_average_rp)), ("I", t.I - 1.55 * (t.V - t.I))):
    # Fit W = a + b log10 P + c [Fe/H] on the core with 3-sigma clipping; dmu = W - fit (lmcold.rrlyrae.relative_modulus).
    dmu, coef, fit = relative_modulus(W.to_numpy(), t.P.to_numpy(), feh, core)
    t["dmu_" + name] = dmu
    fits.append(dict(wesenheit=name, a=coef[0], b_logP=coef[1], c_feh=coef[2], N=fit.sum(),
                     rsd=1.4826 * np.median(np.abs(dmu[fit] - np.median(dmu[fit])))))
pd.DataFrame(fits).to_csv(DATA.parent / "results" / "rrab_pw_fit.csv", index=False, float_format="%.4f")
t["dmu"] = t.dmu_I.fillna(t.dmu_G)  # OGLE I Wesenheit preferred (smaller scatter)
t["dmu_src"] = np.where(t.dmu_I.notna(), "I", np.where(t.dmu_G.notna(), "G", ""))
```

#### Members
Inside the science radius, outside the SMC mask and within the $\Delta\mu$ window; the catalogue is written
to `data/rrab.parquet`.

```python
# Members: inside the science radius, outside the SMC mask and within the distance window (no proper-motion cut, so the
# kinematics stay unbiased). The 20 deg sample is kept for the mask and the foreground estimate.
t["smc_masked"] = smc_masked(t.ra.to_numpy(), t.dec.to_numpy())
t["member"] = (t.rho < RHO_MAX) & ~t.smc_masked & (t.dmu.abs() < DMU_WINDOW)
t.to_parquet(DATA / "rrab.parquet", index=False)
print(f"{len(t)} RRab: OGLE and Gaia {(t.in_ogle & t.in_gaia).sum()}, OGLE only {(t.in_ogle & ~t.in_gaia).sum()}, "
      f"Gaia only {(~t.in_ogle).sum()}; G shift {shift:+.4f}; adopted [Fe/H] from {t.feh_src.value_counts().to_dict()}; "
      f"dmu from {t.dmu_src.value_counts().to_dict()}; SMC-masked {t.smc_masked.sum()}; members {t.member.sum()}")
```

## Red giants

The giants come from APOGEE DR17 ([Majewski et al. 2017](https://doi.org/10.3847/1538-3881/aa784d);
[Wilson et al. 2019](https://doi.org/10.1088/1538-3873/ab0075); [Abdurro'uf et al. 2022](https://doi.org/10.3847/1538-4365/ac4414)),
a high-resolution near-infrared survey whose stellar parameters come from the APOGEE Stellar Parameters and Chemical Abundances
Pipeline (ASPCAP; [García Pérez et al. 2016](https://doi.org/10.3847/0004-6256/151/6/144)), and from SDSS-V DR20
([Kollmeier et al. 2026](https://doi.org/10.3847/1538-3881/ae0576); [SDSS Collaboration 2026](https://doi.org/10.48550/arXiv.2607.26149)),
with low-resolution optical BOSS spectra ([Smee et al. 2013](https://doi.org/10.1088/0004-6256/146/2/32)) analysed by BOSSNet, a
neural network that estimates stellar parameters from the spectra ([Sizemore et al. 2024](https://doi.org/10.3847/1538-3881/ad291d)).
In the LMC, APOGEE observed red giants in 35 fields (LMC1–35;
[Nidever et al. 2020](https://doi.org/10.3847/1538-4357/ab7305)), and SDSS-V observed red giants to $G$ = 17.5
over the whole galaxy in the Magellanic Genesis survey ([Nidever et al. 2026](https://doi.org/10.3847/1538-3881/ae7c76)). We keep
stars within 12° of the centre with heliocentric line-of-sight velocity 100 < $v_{\rm los}$ <
450 km/s. A good APOGEE spectrum is not a duplicate observation, has the ASPCAP flag STAR_BAD unset and a
finite [Fe/H], with signal-to-noise ratio S/N ≥ 50 for the kinematic samples and S/N ≥
30 for the selection function and the mass (selection-function chapter); a good BOSS spectrum has the
BOSSNet quality flag unset, a finite [Fe/H] and velocity, S/N ≥ 16 and a velocity error below
10 km/s. Below S/N = 16 BOSSNet [Fe/H] follows APOGEE poorly on the stars observed by
both: the slope of BOSSNet against APOGEE [Fe/H] is 0.16 ± 0.23 at S/N
10–13 and 0.51 ± 0.10 at
13–16, against 0.79 ± 0.02 above
40, and the calibrated values run 0.06–0.16 dex high (see the chapter on
RR Lyrae and BOSS calibration). Both samples keep giants only (effective temperature $T_{\rm eff}$ < 5500 K and
surface gravity $\log g$ < 3.5, with $g$ in cm s⁻²). For kinematics, APOGEE stars also need velocity errors
below 1 km/s and a visit-to-visit scatter below 1.5 km/s; all stars need *Gaia* DR3 proper
motions with renormalised unit weight error (RUWE, a measure of how well a single-star astrometric model fits) ≤
1.4 and a parallax $\varpi$ (with error $\sigma_\varpi$, both in mas) consistent with the LMC,
$(\varpi+$ 0.017$)/\sigma_\varpi$ < 3, where 0.017 mas corrects the
*Gaia* parallax zero point ([Lindegren et al. 2021](https://doi.org/10.1051/0004-6361/202039653)).

We place BOSS on the APOGEE scale with the 2 097 stars observed by both surveys. With
$t=(T_{\rm eff,BOSS}-$ 4200 K$)/$ 1000 K and $s=$ 20 $/({\rm S/N})_{\rm BOSS}-1$,
robust fits with Huber weights (which reduce the weight of outlying residuals) give

$$
v_{\rm BOSS}-v_{\rm APOGEE}=d_0+d_1\,t,\qquad
[{\rm Fe/H}]_{\rm BOSS}=c_0+c_1\,[{\rm Fe/H}]_{\rm APOGEE}+c_2\,t+c_3\,s,
$$

with $d_0$ = 3.98 km/s, $d_1$ = −4.75 km/s, $c_0$ = -0.069 dex, $c_1$ =
0.733, $c_2$ = −0.043 dex and $c_3$ = 0.053 dex. The [Fe/H] relation is fitted with
the more precise APOGEE value as the independent variable and inverted for BOSS-only stars; this undoes the compression of the
BOSSNet scale without the attenuation of a fit in the other direction. Fitted on half of the common stars and tested on the other
half, the calibrated BOSS values differ from APOGEE by a median of -0.004 dex with a robust scatter (the median absolute
deviation scaled to the standard deviation of a Gaussian) of 0.094 dex, and the velocities by +0.05 km/s; the median residual is
within 0.04 dex in every bin of S/N and of $T_{\rm eff}$. The calibration still leaves a bias of
−0.15 to −0.17 dex at -2 < [Fe/H] < -1.5, which the stellar-mass chapter measures and corrects.
The chapter on RR Lyrae and BOSS calibration shows the fits on the common stars. The kinematic samples hold 8 452
APOGEE and 57 128 BOSS giants (1 926 in both).

#### Reading one summary file
`read` takes the chosen columns of one FITS file and keeps the rows within 12 deg of the LMC centre with
heliocentric velocity between 100 and 450 km/s.

```python
def read(name, cols, hdu):
    """Read the selected columns of one SDSS summary file and apply the loose position and velocity window.

    Parameters
    ----------
    name : str
        sources.yaml entry of the FITS file.
    cols : dict
        File column name -> common column name.
    hdu : int
        FITS extension holding the table.

    Returns
    -------
    pandas.DataFrame
        Rows within 12 deg of the LMC centre (constants.LMC) with 100 < v < 450 km/s, columns renamed by `cols`.
    """
    with fits.open(path(name), memmap=True) as f:
        d = f[hdu].data
        # Columns are converted to native byte order (FITS is big-endian), which pandas needs.
        t = pd.DataFrame({new: np.asarray(d[old]).astype(np.asarray(d[old]).dtype.newbyteorder("=")) for old, new in cols.items()})
    # 12 deg: the radius of the giant samples; 100-450 km/s: a loose heliocentric window around the LMC velocity
    # (rows with NaN v fail it).
    keep = (angular_distance(t.ra, t.dec, LMC["ra0"], LMC["dec0"]) < 12) & (t.v > 100) & (t.v < 450)
    return t[keep.to_numpy()]
```

#### The query
The rough cuts of the carton, run in the Gaia archive: a superset of the stars the
carton selects.

```python
# WHERE (the carton's rough cuts, a superset of the eligible stars):
#   - within 12 deg of the adopted LMC centre (constants.LMC; the radius of the giant samples);
#   - G <= 17.5 (the carton's final magnitude limit) and observed BP - RP >= 1 (red giants);
#   - -3 <= pmra, pmdec <= 3 mas/yr (the carton's PM box);
#   - parallax not significant: parallax <= 0, or parallax + 0.025 mas (the carton's zero-point offset) <= 5 parallax_error.
#     Rows with NULL parallax or proper motion fail these tests and are not returned.
QUERY = f"""
SELECT source_id, ra, dec, pmra, pmdec, pmra_error, pmdec_error, pmra_pmdec_corr, parallax, parallax_error, ruwe,
       phot_g_mean_mag AS g, phot_bp_mean_mag AS bp, phot_rp_mean_mag AS rp
FROM gaiadr3.gaia_source
WHERE 1 = CONTAINS(POINT('ICRS', ra, dec), CIRCLE('ICRS', {LMC['ra0']}, {LMC['dec0']}, 12))
  AND phot_g_mean_mag <= 17.5 AND phot_bp_mean_mag - phot_rp_mean_mag >= 1
  AND pmra BETWEEN -3 AND 3 AND pmdec BETWEEN -3 AND 3
  AND (parallax <= 0 OR parallax + 0.025 <= 5 * parallax_error)
"""
```

#### Load the data
The Gaia DR3 stars passing the rough cuts, from the Gaia@AIP TAP service, sorted by source_id.

```python
job.run().wait(phases=["COMPLETED", "ERROR", "ABORTED"])
job.raise_if_error()
```

```python
def native(a):
    """Return a FITS column as a numpy array in native byte order (FITS data are big-endian; pandas needs native order).

    Parameters
    ----------
    a : array_like
        Column of a FITS table.

    Returns
    -------
    numpy.ndarray
```

#### Helper
`read` returns the rows of a targetdb table whose key is in a given list, filtered while scanning.

```python
def read(name, cols, key, values):
    """Rows of a targetdb table whose `key` is in `values`, filtered while scanning.

    Parameters
    ----------
    name : str
        Table name (file ~/data/sdss/dr20/targetdb/<name>.parquet).
    cols : list of str
        Columns to read.
    key : str
        Column the filter is applied to.
    values : array_like

    -------
    pandas.DataFrame
```

#### Helpers
`native` turns FITS columns into a table; `field_stars` returns the polygon candidates of one field with
their Gaia mates.

```python
def native(d, cols):
    """FITS table columns as a DataFrame: strings stripped, numbers in native byte order.

    Parameters
    ----------
    d : FITS_rec
        Table data.
    cols : dict
        FITS column -> output column.

    Returns
    -------
    pandas.DataFrame
    """
    out = {}
    # Byte strings -> stripped str; big-endian FITS numbers -> native byte order (pandas needs native).
    for old, new in cols.items():
        a = np.asarray(d[old])
        out[new] = np.char.strip(a.astype(str)) if a.dtype.kind in "SU" else a.astype(a.dtype.newbyteorder("="))
    return pd.DataFrame(out)


def field_stars(name, ra, dec):
    """RGB-polygon candidates of one APOGEE-2 LMC field with their Gaia DR3 mate.

    Parameters
    ----------
    name : str
        Field name, 'LMC1' ... 'LMC35'.
    ra, dec : float
        Field centre [deg] (apogee2Field).

    Returns
    -------
    pandas.DataFrame
        Objects of apogee2Object_<name> inside the BrtRGB or FntRGB polygon and within lmcold.apogee_target.RADIUS (0.95 deg)
        of the centre, with the Gaia columns of GAIA (left join: objects without a mate are kept).
    """
    # Targeting file of the field: 2MASS id, position [deg] and J, H, K [mag].
    o = native(fits.getdata(path(f"apogee2_object_{name.lower()}")), {c: c.lower() for c in ("APOGEE_ID", "RA", "DEC", "J", "H", "K")})
    o[["j", "h", "k"]] = o[["j", "h", "k"]].astype(float).round(3)  # 2MASS magnitudes are quoted to 1e-3; float32 breaks edge ties
    o["nbr"] = at.bright_neighbour(o.ra.to_numpy(), o.dec.to_numpy(), o.h.to_numpy())  # on the full object file
    # RGB class in (J - K, H): 0 outside, 1 BrtRGB, 2 FntRGB polygon (H limit of this field); distance from centre [deg].
    o["cls"] = at.rgb_class((o.j - o.k).to_numpy(), o.h.to_numpy(), at.MAGLIM[name])
    o["rfield"] = angular_distance(o.ra, o.dec, ra, dec)
    # Keep polygon members within the field radius (lmcold.apogee_target.RADIUS, 0.95 deg).
    o = o[(o.cls > 0) & (o.rfield <= at.RADIUS)].assign(field=name, maglim=at.MAGLIM[name])
    # Asynchronous TAP job at Gaia@AIP for the cone of the field (wait up to 900 s); the result table as a DataFrame.
    job = pyvo.dal.TAPService("https://gaia.aip.de/tap").submit_job(GAIA.format(ra=ra, dec=dec, r=at.RADIUS + 0.01), maxrec=2_000_000)
    job.run().wait(phases=["COMPLETED", "ERROR", "ABORTED"], timeout=900)
    job.raise_if_error()
    g = job.fetch_result().to_table().to_pandas()
    # Attach the Gaia mate by 2MASS designation (nearest where several Gaia sources share one 2MASS source).
    return at.join_gaia(o, g)
```

### XP metallicities

`scripts/build_xp_andrae.py` takes [M/H]$_{\rm XP}$ of [Andrae et al. (2023)](https://doi.org/10.3847/1538-4365/acd53e) for the
stars of both parents and keeps it only for cool giants, where it is reliable. These values define the pass groups of both
selection functions and the tests of the next chapter.
